```python
"""Bucholtz Date-Driven Trade Idea Engine

Research implementation inspired by the repeatable timing and confirmation concepts
in Malcolm G. Bucholtz's Financial Astrology Almanac series supplied by the user.

This tool is for research/education. Astrology-based market timing is not established
as a scientifically validated forecasting method. The engine therefore treats cosmic
configurations as timing windows and requires market-price confirmation before producing
trade ideas.
"""
from __future__ import annotations

import math
import json
import warnings
import re
from dataclasses import dataclass, asdict
from datetime import datetime, timedelta, timezone
from pathlib import Path
from zoneinfo import ZoneInfo
from typing import Dict, List, Tuple, Optional

import numpy as np
import pandas as pd

try:
    import yfinance as yf
except Exception:
    yf = None

try:
    import swisseph as swe
except Exception:
    swe = None

# -------------------------
# Source-derived rulebook
# -------------------------
RULEBOOK = [
    {
        "id":"TREND_MA",
        "family":"Technical confirmation",
        "rule":"Use the 200-day moving average as a broader trend measure and the 50-day moving average as a shorter-term trend measure; price crossing above/below these averages can indicate trend change.",
        "source":"Financial Astrology Almanac 2024 – Trend Changes",
        "automatable":True,
    },
    {
        "id":"STOCH",
        "family":"Technical confirmation",
        "rule":"Slow Stochastic: bullish crossover from below 20 is a potential buy confirmation; move above 80 followed by a cross back beneath the upper region is a sell warning. Prefer price confirmation rather than oscillator alone.",
        "source":"Financial Astrology Almanac 2024 – Trend Changes",
        "automatable":True,
    },
    {
        "id":"TSI",
        "family":"Technical confirmation",
        "rule":"True Strength Index / oscillator crossovers may be used to confirm counter-trend or trend-change opportunities.",
        "source":"Financial Astrology Almanac 2024 – Trend Changes",
        "automatable":True,
    },
    {
        "id":"NATAL_HARD",
        "family":"Natal/first-trade transits",
        "rule":"Watch 0°, 90° and 180° transits from Sun, Mars, Jupiter, Saturn, Uranus and Neptune to key first-trade/natal points (Sun, Moon, Jupiter, Ascendant, Mid-Heaven). These are timing alerts for trend change, volatility or breakout; direction requires technical confirmation.",
        "source":"Financial Astrology Almanac 2024 – McWhirter methodology",
        "automatable":True,
    },
    {
        "id":"MOON_NATAL",
        "family":"Lunar/natal transits",
        "rule":"Moon passages over important natal/first-trade points can define short timing windows. Bucholtz combines these with other planetary events and technical evidence.",
        "source":"Financial Astrology Almanac 2024 – Lunar Cycles / commodity chapters",
        "automatable":True,
    },
    {
        "id":"RETROGRADE",
        "family":"Planetary cycles",
        "rule":"Mercury, Venus and Mars retrograde periods are recurring timing windows in the almanacs. They are not direction by themselves; use technical trend-change confirmation.",
        "source":"Financial Astrology Almanac series 2017–2026",
        "automatable":True,
    },
    {
        "id":"DECLINATION",
        "family":"Declination",
        "rule":"Zero-declination crossings, declination extrema, and returns to natal declination levels are treated as potential timing events, particularly when they cluster with other signals.",
        "source":"Financial Astrology Almanac series – declination sections",
        "automatable":True,
    },
    {
        "id":"ELONGATION",
        "family":"Elongation",
        "rule":"Mercury/Venus greatest eastern or western elongation can coincide with short-term turning windows and should be examined together with price action.",
        "source":"Financial Astrology Almanac series – elongation sections",
        "automatable":True,
    },
    {
        "id":"LUNATION",
        "family":"Lunar cycles",
        "rule":"New-Moon cycles and hard contacts to important market natal points are used as timing frameworks for equity-market volatility and trend changes.",
        "source":"Financial Astrology Almanac 2024/2025 – Lunation and NYSE sections",
        "automatable":True,
    },
    {
        "id":"ASPECTS",
        "family":"Planetary aspects",
        "rule":"Planetary conjunctions and major 60°, 90°, 120° and 180° aspects are examined for timing confluence; hard aspects receive greater timing weight in this implementation.",
        "source":"Financial Astrology Almanac series – aspects/cosmic events sections",
        "automatable":True,
    },
    {
        "id":"PERI_APH",
        "family":"Orbital distance",
        "rule":"Mercury/Venus perihelion and aphelion are treated as cosmic-event timing windows, especially when they coincide with other events.",
        "source":"Financial Astrology Almanac 2024/2025 – shorter cycles",
        "automatable":True,
    },
    {
        "id":"PARALLEL",
        "family":"Declination",
        "rule":"Planetary declination parallels and contra-parallels within about 1.5 degrees are recurring commodity timing relationships in Bucholtz's examples.",
        "source":"Financial Astrology Almanac 2018 and later commodity chapters",
        "automatable":True,
    },
    {
        "id":"QUANTUM",
        "family":"Quantum Lines",
        "rule":"Quantum Line=(N×360)+PSO where PSO=heliocentric planetary longitude×Conversion Scale; nearby levels are treated as secondary support/resistance confluence.",
        "source":"Financial Astrology Almanac 2021/2024/2025 – Quantum Science",
        "automatable":True,
    },
]

PLANETS = {
    "Sun": 0,
    "Moon": 1,
    "Mercury": 2,
    "Venus": 3,
    "Mars": 4,
    "Jupiter": 5,
    "Saturn": 6,
    "Uranus": 7,
    "Neptune": 8,
    "Pluto": 9,
}

# Commodity futures used by Yahoo Finance. First-trade dates are used to construct
# research natal charts. Some markets have multiple valid historical reference dates;
# the app labels these as modelling anchors, not immutable facts.
COMMODITIES = {
    "Gold": {"ticker":"GC=F", "first_trade":"1974-12-31", "category":"Metals"},
    "Silver": {"ticker":"SI=F", "first_trade":"1933-07-05", "category":"Metals"},
    "Copper": {"ticker":"HG=F", "first_trade":"1958-07-29", "category":"Metals"},
    "Platinum": {"ticker":"PL=F", "first_trade":"1956-12-03", "category":"Metals"},
    "Palladium": {"ticker":"PA=F", "first_trade":"1968-01-22", "category":"Metals"},
    "WTI Crude Oil": {"ticker":"CL=F", "first_trade":"1983-03-30", "category":"Energy"},
    "Natural Gas": {"ticker":"NG=F", "first_trade":"1990-04-03", "category":"Energy"},
    "Corn": {"ticker":"ZC=F", "first_trade":"1877-01-02", "category":"Grains"},
    "Wheat": {"ticker":"ZW=F", "first_trade":"1877-01-02", "category":"Grains"},
    "Oats": {"ticker":"ZO=F", "first_trade":"1877-01-02", "category":"Grains"},
    "Soybeans": {"ticker":"ZS=F", "first_trade":"1936-10-05", "category":"Grains"},
    "Live Cattle": {"ticker":"LE=F", "first_trade":"1964-11-30", "category":"Meats"},
    "Feeder Cattle": {"ticker":"GF=F", "first_trade":"1971-11-30", "category":"Meats"},
    "Lean Hogs": {"ticker":"HE=F", "first_trade":"1966-02-28", "category":"Meats"},
    "Coffee": {"ticker":"KC=F", "first_trade":"1882-03-07", "category":"Softs"},
    "Sugar": {"ticker":"SB=F", "first_trade":"1914-01-01", "category":"Softs"},
    "Cocoa": {"ticker":"CC=F", "first_trade":"1925-10-01", "category":"Softs"},
    "Cotton": {"ticker":"CT=F", "first_trade":"1870-04-10", "category":"Softs"},
    "30Y Treasury Bond": {"ticker":"ZB=F", "first_trade":"1977-08-22", "category":"Rates"},
    "10Y Treasury Note": {"ticker":"ZN=F", "first_trade":"1982-05-03", "category":"Rates"},
    "British Pound": {"ticker":"6B=F", "first_trade":"1972-05-16", "category":"FX"},
    "Euro FX": {"ticker":"6E=F", "first_trade":"1999-01-04", "category":"FX"},
    "Australian Dollar": {"ticker":"6A=F", "first_trade":"1987-01-13", "category":"FX"},
    "Canadian Dollar": {"ticker":"6C=F", "first_trade":"1972-05-16", "category":"FX"},
}

DEFAULT_STOCKS = [
    "AAPL","MSFT","NVDA","AMZN","GOOGL","META","TSLA","AVGO","BRK-B","JPM",
    "V","MA","XOM","LLY","WMT","COST","NFLX","AMD","PLTR","ORCL","CRM","INTC",
    "BAC","GS","UNH","JNJ","PG","HD","CAT","GE","BA","DIS","KO","PEP","MCD",
    "CVX","COP","SLB","NKE","SBUX","UBER","SHOP","PYPL","COIN","MSTR","MU","QCOM",
    "AMAT","LRCX","SMCI"
]

# -------------------------
# Astronomy helpers
# -------------------------
def _jd(dt: datetime) -> float:
    return swe.julday(dt.year, dt.month, dt.day, dt.hour + dt.minute/60 + dt.second/3600)

def _norm(x: float) -> float:
    return x % 360.0

def angular_distance(a: float, b: float) -> float:
    d = abs((_norm(a) - _norm(b) + 180) % 360 - 180)
    return d

def aspect_distance(a: float, b: float, aspect: float) -> float:
    return abs(angular_distance(a,b) - aspect)

def _planet(jd: float, pid: int, helio: bool=False):
    flags = swe.FLG_SWIEPH | swe.FLG_SPEED | swe.FLG_EQUATORIAL
    # longitude needs ecliptic, so get twice
    ecl_flags = swe.FLG_SWIEPH | swe.FLG_SPEED
    if helio:
        ecl_flags |= swe.FLG_HELCTR
    ecl, _ = swe.calc_ut(jd, pid, ecl_flags)
    equ, _ = swe.calc_ut(jd, pid, flags)
    return {
        "lon": _norm(ecl[0]),
        "lat": ecl[1],
        "dist": ecl[2],
        "speed_lon": ecl[3],
        "decl": equ[1],
    }

def astro_snapshot(date: pd.Timestamp) -> Dict[str,dict]:
    if swe is None:
        raise RuntimeError("pyswisseph is required. Install with: pip install pyswisseph")
    dt = datetime(date.year,date.month,date.day,12,0,0,tzinfo=timezone.utc)
    jd = _jd(dt)
    return {name:_planet(jd,pid) for name,pid in PLANETS.items()}

def _snapshot_offset(date: pd.Timestamp, days: int):
    return astro_snapshot(date + pd.Timedelta(days=days))

def detect_cosmic_events(date, orb=1.5) -> List[dict]:
    date = pd.Timestamp(date).normalize()
    snap = astro_snapshot(date)
    prev = _snapshot_offset(date,-1)
    nxt = _snapshot_offset(date,1)
    events=[]

    # Retrogrades
    for p in ["Mercury","Venus","Mars"]:
        if snap[p]["speed_lon"] < 0:
            events.append({"type":"retrograde","planet":p,"weight":8,
                           "reason":f"{p} is retrograde: Bucholtz treats retrograde periods as timing windows requiring chart confirmation."})

    # New / Full Moon
    phase = angular_distance(snap["Sun"]["lon"], snap["Moon"]["lon"])
    if phase <= 8:
        events.append({"type":"new_moon","planet":"Moon","weight":8,
                       "reason":f"New-Moon window (Sun–Moon separation {phase:.1f}°), used by Bucholtz/McWhirter as an equity timing framework."})
    if abs(phase-180) <= 8:
        events.append({"type":"full_moon","planet":"Moon","weight":5,
                       "reason":f"Full-Moon window (separation {phase:.1f}°), added as lunar timing confluence."})

    # Major planetary aspects
    names=["Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
    aspects=[(0,9,"conjunction"),(60,4,"sextile"),(90,8,"square"),(120,5,"trine"),(180,8,"opposition")]
    for i,a in enumerate(names):
        for b in names[i+1:]:
            for deg,w,label in aspects:
                err=aspect_distance(snap[a]["lon"],snap[b]["lon"],deg)
                if err <= orb:
                    events.append({"type":"aspect","planet":f"{a}-{b}","weight":w,
                                   "reason":f"{a}–{b} {label} near {deg}° (orb {err:.2f}°), a major aspect used as cosmic-event timing confluence."})

    # Declination zero crossings / local extrema
    for p in ["Mercury","Venus","Mars"]:
        d0,d1,d2=prev[p]["decl"],snap[p]["decl"],nxt[p]["decl"]
        if d0*d2 <= 0 or abs(d1) < 0.35:
            events.append({"type":"declination_zero","planet":p,"weight":7,
                           "reason":f"{p} is crossing/near 0° declination ({d1:.2f}°), a Bucholtz trend-change timing condition."})
        if (d1>d0 and d1>d2) or (d1<d0 and d1<d2):
            events.append({"type":"declination_extreme","planet":p,"weight":6,
                           "reason":f"{p} is at/near a local declination extreme ({d1:.2f}°), used as a timing alert in the almanacs."})

    # Mercury/Venus elongation local extrema
    for p in ["Mercury","Venus"]:
        e0=angular_distance(prev[p]["lon"],prev["Sun"]["lon"])
        e1=angular_distance(snap[p]["lon"],snap["Sun"]["lon"])
        e2=angular_distance(nxt[p]["lon"],nxt["Sun"]["lon"])
        if e1>e0 and e1>e2:
            events.append({"type":"elongation","planet":p,"weight":6,
                           "reason":f"{p} is near greatest elongation ({e1:.1f}° from Sun), a shorter-cycle timing event in Bucholtz."})

    # Perihelion/aphelion approximate from heliocentric radius local extrema
    for p in ["Mercury","Venus"]:
        pid=PLANETS[p]
        dt=datetime(date.year,date.month,date.day,12,tzinfo=timezone.utc)
        r0=_planet(_jd(dt-timedelta(days=1)),pid,True)["dist"]
        r1=_planet(_jd(dt),pid,True)["dist"]
        r2=_planet(_jd(dt+timedelta(days=1)),pid,True)["dist"]
        if r1<r0 and r1<r2:
            events.append({"type":"perihelion","planet":p,"weight":6,
                           "reason":f"{p} is near perihelion, treated by Bucholtz as a cosmic-event timing window."})
        if r1>r0 and r1>r2:
            events.append({"type":"aphelion","planet":p,"weight":6,
                           "reason":f"{p} is near aphelion, treated by Bucholtz as a cosmic-event timing window."})
    events.extend(detect_declination_parallels(date))
    return events


# -------------------------
# V4 verified first-trade database
# -------------------------
FIRST_TRADE_DB = None
FIRST_TRADE_DB_PATH = None

def load_first_trade_database(path=None):
    """Load first-trade/IPO data with an embedded clean-database safety net.

    V9.2.4 order:
      1) explicit file, if supplied and valid
      2) external CLEAN CSV, strict parsing
      3) EMBEDDED CLEAN DATABASE bundled inside this notebook
      4) external original CSV, strict parsing
      5) recovery parsing only as a last resort

    The embedded database makes Chapter IV independent of Colab's temporary file state.
    """
    global FIRST_TRADE_DB, FIRST_TRADE_DB_PATH

    explicit=[Path(path)] if path else []
    clean_candidates=[
        Path("/content/first_trades_IPOs_clean.csv"),
        Path("first_trades_IPOs_clean.csv"),
        Path("/mnt/data/first_trades_IPOs_clean.csv"),
    ]
    original_candidates=[
        Path("/content/first_trades_IPOs.csv"),
        Path("first_trades_IPOs.csv"),
        Path("/mnt/data/first_trades_IPOs.csv"),
    ]

    attempts=[]
    db=None
    found=None
    accepted_mode=None

    # A. Explicit path, then clean external files.
    strict_primary=explicit+clean_candidates
    seen=set()
    strict_primary=[p for p in strict_primary if not (str(p) in seen or seen.add(str(p)))]

    for candidate in strict_primary:
        if not candidate.exists():
            continue
        for enc in ("utf-8-sig","utf-8","cp1252","latin1"):
            try:
                trial=pd.read_csv(candidate,encoding=enc)
                if {"Ticker","Date"}.issubset(trial.columns) and len(trial)>0:
                    db=trial
                    found=candidate
                    accepted_mode=f"STRICT/{enc}"
                    break
            except Exception as e:
                attempts.append(f"{candidate} [{enc} strict]: {type(e).__name__}: {e}")
        if db is not None:
            break

    # B. Embedded clean database. This is preferred over recovering a malformed original.
    if db is None:
        try:
            trial=_embedded_first_trade_dataframe()
            if {"Ticker","Date"}.issubset(trial.columns) and len(trial)>0:
                db=trial
                found=Path("EMBEDDED:first_trades_IPOs_clean.csv")
                accepted_mode="EMBEDDED_CLEAN"
        except Exception as e:
            attempts.append(f"embedded clean database: {type(e).__name__}: {e}")

    # C. Strict original files.
    if db is None:
        for candidate in original_candidates:
            if not candidate.exists():
                continue
            for enc in ("utf-8-sig","utf-8","cp1252","latin1"):
                try:
                    trial=pd.read_csv(candidate,encoding=enc)
                    if {"Ticker","Date"}.issubset(trial.columns) and len(trial)>0:
                        db=trial
                        found=candidate
                        accepted_mode=f"STRICT_ORIGINAL/{enc}"
                        break
                except Exception as e:
                    attempts.append(f"{candidate} [{enc} strict]: {type(e).__name__}: {e}")
            if db is not None:
                break

    # D. Recovery only as absolute last resort.
    if db is None:
        recovery_candidates=clean_candidates+explicit+original_candidates
        seen=set()
        recovery_candidates=[p for p in recovery_candidates if not (str(p) in seen or seen.add(str(p)))]
        for candidate in recovery_candidates:
            if not candidate.exists():
                continue
            for enc in ("utf-8-sig","cp1252","latin1"):
                try:
                    trial=pd.read_csv(candidate,encoding=enc,engine="python",on_bad_lines="skip")
                    if {"Ticker","Date"}.issubset(trial.columns) and len(trial)>0:
                        db=trial
                        found=candidate
                        accepted_mode=f"RECOVERED/{enc}"
                        break
                except Exception as e:
                    attempts.append(f"{candidate} [{enc} recovery]: {type(e).__name__}: {e}")
            if db is not None:
                break

    if db is None:
        FIRST_TRADE_DB=None
        FIRST_TRADE_DB_PATH=None
        print("WARNING: no valid first-trade database could be loaded.")
        for msg in attempts[-8:]:
            print(" -",msg)
        return None

    db=db.loc[:,~db.columns.astype(str).str.startswith("Unnamed:")].copy()
    for col in ["name","Place","ipo_time","description"]:
        if col not in db.columns:
            db[col]=np.nan

    db["Ticker"]=db["Ticker"].astype(str).str.upper().str.strip().str.replace(".","-",regex=False)
    db=db[(db["Ticker"]!="") & (db["Ticker"]!="NAN")].copy()
    db["_parsed_date"]=pd.to_datetime(db["Date"],errors="coerce",dayfirst=False)
    db=db[db["_parsed_date"].notna()].copy()

    if db.empty:
        FIRST_TRADE_DB=None
        FIRST_TRADE_DB_PATH=None
        print("WARNING: selected first-trade database contains no usable dated records.")
        return None

    FIRST_TRADE_DB=db
    FIRST_TRADE_DB_PATH=str(found)
    print(f"Loaded {len(db):,} first-trade records from {FIRST_TRADE_DB_PATH} using {accepted_mode} "
          f"({db['Ticker'].nunique():,} unique tickers).")
    return db

def _normalize_hhmm(v):
    if pd.isna(v):
        return None
    s = re.sub(r"\D","",str(v))
    if not s:
        return None
    s = s.zfill(4)[-4:]
    hh, mm = int(s[:2]), int(s[2:])
    if 0 <= hh <= 23 and 0 <= mm <= 59:
        return f"{hh:02d}:{mm:02d}"
    return None

def lookup_first_trade_record(ticker: str):
    """Resolve a ticker conservatively.

    Exact duplicate rows with the same date/time are collapsed.
    If multiple *different* dates remain for the ticker, the result is AMBIGUOUS;
    the engine will not silently choose one.
    """
    global FIRST_TRADE_DB
    if FIRST_TRADE_DB is None:
        load_first_trade_database()
    if FIRST_TRADE_DB is None:
        return {"quality":"MISSING_DATABASE","ticker":ticker}

    t = str(ticker).upper().strip().replace(".", "-")
    x = FIRST_TRADE_DB[FIRST_TRADE_DB["Ticker"].str.replace(".","-",regex=False)==t].copy()
    if x.empty:
        return {"quality":"NOT_FOUND","ticker":ticker}

    x["time_norm"] = x["ipo_time"].map(_normalize_hhmm)
    x["date_norm"] = x["_parsed_date"].dt.strftime("%Y-%m-%d")
    # Drop duplicated companion rows that repeat the exact date/time.
    x = x.sort_values(["date_norm","time_norm","name"], na_position="last")
    unique_keys = x[["date_norm","time_norm"]].drop_duplicates()

    # If dates conflict, do not guess.
    dates = sorted(x["date_norm"].dropna().unique().tolist())
    if len(dates) > 1:
        return {
            "quality":"AMBIGUOUS",
            "ticker":ticker,
            "candidate_count":len(x),
            "candidate_dates":dates,
            "candidates":x[["Ticker","name","Date","ipo_time","Place"]].to_dict("records")
        }

    # Prefer the row with a company name and time.
    x["_rank"] = x["name"].notna().astype(int)*2 + x["time_norm"].notna().astype(int)
    row = x.sort_values("_rank", ascending=False).iloc[0]
    quality = "VERIFIED_TIME" if row["time_norm"] else "VERIFIED_DATE"
    return {
        "quality":quality,
        "ticker":ticker,
        "company":None if pd.isna(row["name"]) else str(row["name"]),
        "place":None if pd.isna(row["Place"]) else str(row["Place"]),
        "date":row["date_norm"],
        "time":row["time_norm"],
        "description":None if pd.isna(row["description"]) else str(row["description"]),
        "candidate_count":len(x),
        "source":"user-supplied first_trades_IPOs.csv",
    }

def first_trade_datetime_utc(record, market_tz="America/New_York"):
    if not record or record.get("quality") not in ("VERIFIED_TIME","VERIFIED_DATE"):
        return None
    d = pd.Timestamp(record["date"])
    hh, mm = 12, 0
    if record.get("time"):
        hh, mm = map(int, record["time"].split(":"))
    # The dataset does not provide exchange/time-zone fields. For US-listed stock scans
    # we therefore use America/New_York as an explicit modelling assumption.
    local = datetime(d.year,d.month,d.day,hh,mm,tzinfo=ZoneInfo(market_tz))
    return local.astimezone(timezone.utc)

def astro_snapshot_datetime(dt: datetime) -> Dict[str,dict]:
    if swe is None:
        raise RuntimeError("pyswisseph is required")
    if dt.tzinfo is None:
        dt = dt.replace(tzinfo=timezone.utc)
    dt = dt.astimezone(timezone.utc)
    jd = _jd(dt)
    return {name:_planet(jd,pid) for name,pid in PLANETS.items()}

def natal_chart_from_record(record, market_tz="America/New_York",
                            latitude=40.7128, longitude=-74.0060):
    """Return time-resolved natal longitudes and, when time is available, ASC/MC.

    New York coordinates are used as the exchange-location assumption for US-listed stocks.
    """
    dt = first_trade_datetime_utc(record, market_tz)
    if dt is None:
        return None
    snap = astro_snapshot_datetime(dt)
    chart = {p:snap[p]["lon"] for p in ["Sun","Moon","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]}
    chart["_datetime_utc"] = dt.isoformat()
    chart["_time_resolved"] = record.get("quality") == "VERIFIED_TIME"
    if chart["_time_resolved"]:
        try:
            cusps, ascmc = swe.houses_ex(_jd(dt), latitude, longitude, b'P')
            chart["Ascendant"] = _norm(float(ascmc[0]))
            chart["MidHeaven"] = _norm(float(ascmc[1]))
        except Exception:
            pass
    return chart

def natal_transit_events_record(date, record, orb=1.5):
    chart = natal_chart_from_record(record)
    if not chart:
        return []
    cur = astro_snapshot(pd.Timestamp(date))
    transiting=["Sun","Moon","Mars","Jupiter","Saturn","Uranus","Neptune"]
    natal_pts=["Sun","Moon","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
    if "Ascendant" in chart:
        natal_pts += ["Ascendant","MidHeaven"]
    events=[]
    for tp in transiting:
        for np_ in natal_pts:
            for deg,w,label in [(0,12,"conjunction"),(90,10,"square"),(180,10,"opposition")]:
                err=aspect_distance(cur[tp]["lon"],chart[np_],deg)
                allowed=min(orb,1.0) if tp=="Moon" else orb
                if err<=allowed:
                    angle_note = " time-resolved" if np_ in ("Ascendant","MidHeaven") else ""
                    events.append({
                        "type":"natal_hard_aspect","planet":tp,"natal_point":np_,"weight":w,
                        "reason":f"Transiting {tp} is {label} the{angle_note} first-trade natal {np_} (orb {err:.2f}°). Bucholtz/McWhirter treat hard contacts to natal points as timing alerts for trend change, volatility or breakout."
                    })
    return events


# -------------------------
# Natal / first-trade logic
# -------------------------
def natal_longitudes(first_trade: str) -> Dict[str,float]:
    d=pd.Timestamp(first_trade)
    snap=astro_snapshot(d)
    return {p:snap[p]["lon"] for p in ["Sun","Moon","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]}

def natal_transit_events(date, first_trade: str, orb=1.5) -> List[dict]:
    cur=astro_snapshot(pd.Timestamp(date))
    natal=natal_longitudes(first_trade)
    transiting=["Sun","Moon","Mars","Jupiter","Saturn","Uranus","Neptune"]
    natal_pts=["Sun","Moon","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
    events=[]
    for tp in transiting:
        for np_ in natal_pts:
            for deg,w,label in [(0,12,"conjunction"),(90,10,"square"),(180,10,"opposition")]:
                err=aspect_distance(cur[tp]["lon"],natal[np_],deg)
                # Moon gets tighter orb
                allowed = min(orb,1.0) if tp=="Moon" else orb
                if err<=allowed:
                    events.append({
                        "type":"natal_hard_aspect","planet":tp,"natal_point":np_,"weight":w,
                        "reason":f"Transiting {tp} is {label} the first-trade natal {np_} (orb {err:.2f}°). Bucholtz/McWhirter treat hard contacts to natal points as timing alerts for trend change, volatility or breakout."
                    })
    return events



def detect_declination_parallels(date, orb=1.5) -> List[dict]:
    """Parallel / contra-parallel declination events used repeatedly in the almanacs."""
    snap=astro_snapshot(pd.Timestamp(date))
    names=["Sun","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
    out=[]
    for i,a in enumerate(names):
        for b in names[i+1:]:
            da,db=snap[a]["decl"],snap[b]["decl"]
            same=abs(da-db)
            contra=abs(da+db)
            if same<=orb:
                out.append({"type":"declination_parallel","planet":f"{a}-{b}","weight":5,
                            "reason":f"{a} and {b} are in declination parallel (difference {same:.2f}°), a relationship Bucholtz repeatedly compares with price inflections in commodities."})
            elif contra<=orb:
                out.append({"type":"declination_contra_parallel","planet":f"{a}-{b}","weight":5,
                            "reason":f"{a} and {b} are in declination contra-parallel (opposition in declination, error {contra:.2f}°), a Bucholtz timing relationship."})
    return out

def quantum_levels(date, price, planets=("Jupiter","Saturn","Uranus","Neptune","Pluto")) -> List[dict]:
    """Find nearby Oreste/Bucholtz Quantum Line levels from heliocentric positions.

    Formula in the source: QL=(N*360)+PSO, PSO=heliocentric longitude*conversion scale.
    For prices below 360, inverse N form is also considered. Because many N/CS
    combinations exist, this is intentionally a *secondary* confluence measure only.
    """
    if swe is None or not np.isfinite(price): return []
    d=pd.Timestamp(date)
    dt=datetime(d.year,d.month,d.day,12,tzinfo=timezone.utc); jd=_jd(dt)
    levels=[]
    for name in planets:
        lon=_planet(jd,PLANETS[name],True)["lon"]
        for cs in [1,2,4,8,16,32]:
            pso=lon*cs
            for n in [1,2,3,4,5,6,8,16]:
                q=(n*360)+pso
                levels.append((abs(q-price),q,name,n,cs,lon))
                if price<360:
                    qi=(360/n)+pso
                    levels.append((abs(qi-price),qi,name,n,cs,lon))
    levels.sort(key=lambda x:x[0])
    out=[]
    for diff,q,name,n,cs,lon in levels[:3]:
        pct=diff/max(abs(price),1)*100
        if pct<=2.0:
            out.append({"type":"quantum_line","planet":name,"weight":3,
                        "level":q,"distance_pct":pct,
                        "reason":f"{name} heliocentric Quantum Line {q:.2f} (N={n}, CS={cs}) is {pct:.2f}% from price. Bucholtz presents Quantum Lines as potential support/resistance; this engine gives them secondary weight because N/CS choice is flexible."})
    return out

# -------------------------
# Market data / indicators
# -------------------------
def fetch_history(ticker: str, asof, years=3) -> pd.DataFrame:
    if yf is None:
        raise RuntimeError("yfinance is required. Install with: pip install yfinance")
    asof=pd.Timestamp(asof).tz_localize(None)
    end=min(pd.Timestamp.utcnow().tz_localize(None),asof)+pd.Timedelta(days=1)
    start=end-pd.Timedelta(days=int(365.25*years))
    try:
        df=yf.download(ticker,start=start.strftime('%Y-%m-%d'),end=end.strftime('%Y-%m-%d'),
                       auto_adjust=True,progress=False,threads=False)
    except Exception:
        return pd.DataFrame()
    if df is None or len(df)==0:
        return pd.DataFrame()
    if isinstance(df.columns,pd.MultiIndex):
        df.columns=[c[0] for c in df.columns]
    return df.dropna(how='all')

def _tsi(close, r=25, s=13, signal=7):
    m=close.diff()
    ema1=m.ewm(span=r,adjust=False).mean(); ema2=ema1.ewm(span=s,adjust=False).mean()
    a1=m.abs().ewm(span=r,adjust=False).mean(); a2=a1.ewm(span=s,adjust=False).mean()
    tsi=100*ema2/a2.replace(0,np.nan)
    sig=tsi.ewm(span=signal,adjust=False).mean()
    return tsi,sig

def _stoch(df,k=14,d=3):
    lo=df['Low'].rolling(k).min(); hi=df['High'].rolling(k).max()
    K=100*(df['Close']-lo)/(hi-lo).replace(0,np.nan)
    D=K.rolling(d).mean()
    return K,D

def technical_snapshot(df: pd.DataFrame) -> dict:
    if df is None or len(df)<60:
        return {"valid":False,"reason":"Insufficient price history"}
    c=df['Close'].astype(float)
    ma50=c.rolling(50).mean(); ma200=c.rolling(200).mean()
    K,D=_stoch(df)
    tsi,tsis=_tsi(c)
    last=-1; prev=-2
    price=float(c.iloc[last])
    ret5=float(c.pct_change(5).iloc[last]) if len(c)>6 else np.nan
    high20=float(df['High'].rolling(20).max().shift(1).iloc[last])
    low20=float(df['Low'].rolling(20).min().shift(1).iloc[last])
    high50=float(df['High'].rolling(50).max().shift(1).iloc[last]) if len(df)>=51 else np.nan
    low50=float(df['Low'].rolling(50).min().shift(1).iloc[last]) if len(df)>=51 else np.nan
    high100=float(df['High'].rolling(100).max().shift(1).iloc[last]) if len(df)>=101 else np.nan
    low100=float(df['Low'].rolling(100).min().shift(1).iloc[last]) if len(df)>=101 else np.nan
    atr=float((pd.concat([
        df['High']-df['Low'],
        (df['High']-c.shift()).abs(),
        (df['Low']-c.shift()).abs()],axis=1).max(axis=1).rolling(14).mean().iloc[last]))
    bull=[]; bear=[]
    score=0
    if not np.isnan(ma200.iloc[last]):
        if price>ma200.iloc[last]: bull.append("Price is above the 200-day average (broader trend bullish)."); score+=12
        else: bear.append("Price is below the 200-day average (broader trend bearish)."); score-=12
    if not np.isnan(ma50.iloc[last]):
        if price>ma50.iloc[last]: bull.append("Price is above the 50-day average (shorter-term trend bullish)."); score+=10
        else: bear.append("Price is below the 50-day average (shorter-term trend bearish)."); score-=10
    # stochastic crossover
    if not np.isnan(K.iloc[prev]) and not np.isnan(D.iloc[prev]):
        if K.iloc[prev] <= D.iloc[prev] and K.iloc[last] > D.iloc[last]:
            if K.iloc[last] < 30: score+=15; bull.append(f"Slow Stochastic bullish crossover from a low region (K {K.iloc[last]:.1f}).")
            else: score+=7; bull.append("Slow Stochastic bullish crossover.")
        if K.iloc[prev] >= D.iloc[prev] and K.iloc[last] < D.iloc[last]:
            if K.iloc[last] > 70: score-=15; bear.append(f"Slow Stochastic bearish crossover from a high region (K {K.iloc[last]:.1f}).")
            else: score-=7; bear.append("Slow Stochastic bearish crossover.")
    if not np.isnan(tsi.iloc[prev]) and not np.isnan(tsis.iloc[prev]):
        if tsi.iloc[prev]<=tsis.iloc[prev] and tsi.iloc[last]>tsis.iloc[last]: score+=10; bull.append("True Strength Index crossed above its signal line.")
        if tsi.iloc[prev]>=tsis.iloc[prev] and tsi.iloc[last]<tsis.iloc[last]: score-=10; bear.append("True Strength Index crossed below its signal line.")
    if price>high20: score+=12; bull.append("Price broke above the prior 20-session high.")
    if price<low20: score-=12; bear.append("Price broke below the prior 20-session low.")
    if not np.isnan(ret5):
        score += 4 if ret5>0 else -4
    bias="LONG" if score>=18 else "SHORT" if score<=-18 else "WATCH"
    # entry, invalidation, target as research levels
    if bias=="LONG":
        entry=max(price,float(ma50.iloc[last]) if not np.isnan(ma50.iloc[last]) else price)
        stop=price-1.5*atr
        target=price+2.5*atr
    elif bias=="SHORT":
        entry=min(price,float(ma50.iloc[last]) if not np.isnan(ma50.iloc[last]) else price)
        stop=price+1.5*atr
        target=price-2.5*atr
    else:
        entry=price; stop=price-1.5*atr; target=price+1.5*atr
    return {
        "valid":True,"price":price,"ma50":float(ma50.iloc[last]) if not np.isnan(ma50.iloc[last]) else None,
        "ma200":float(ma200.iloc[last]) if not np.isnan(ma200.iloc[last]) else None,
        "stoch_k":float(K.iloc[last]) if not np.isnan(K.iloc[last]) else None,
        "stoch_d":float(D.iloc[last]) if not np.isnan(D.iloc[last]) else None,
        "tsi":float(tsi.iloc[last]) if not np.isnan(tsi.iloc[last]) else None,
        "tsi_signal":float(tsis.iloc[last]) if not np.isnan(tsis.iloc[last]) else None,
        "atr":atr,"score":score,"bias":bias,"bull":bull,"bear":bear,
        "entry":entry,"stop":stop,"target":target,"ret5":ret5,
        "high20":high20,"low20":low20,
        "high50":None if np.isnan(high50) else high50,
        "low50":None if np.isnan(low50) else low50,
        "high100":None if np.isnan(high100) else high100,
        "low100":None if np.isnan(low100) else low100,
        "asof":pd.Timestamp(df.index[-1]).strftime('%Y-%m-%d')
    }

# -------------------------
# Stock universe
# -------------------------
def get_sp500_tickers() -> List[str]:
    try:
        tables=pd.read_html('https://en.wikipedia.org/wiki/List_of_S%26P_500_companies')
        syms=tables[0]['Symbol'].astype(str).str.replace('.','-',regex=False).tolist()
        return syms
    except Exception:
        return DEFAULT_STOCKS.copy()


# -------------------------
# V5 Lunar Confluence Engine
# -------------------------
ZODIAC_SIGNS = [
    "Aries","Taurus","Gemini","Cancer","Leo","Virgo",
    "Libra","Scorpio","Sagittarius","Capricorn","Aquarius","Pisces"
]

def zodiac_position(lon: float) -> dict:
    lon = _norm(float(lon))
    sign_index = int(lon // 30)
    deg = lon % 30
    return {
        "longitude": round(lon, 3),
        "sign": ZODIAC_SIGNS[sign_index],
        "degree_in_sign": round(deg, 2),
        "display": f"{deg:.2f}° {ZODIAC_SIGNS[sign_index]}"
    }

def lunar_phase_detail(date) -> dict:
    snap = astro_snapshot(pd.Timestamp(date))
    sun = snap["Sun"]["lon"]
    moon = snap["Moon"]["lon"]
    angle = (moon - sun) % 360.0

    phases = [(0,"New Moon"),(90,"First Quarter"),(180,"Full Moon"),(270,"Last Quarter")]
    nearest_deg, nearest_name = min(phases, key=lambda x: min(abs(angle-x[0]), 360-abs(angle-x[0])))
    sep = min(abs(angle-nearest_deg), 360-abs(angle-nearest_deg))

    if angle < 90:
        phase_name = "Waxing Crescent"
    elif angle < 180:
        phase_name = "Waxing Gibbous"
    elif angle < 270:
        phase_name = "Waning Gibbous"
    else:
        phase_name = "Waning Crescent"
    if sep <= 8:
        phase_name = nearest_name

    illumination = (1 - math.cos(math.radians(angle))) / 2.0
    return {
        "angle": round(angle,2),
        "phase": phase_name,
        "illumination_pct": round(illumination*100,1),
        "nearest_cardinal_phase": nearest_name,
        "distance_to_cardinal_phase": round(sep,2),
    }

def _lunar_aspect_class(other_planet: str, aspect_label: str) -> str:
    if aspect_label in ("sextile","trine"):
        if other_planet in ("Venus","Jupiter"):
            return "Supportive / expansionary"
        return "Supportive / trend-flow"
    if aspect_label in ("square","opposition"):
        if other_planet in ("Uranus","Pluto"):
            return "Tension / reversal-sensitive"
        return "Tension / volatility"
    if other_planet in ("Venus","Jupiter"):
        return "Supportive / expansionary"
    if other_planet in ("Mars","Saturn","Uranus","Pluto"):
        return "Volatility / reversal-sensitive"
    return "High-sensitivity timing"

def transiting_moon_aspects(date, orb=2.0) -> List[dict]:
    snap = astro_snapshot(pd.Timestamp(date))
    moon_lon = snap["Moon"]["lon"]
    planets = ["Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
    aspects = [
        (0,"conjunction",5.0),
        (60,"sextile",3.0),
        (90,"square",4.5),
        (120,"trine",3.5),
        (180,"opposition",4.5),
    ]
    out=[]
    for p in planets:
        for deg,label,base_w in aspects:
            err = aspect_distance(moon_lon, snap[p]["lon"], deg)
            if err <= orb:
                category = _lunar_aspect_class(p, label)
                exactness = max(0.0, 1.0 - err/max(orb,0.001))
                weight = round(base_w*(0.65+0.35*exactness),2)
                out.append({
                    "type":"lunar_transit_aspect",
                    "planet":p,
                    "aspect":label,
                    "angle":deg,
                    "orb":round(err,2),
                    "classification":category,
                    "weight":weight,
                    "reason":f"Transiting Moon {label} transiting {p} (orb {err:.2f}°): {category}."
                })
    return sorted(out, key=lambda x:(-x["weight"],x["orb"]))

def moon_to_natal_events(date, record=None, first_trade=None, orb=1.5) -> List[dict]:
    chart = None
    source = None
    if record and record.get("quality") in ("VERIFIED_TIME","VERIFIED_DATE"):
        chart = natal_chart_from_record(record)
        source = "time-resolved first-trade chart" if record.get("quality")=="VERIFIED_TIME" else "first-trade date chart"
    elif first_trade:
        chart = natal_longitudes(first_trade)
        source = "first-trade date chart"
    if not chart:
        return []

    cur = astro_snapshot(pd.Timestamp(date))
    moon = cur["Moon"]["lon"]
    natal_pts = ["Sun","Moon","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
    if "Ascendant" in chart:
        natal_pts += ["Ascendant","MidHeaven"]

    aspects = [
        (0,"conjunction",7.0,"High-priority timing"),
        (60,"sextile",2.5,"Supportive"),
        (90,"square",6.0,"Tension / reversal-sensitive"),
        (120,"trine",3.0,"Supportive"),
        (180,"opposition",6.0,"Tension / reversal-sensitive"),
    ]
    events=[]
    for np_ in natal_pts:
        if np_ not in chart or not isinstance(chart[np_], (int,float,np.floating)):
            continue
        for deg,label,base_w,cls in aspects:
            err = aspect_distance(moon, chart[np_], deg)
            allowed = min(orb,1.0) if np_ in ("Ascendant","MidHeaven") else orb
            if err <= allowed:
                exactness = max(0.0,1.0-err/max(allowed,0.001))
                w = round(base_w*(0.7+0.3*exactness),2)
                events.append({
                    "type":"lunar_natal_aspect",
                    "natal_point":np_,
                    "aspect":label,
                    "angle":deg,
                    "orb":round(err,2),
                    "classification":cls,
                    "weight":w,
                    "reason":f"Transiting Moon {label} natal {np_} (orb {err:.2f}°) in the {source}: {cls}."
                })
    return sorted(events, key=lambda x:(-x["weight"],x["orb"]))

def lunar_declination_detail(date) -> dict:
    d = pd.Timestamp(date).normalize()
    snap = astro_snapshot(d)
    prev = _snapshot_offset(d,-1)
    nxt = _snapshot_offset(d,1)
    dec = float(snap["Moon"]["decl"])
    pdec = float(prev["Moon"]["decl"])
    ndec = float(nxt["Moon"]["decl"])
    return {
        "declination": round(dec,2),
        "zero_crossing": bool((pdec < 0 <= dec) or (pdec > 0 >= dec)),
        "local_extremum": bool((dec >= pdec and dec >= ndec) or (dec <= pdec and dec <= ndec)),
    }

def lunar_confluence(date, record=None, first_trade=None) -> dict:
    snap = astro_snapshot(pd.Timestamp(date))
    moon_pos = zodiac_position(snap["Moon"]["lon"])
    phase = lunar_phase_detail(date)
    dec = lunar_declination_detail(date)
    transit_events = transiting_moon_aspects(date)
    natal_events = moon_to_natal_events(date, record=record, first_trade=first_trade)

    natal_weights = sorted([e["weight"] for e in natal_events], reverse=True)
    transit_weights = sorted([e["weight"] for e in transit_events], reverse=True)

    natal_score = 0.0
    if natal_weights:
        natal_score = natal_weights[0] + (0.45*natal_weights[1] if len(natal_weights)>1 else 0)
        natal_score = min(10.0,natal_score)

    transit_score = 0.0
    if transit_weights:
        transit_score = transit_weights[0] + (0.35*transit_weights[1] if len(transit_weights)>1 else 0)
        transit_score = min(7.0,transit_score)

    if phase["distance_to_cardinal_phase"] <= 3:
        phase_score = 4.0
    elif phase["distance_to_cardinal_phase"] <= 8:
        phase_score = 2.5
    else:
        phase_score = 0.0

    decl_score = 0.0
    if dec["zero_crossing"]:
        decl_score += 2.0
    if dec["local_extremum"]:
        decl_score += 2.0
    decl_score = min(3.0,decl_score)

    total = round(min(20.0,natal_score+transit_score+phase_score+decl_score),1)

    supportive = sum(1 for e in transit_events+natal_events if "Supportive" in e["classification"])
    tension = sum(1 for e in transit_events+natal_events if ("Tension" in e["classification"] or "Volatility" in e["classification"]))
    reversal = sum(1 for e in transit_events+natal_events if "reversal-sensitive" in e["classification"])

    if total >= 14:
        label = "STRONG"
    elif total >= 9:
        label = "MODERATE"
    elif total >= 5:
        label = "LIGHT"
    else:
        label = "QUIET"

    if tension > supportive and reversal:
        environment = "Tension / reversal-sensitive"
    elif tension > supportive:
        environment = "Volatility / tension"
    elif supportive > tension:
        environment = "Supportive / flowing"
    else:
        environment = "Mixed / balanced"

    reasons = [
        f"Moon position: {moon_pos['display']}; declination {dec['declination']:+.2f}°.",
        f"Lunar phase: {phase['phase']} ({phase['illumination_pct']:.1f}% illuminated; Sun-Moon angle {phase['angle']:.1f}°).",
    ]
    if dec["zero_crossing"]:
        reasons.append("Moon is crossing zero declination, a lunar timing transition.")
    if dec["local_extremum"]:
        reasons.append("Moon is near a local declination extreme, a potential timing/volatility marker.")
    reasons += [e["reason"] for e in natal_events[:4]]
    reasons += [e["reason"] for e in transit_events[:5]]

    return {
        "score": total,
        "label": label,
        "environment": environment,
        "moon_position": moon_pos,
        "phase": phase,
        "declination": dec,
        "natal_score": round(natal_score,1),
        "transit_score": round(transit_score,1),
        "phase_score": round(phase_score,1),
        "declination_score": round(decl_score,1),
        "supportive_count": supportive,
        "tension_count": tension,
        "reversal_sensitive_count": reversal,
        "natal_events": natal_events,
        "transit_events": transit_events,
        "reasons": reasons,
    }


# -------------------------
# Scanning / scoring
# -------------------------


# -------------------------
# V5 Lunar Confluence Engine
# -------------------------
ZODIAC_SIGNS = [
    "Aries","Taurus","Gemini","Cancer","Leo","Virgo",
    "Libra","Scorpio","Sagittarius","Capricorn","Aquarius","Pisces"
]

def zodiac_position(lon: float) -> dict:
    lon = _norm(float(lon))
    sign_index = int(lon // 30)
    deg = lon % 30
    return {
        "longitude": round(lon, 3),
        "sign": ZODIAC_SIGNS[sign_index],
        "degree_in_sign": round(deg, 2),
        "display": f"{deg:.2f}° {ZODIAC_SIGNS[sign_index]}"
    }

def lunar_phase_detail(date) -> dict:
    snap = astro_snapshot(pd.Timestamp(date))
    sun = snap["Sun"]["lon"]
    moon = snap["Moon"]["lon"]
    angle = (moon - sun) % 360.0

    phases = [(0,"New Moon"),(90,"First Quarter"),(180,"Full Moon"),(270,"Last Quarter")]
    nearest_deg, nearest_name = min(phases, key=lambda x: min(abs(angle-x[0]), 360-abs(angle-x[0])))
    sep = min(abs(angle-nearest_deg), 360-abs(angle-nearest_deg))

    if angle < 90:
        phase_name = "Waxing Crescent"
    elif angle < 180:
        phase_name = "Waxing Gibbous"
    elif angle < 270:
        phase_name = "Waning Gibbous"
    else:
        phase_name = "Waning Crescent"
    if sep <= 8:
        phase_name = nearest_name

    illumination = (1 - math.cos(math.radians(angle))) / 2.0
    return {
        "angle": round(angle,2),
        "phase": phase_name,
        "illumination_pct": round(illumination*100,1),
        "nearest_cardinal_phase": nearest_name,
        "distance_to_cardinal_phase": round(sep,2),
    }

def _lunar_aspect_class(other_planet: str, aspect_label: str) -> str:
    if aspect_label in ("sextile","trine"):
        if other_planet in ("Venus","Jupiter"):
            return "Supportive / expansionary"
        return "Supportive / trend-flow"
    if aspect_label in ("square","opposition"):
        if other_planet in ("Uranus","Pluto"):
            return "Tension / reversal-sensitive"
        return "Tension / volatility"
    if other_planet in ("Venus","Jupiter"):
        return "Supportive / expansionary"
    if other_planet in ("Mars","Saturn","Uranus","Pluto"):
        return "Volatility / reversal-sensitive"
    return "High-sensitivity timing"

def transiting_moon_aspects(date, orb=2.0) -> List[dict]:
    snap = astro_snapshot(pd.Timestamp(date))
    moon_lon = snap["Moon"]["lon"]
    planets = ["Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
    aspects = [
        (0,"conjunction",5.0),
        (60,"sextile",3.0),
        (90,"square",4.5),
        (120,"trine",3.5),
        (180,"opposition",4.5),
    ]
    out=[]
    for p in planets:
        for deg,label,base_w in aspects:
            err = aspect_distance(moon_lon, snap[p]["lon"], deg)
            if err <= orb:
                category = _lunar_aspect_class(p, label)
                exactness = max(0.0, 1.0 - err/max(orb,0.001))
                weight = round(base_w*(0.65+0.35*exactness),2)
                out.append({
                    "type":"lunar_transit_aspect",
                    "planet":p,
                    "aspect":label,
                    "angle":deg,
                    "orb":round(err,2),
                    "classification":category,
                    "weight":weight,
                    "reason":f"Transiting Moon {label} transiting {p} (orb {err:.2f}°): {category}."
                })
    return sorted(out, key=lambda x:(-x["weight"],x["orb"]))

def moon_to_natal_events(date, record=None, first_trade=None, orb=1.5) -> List[dict]:
    chart = None
    source = None
    if record and record.get("quality") in ("VERIFIED_TIME","VERIFIED_DATE"):
        chart = natal_chart_from_record(record)
        source = "time-resolved first-trade chart" if record.get("quality")=="VERIFIED_TIME" else "first-trade date chart"
    elif first_trade:
        chart = natal_longitudes(first_trade)
        source = "first-trade date chart"
    if not chart:
        return []

    cur = astro_snapshot(pd.Timestamp(date))
    moon = cur["Moon"]["lon"]
    natal_pts = ["Sun","Moon","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
    if "Ascendant" in chart:
        natal_pts += ["Ascendant","MidHeaven"]

    aspects = [
        (0,"conjunction",7.0,"High-priority timing"),
        (60,"sextile",2.5,"Supportive"),
        (90,"square",6.0,"Tension / reversal-sensitive"),
        (120,"trine",3.0,"Supportive"),
        (180,"opposition",6.0,"Tension / reversal-sensitive"),
    ]
    events=[]
    for np_ in natal_pts:
        if np_ not in chart or not isinstance(chart[np_], (int,float,np.floating)):
            continue
        for deg,label,base_w,cls in aspects:
            err = aspect_distance(moon, chart[np_], deg)
            allowed = min(orb,1.0) if np_ in ("Ascendant","MidHeaven") else orb
            if err <= allowed:
                exactness = max(0.0,1.0-err/max(allowed,0.001))
                w = round(base_w*(0.7+0.3*exactness),2)
                events.append({
                    "type":"lunar_natal_aspect",
                    "natal_point":np_,
                    "aspect":label,
                    "angle":deg,
                    "orb":round(err,2),
                    "classification":cls,
                    "weight":w,
                    "reason":f"Transiting Moon {label} natal {np_} (orb {err:.2f}°) in the {source}: {cls}."
                })
    return sorted(events, key=lambda x:(-x["weight"],x["orb"]))

def lunar_declination_detail(date) -> dict:
    d = pd.Timestamp(date).normalize()
    snap = astro_snapshot(d)
    prev = _snapshot_offset(d,-1)
    nxt = _snapshot_offset(d,1)
    dec = float(snap["Moon"]["decl"])
    pdec = float(prev["Moon"]["decl"])
    ndec = float(nxt["Moon"]["decl"])
    return {
        "declination": round(dec,2),
        "zero_crossing": bool((pdec < 0 <= dec) or (pdec > 0 >= dec)),
        "local_extremum": bool((dec >= pdec and dec >= ndec) or (dec <= pdec and dec <= ndec)),
    }

def lunar_confluence(date, record=None, first_trade=None) -> dict:
    snap = astro_snapshot(pd.Timestamp(date))
    moon_pos = zodiac_position(snap["Moon"]["lon"])
    phase = lunar_phase_detail(date)
    dec = lunar_declination_detail(date)
    transit_events = transiting_moon_aspects(date)
    natal_events = moon_to_natal_events(date, record=record, first_trade=first_trade)

    natal_weights = sorted([e["weight"] for e in natal_events], reverse=True)
    transit_weights = sorted([e["weight"] for e in transit_events], reverse=True)

    natal_score = 0.0
    if natal_weights:
        natal_score = natal_weights[0] + (0.45*natal_weights[1] if len(natal_weights)>1 else 0)
        natal_score = min(10.0,natal_score)

    transit_score = 0.0
    if transit_weights:
        transit_score = transit_weights[0] + (0.35*transit_weights[1] if len(transit_weights)>1 else 0)
        transit_score = min(7.0,transit_score)

    if phase["distance_to_cardinal_phase"] <= 3:
        phase_score = 4.0
    elif phase["distance_to_cardinal_phase"] <= 8:
        phase_score = 2.5
    else:
        phase_score = 0.0

    decl_score = 0.0
    if dec["zero_crossing"]:
        decl_score += 2.0
    if dec["local_extremum"]:
        decl_score += 2.0
    decl_score = min(3.0,decl_score)

    total = round(min(20.0,natal_score+transit_score+phase_score+decl_score),1)

    supportive = sum(1 for e in transit_events+natal_events if "Supportive" in e["classification"])
    tension = sum(1 for e in transit_events+natal_events if ("Tension" in e["classification"] or "Volatility" in e["classification"]))
    reversal = sum(1 for e in transit_events+natal_events if "reversal-sensitive" in e["classification"])

    if total >= 14:
        label = "STRONG"
    elif total >= 9:
        label = "MODERATE"
    elif total >= 5:
        label = "LIGHT"
    else:
        label = "QUIET"

    if tension > supportive and reversal:
        environment = "Tension / reversal-sensitive"
    elif tension > supportive:
        environment = "Volatility / tension"
    elif supportive > tension:
        environment = "Supportive / flowing"
    else:
        environment = "Mixed / balanced"

    reasons = [
        f"Moon position: {moon_pos['display']}; declination {dec['declination']:+.2f}°.",
        f"Lunar phase: {phase['phase']} ({phase['illumination_pct']:.1f}% illuminated; Sun-Moon angle {phase['angle']:.1f}°).",
    ]
    if dec["zero_crossing"]:
        reasons.append("Moon is crossing zero declination, a lunar timing transition.")
    if dec["local_extremum"]:
        reasons.append("Moon is near a local declination extreme, a potential timing/volatility marker.")
    reasons += [e["reason"] for e in natal_events[:4]]
    reasons += [e["reason"] for e in transit_events[:5]]

    return {
        "score": total,
        "label": label,
        "environment": environment,
        "moon_position": moon_pos,
        "phase": phase,
        "declination": dec,
        "natal_score": round(natal_score,1),
        "transit_score": round(transit_score,1),
        "phase_score": round(phase_score,1),
        "declination_score": round(decl_score,1),
        "supportive_count": supportive,
        "tension_count": tension,
        "reversal_sensitive_count": reversal,
        "natal_events": natal_events,
        "transit_events": transit_events,
        "reasons": reasons,
    }


# -------------------------
# Scanning / scoring — V2 grouped-confluence model
# -------------------------
def infer_stock_first_trade(ticker: str) -> Optional[str]:
    """Approximate a listed stock's first-trading date from the oldest Yahoo Finance record.

    This is explicitly labelled as an approximation in the output because corporate
    restructurings, ticker changes and missing vendor history can make the oldest
    available record differ from the legally exact first-trade date.
    """
    if yf is None:
        return None
    try:
        hist = yf.Ticker(ticker).history(period='max', auto_adjust=False)
        if hist is None or len(hist) == 0:
            return None
        return pd.Timestamp(hist.index[0]).date().isoformat()
    except Exception:
        return None


def _event_bucket(event: dict) -> str:
    """Map raw events into independent evidence families to prevent score inflation."""
    t = event.get("type", "")
    if t == "natal_hard_aspect":
        return "first_trade_natal"
    if t in ("new_moon", "full_moon"):
        return "lunation"
    if t == "retrograde":
        return "retrograde"
    if t in ("declination_zero", "declination_extreme", "declination_parallel", "declination_contra_parallel"):
        return "declination"
    if t == "aspect":
        return "planetary_aspects"
    if t in ("elongation", "perihelion", "aphelion"):
        return "orbital_cycles"
    if t == "quantum_line":
        return "quantum_lines"
    return "other"


BUCKET_CAPS = {
    # Primary Bucholtz timing families
    "first_trade_natal": 28,
    "lunation": 10,
    "retrograde": 10,
    # Secondary timing families
    "declination": 12,
    "planetary_aspects": 10,
    "orbital_cycles": 8,
    # Price geometry
    "quantum_lines": 4,
    "other": 4,
}

PRIMARY_BUCKETS = {"first_trade_natal", "lunation", "retrograde"}
SECONDARY_BUCKETS = {"declination", "planetary_aspects", "orbital_cycles"}


def grouped_timing_score(events: List[dict]) -> dict:
    """Score *independent families*, not every raw event equally.

    Within each family the score is capped. This prevents five closely-related
    aspects or declination observations from overwhelming one genuinely strong
    first-trade/natal contact.
    """
    grouped = {}
    for e in events:
        b = _event_bucket(e)
        grouped.setdefault(b, []).append(e)

    contributions = {}
    for b, items in grouped.items():
        # Use strongest event + diminishing support from further events.
        weights = sorted([max(0, float(i.get("weight", 0))) for i in items], reverse=True)
        raw = 0.0
        if weights:
            raw += weights[0]
            if len(weights) > 1:
                raw += 0.45 * weights[1]
            if len(weights) > 2:
                raw += 0.20 * sum(weights[2:4])
        contributions[b] = round(min(raw, BUCKET_CAPS.get(b, 4)), 1)

    primary = round(sum(contributions.get(b, 0) for b in PRIMARY_BUCKETS), 1)
    secondary = round(sum(contributions.get(b, 0) for b in SECONDARY_BUCKETS), 1)
    quantum = round(contributions.get("quantum_lines", 0), 1)
    total = round(primary + secondary + quantum + contributions.get("other", 0), 1)

    # Strong primary = a first-trade hard contact OR one of the canonical
    # timing windows with meaningful weight.
    strong_primary = False
    strongest_primary = None
    for e in events:
        if _event_bucket(e) in PRIMARY_BUCKETS and float(e.get("weight", 0)) >= 8:
            strong_primary = True
            if strongest_primary is None or e.get("weight", 0) > strongest_primary.get("weight", 0):
                strongest_primary = e

    return {
        "total": total,
        "primary": primary,
        "secondary": secondary,
        "quantum": quantum,
        "contributions": contributions,
        "strong_primary": strong_primary,
        "strongest_primary": strongest_primary,
        "families_present": sorted([k for k,v in contributions.items() if v > 0]),
    }


def _technical_strength(tech: dict) -> dict:
    score = float(tech.get("score", 0))
    direction = "LONG" if score >= 18 else "SHORT" if score <= -18 else "WATCH"
    strength = abs(score)
    if strength >= 35:
        label = "STRONG"
    elif strength >= 24:
        label = "MODERATE"
    elif strength >= 18:
        label = "BASIC"
    else:
        label = "WEAK"
    return {"direction": direction, "strength": strength, "label": label}


def _status_for(mode: str, requested_date, data_asof: str, eligible: bool, tech_direction: str) -> str:
    req = pd.Timestamp(requested_date).date()
    data_day = pd.Timestamp(data_asof).date()
    provisional = req > data_day

    if not eligible:
        return "NO TRADE"

    if mode == "before_market":
        # Before-market mode is deliberately a watchlist, never a claim that
        # the future session has already confirmed.
        return "WATCH DATE" if provisional else "WATCH / SETUP"

    # after_market
    if provisional:
        return "PENDING CONFIRMATION"
    return "CONFIRMED SETUP"


def _quality_label(timing: dict, tech: dict, eligible: bool) -> str:
    if not eligible:
        return "NO TRADE"
    t = timing["total"]
    s = abs(float(tech.get("score", 0)))
    if timing["strong_primary"] and t >= 30 and s >= 30:
        return "HIGH"
    if timing["strong_primary"] and t >= 22 and s >= 22:
        return "MEDIUM"
    return "LOW"


def _build_reason_sections(events: List[dict], tech: dict) -> dict:
    buckets = {}
    for e in events:
        buckets.setdefault(_event_bucket(e), []).append(e)

    def top_reasons(names, limit):
        pool = []
        for n in names:
            pool.extend(buckets.get(n, []))
        pool = sorted(pool, key=lambda x: x.get("weight", 0), reverse=True)
        return [x["reason"] for x in pool[:limit]]

    technical = tech["bull"] if tech["bias"] == "LONG" else tech["bear"] if tech["bias"] == "SHORT" else (tech["bull"] + tech["bear"])

    return {
        "primary": top_reasons(["first_trade_natal", "lunation", "retrograde"], 5),
        "secondary": top_reasons(["declination", "planetary_aspects", "orbital_cycles"], 5),
        "quantum": top_reasons(["quantum_lines"], 2),
        "technical": technical[:6],
    }


def score_idea(name, ticker, date, kind="stock", first_trade=None,
               astro_events=None, long_only=False, mode="before_market"):
    astro_events = list(astro_events or detect_cosmic_events(date))
    natal = []
    if first_trade:
        try:
            natal = natal_transit_events(date, first_trade)
        except Exception:
            natal = []

    df = fetch_history(ticker, date, 3)
    tech = technical_snapshot(df)
    if not tech.get("valid"):
        return None

    qevents = quantum_levels(date, tech["price"])
    all_events = astro_events + natal + qevents
    timing = grouped_timing_score(all_events)
    tstrength = _technical_strength(tech)

    # Hard gate:
    # 1) at least one strong PRIMARY Bucholtz timing family;
    # 2) directional technical confirmation must reach the book-inspired threshold;
    # 3) direction must be actual LONG/SHORT, not WATCH.
    eligible = bool(
        timing["strong_primary"]
        and timing["primary"] >= 8
        and abs(float(tech["score"])) >= 18
        and tech["bias"] in ("LONG", "SHORT")
    )

    if long_only and tech["bias"] == "SHORT":
        eligible = False

    status = _status_for(mode, date, tech["asof"], eligible, tech["bias"])
    bias = tech["bias"] if eligible else "NO TRADE"
    quality = _quality_label(timing, tech, eligible)

    # Ranking score with controlled, non-duplicative timing weight.
    lunar = lunar_confluence(date, first_trade=first_trade)
    combined = round(
        timing["primary"] * 1.6 +
        lunar["score"] * 0.75 +
        timing["secondary"] * 0.8 +
        timing["quantum"] * 0.4 +
        abs(float(tech["score"])) * 1.0,
        1
    )
    if not eligible:
        combined -= 20

    sections = _build_reason_sections(all_events, tech)

    return {
        "name": name,
        "ticker": ticker,
        "kind": kind,
        "date": str(pd.Timestamp(date).date()),
        "mode": mode,
        "first_trade": first_trade,
        "first_trade_note": "Yahoo oldest-history approximation" if kind == "stock" and first_trade else "published/market-history modelling anchor",
        "timing_score": timing["total"],
        "primary_score": timing["primary"],
        "secondary_score": timing["secondary"],
        "lunar_score": lunar["score"],
        "lunar_label": lunar["label"],
        "lunar_environment": lunar["environment"],
        "lunar_detail": lunar,
        "quantum_score": timing["quantum"],
        "timing_families": timing["contributions"],
        "strong_primary": timing["strong_primary"],
        "technical_score": tech["score"],
        "technical_strength": tstrength["label"],
        "combined_score": combined,
        "bias": bias,
        "status": status,
        "quality": quality,
        "price": tech["price"],
        "data_asof": tech["asof"],
        "entry": tech["entry"],
        "stop": tech["stop"],
        "target": tech["target"],
        "atr": tech["atr"],
        "reason_sections": sections,
        "astro_events": all_events,
        "technical": tech,
        "provisional": pd.Timestamp(date).date() > pd.Timestamp(tech["asof"]).date(),
    }


def scan_commodities(date, long_only=False, top_n=5, mode="before_market", include_no_trade=True):
    astro = detect_cosmic_events(date)
    out = []
    for name, meta in COMMODITIES.items():
        idea = score_idea(
            name, meta["ticker"], date, "commodity", meta["first_trade"],
            astro, long_only, mode
        )
        if idea:
            if include_no_trade or idea["status"] != "NO TRADE":
                out.append(idea)

    # Trade/watch candidates first, then no-trades by combined research score.
    out.sort(key=lambda x: (x["status"] == "NO TRADE", -x["combined_score"]))
    return out[:top_n] if top_n else out


def scan_stocks(date, universe='liquid', long_only=False, top_n=5,
                max_symbols=120, mode="before_market", include_no_trade=True):
    astro = detect_cosmic_events(date)
    tickers = DEFAULT_STOCKS if universe == 'liquid' else get_sp500_tickers()
    tickers = tickers[:max_symbols]
    out = []

    for t in tickers:
        df = fetch_history(t, date, 3)
        tech = technical_snapshot(df)
        if not tech.get("valid"):
            continue
        if abs(float(tech["score"])) < 10:
            continue

        rec = lookup_first_trade_record(t)
        natal = []
        first = None
        natal_quality = rec.get("quality","NOT_FOUND")
        company_name = rec.get("company") if isinstance(rec,dict) else None
        first_time = rec.get("time") if isinstance(rec,dict) else None

        if natal_quality in ("VERIFIED_TIME","VERIFIED_DATE"):
            first = rec.get("date")
            try:
                natal = natal_transit_events_record(date, rec)
            except Exception:
                natal = []
        elif natal_quality in ("NOT_FOUND","MISSING_DATABASE"):
            # Fallback retained, but explicitly labelled approximate.
            try:
                first = infer_stock_first_trade(t)
            except Exception:
                first = None
            if first:
                natal_quality = "YAHOO_APPROXIMATE"
                try:
                    natal = natal_transit_events(date, first)
                except Exception:
                    natal = []
        # AMBIGUOUS intentionally receives no natal score. We refuse to guess.

        qevents = quantum_levels(date, tech["price"])
        all_events = astro + natal + qevents
        timing = grouped_timing_score(all_events)

        eligible = bool(
            timing["strong_primary"]
            and timing["primary"] >= 8
            and abs(float(tech["score"])) >= 18
            and tech["bias"] in ("LONG", "SHORT")
        )
        if long_only and tech["bias"] == "SHORT":
            eligible = False

        status = _status_for(mode, date, tech["asof"], eligible, tech["bias"])
        bias = tech["bias"] if eligible else "NO TRADE"
        quality = _quality_label(timing, tech, eligible)
        lunar = lunar_confluence(
            date,
            record=rec if natal_quality in ("VERIFIED_TIME","VERIFIED_DATE") else None,
            first_trade=first if natal_quality=="YAHOO_APPROXIMATE" else None
        )
        combined = round(
            timing["primary"] * 1.6 +
            lunar["score"] * 0.75 +
            timing["secondary"] * 0.8 +
            timing["quantum"] * 0.4 +
            abs(float(tech["score"])) -
            (20 if not eligible else 0),
            1
        )
        sections = _build_reason_sections(all_events, tech)

        if natal_quality=="AMBIGUOUS":
            source_note = f"AMBIGUOUS dataset ticker history — candidate dates: {', '.join(rec.get('candidate_dates',[]))}; natal scoring withheld"
        elif natal_quality=="VERIFIED_TIME":
            source_note = f"VERIFIED dataset first trade {first} {first_time} America/New_York assumption; time-resolved natal chart with ASC/MC"
        elif natal_quality=="VERIFIED_DATE":
            source_note = f"VERIFIED dataset first-trade date {first}; no usable time, noon-date natal chart"
        elif natal_quality=="YAHOO_APPROXIMATE":
            source_note = f"Dataset record unavailable; Yahoo oldest-history fallback {first}"
        else:
            source_note = "No usable first-trade record; natal scoring unavailable"

        idea = {
            "name": company_name or t, "ticker": t, "kind": "stock",
            "date": str(pd.Timestamp(date).date()), "mode": mode,
            "first_trade": first,
            "first_trade_time": first_time,
            "natal_data_quality": natal_quality,
            "first_trade_note": source_note,
            "first_trade_record": rec,
            "timing_score": timing["total"],
            "primary_score": timing["primary"],
            "secondary_score": timing["secondary"],
            "lunar_score": lunar["score"],
            "lunar_label": lunar["label"],
            "lunar_environment": lunar["environment"],
            "lunar_detail": lunar,
            "quantum_score": timing["quantum"],
            "timing_families": timing["contributions"],
            "strong_primary": timing["strong_primary"],
            "technical_score": tech["score"],
            "technical_strength": _technical_strength(tech)["label"],
            "combined_score": combined,
            "bias": bias, "status": status, "quality": quality,
            "price": tech["price"], "data_asof": tech["asof"],
            "entry": tech["entry"], "stop": tech["stop"], "target": tech["target"],
            "atr": tech["atr"], "reason_sections": sections,
            "astro_events": all_events, "technical": tech,
            "provisional": pd.Timestamp(date).date() > pd.Timestamp(tech["asof"]).date()
        }
        if include_no_trade or status != "NO TRADE":
            out.append(idea)

    out.sort(key=lambda x: (x["status"] == "NO TRADE", -x["combined_score"]))
    return out[:top_n] if top_n else out

def summary_dataframe(ideas: List[dict]) -> pd.DataFrame:
    rows = []
    for i in ideas:
        rows.append({
            "Market": i["name"],
            "Ticker": i["ticker"],
            "Status": i["status"],
            "Bias": i["bias"],
            "Quality": i["quality"],
            "Primary": i["primary_score"],
            "Lunar": i.get("lunar_score",0),
            "Lunar env": i.get("lunar_environment",""),
            "Secondary": i["secondary_score"],
            "Technical": i["technical_score"],
            "Combined": i["combined_score"],
            "Natal data": i.get("natal_data_quality","PUBLISHED" if i.get("kind")=="commodity" else "UNKNOWN"),
            "Data through": i["data_asof"],
        })
    return pd.DataFrame(rows)


def format_idea(idea: dict) -> str:
    if not idea:
        return "No idea."
    provisional = "PROVISIONAL" if idea["provisional"] else "date-confirmed data available"
    rs = idea["reason_sections"]
    lines = [
        f"{idea['name']} ({idea['ticker']}) — {idea['status']} — {idea['bias']}",
        f"Date: {idea['date']} | Mode: {idea['mode']} | Price data through: {idea['data_asof']} | {provisional}",
        f"Quality: {idea['quality']} | Primary timing: {idea['primary_score']} | Lunar: {idea.get('lunar_score',0)} | Secondary timing: {idea['secondary_score']} | Technical: {idea['technical_score']} | Combined: {idea['combined_score']}",
        f"Natal data quality: {idea.get('natal_data_quality', 'PUBLISHED' if idea.get('kind')=='commodity' else 'UNKNOWN')}",
        f"First-trade/natal reference: {idea.get('first_trade') or 'not available'} ({idea.get('first_trade_note','')})",
        f"Research levels: confirmation/entry {idea['entry']:.2f} | invalidation {idea['stop']:.2f} | initial objective {idea['target']:.2f}",
        "",
        "A. PRIMARY BUCHOLTZ TIMING",
    ]
    lines += [f"  • {r}" for r in rs["primary"]] or ["  • No qualifying primary timing event."]
    lines += ["", "B. LUNAR CONFLUENCE"]
    lunar = idea.get("lunar_detail",{})
    lines += [f"  • {r}" for r in lunar.get("reasons",[])[:8]] or ["  • No material lunar confluence."]
    lines += ["", "C. SECONDARY ASTRONOMICAL CONFIRMATION"]
    lines += [f"  • {r}" for r in rs["secondary"]] or ["  • No secondary timing confluence."]
    lines += ["", "D. QUANTUM / PRICE GEOMETRY"]
    lines += [f"  • {r}" for r in rs["quantum"]] or ["  • No nearby Quantum Line confluence."]
    lines += ["", "E. TECHNICAL CONFIRMATION"]
    lines += [f"  • {r}" for r in rs["technical"]] or ["  • No directional technical confirmation."]
    lines += [
        "",
        "DECISION RULE",
        "  • A setup is only promoted when there is at least one strong PRIMARY Bucholtz timing event plus directional technical confirmation.",
        "  • Lunar confluence is scored separately and capped so multiple Moon contacts cannot manufacture a high score.",
        "  • Multiple related aspects inside the same family are capped so they cannot manufacture a high score.",
        "  • Before-market mode produces a WATCHLIST; after-market mode can label a setup CONFIRMED only when price data for the requested date exists.",
        "  • Financial astrology is treated as a timing hypothesis, not a scientifically established forecasting method.",
    ]
    return "\n".join(lines)
```


## V8 — Bill Meridian + applying-aspect execution layer

The Meridian module is deliberately **independent** from the Bucholtz score. It uses only rules that can be stated consistently from the supplied book: first-trade natal signatures, prominent natal aspects, recurring bullish/bearish transit combinations, and eclipse/station sensitivity. Qualitative case-study observations are not silently converted into hard rules.

Applying aspects are favoured for execution timing. Intraday windows are calculated only for the final shortlisted ideas so the full stock scan stays reasonably fast.


```python

# ================================================================
# V8 MERIDIAN + APPLYING EXECUTION EXTENSION
# ================================================================
from zoneinfo import ZoneInfo
from datetime import datetime, timedelta
import math, copy

V8_VERSION = "8.1"

# User-selected applying-aspect geometry from the earlier configuration screen.
APPLYING_ASPECTS = {
    "conjunction": {"angle":0.0,   "orb":5.0, "score": 5.0},
    "opposition":  {"angle":180.0, "orb":5.0, "score":-5.0},
    "trine":       {"angle":120.0, "orb":3.0, "score": 3.0},
    "sextile":     {"angle":60.0,  "orb":3.0, "score": 2.0},
    "square":      {"angle":90.0,  "orb":3.0, "score":-3.0},
    "quincunx":    {"angle":150.0, "orb":2.0, "score":-1.0},
    "semisextile": {"angle":30.0,  "orb":2.0, "score": 1.0},
    "semisquare":  {"angle":45.0,  "orb":2.0, "score":-1.0},
    "sesquisquare": {"angle":135.0,"orb":2.0, "score":-1.0},
}

# Source-derived natal sign table (Meridian, Part 2: prominent natal signs).
MERIDIAN_NATAL_SIGNS = {
    "Sun":      {"bullish":{"Taurus"},              "bearish":{"Sagittarius","Capricorn"}},
    "Moon":     {"bullish":{"Taurus"},              "bearish":{"Scorpio"}},
    "Mercury":  {"bullish":{"Taurus","Gemini"},    "bearish":{"Cancer"}},
    "Venus":    {"bullish":{"Scorpio"},             "bearish":{"Capricorn","Aquarius"}},
    "Mars":     {"bullish":{"Gemini"},              "bearish":{"Virgo"}},
    "Jupiter":  {"bullish":{"Aries","Pisces"},     "bearish":{"Aquarius"}},
    "Saturn":   {"bullish":{"Scorpio"},             "bearish":{"Scorpio"}},
    "Uranus":   {"bullish":{"Sagittarius"},         "bearish":{"Sagittarius"}},
    "Neptune":  {"bullish":{"Capricorn"},           "bearish":{"Sagittarius"}},
    "Pluto":    {"bullish":{"Scorpio"},             "bearish":{"Libra"}},
    "Ascendant":{"bullish":{"Cancer"},              "bearish":{"Aquarius"}},
    "MidHeaven":{"bullish":{"Aries"},               "bearish":{"Aquarius"}},
}

# Source-derived prominent natal aspect table.  Values are (aspect angle, orb).
# "Sun-Mercury conjunction" is retained as a low-confidence item because the book
# explicitly notes that its apparent bearish frequency is largely structural.
MERIDIAN_NATAL_ASPECTS = {
    "bullish": [
        ("Jupiter","Uranus",0,6,2.4),
        ("Venus","Mars",45,3,1.4),
        ("Mars","Node",60,4,1.6),
        ("Moon","Ascendant",60,4,1.4),
        ("Moon","Jupiter",60,4,1.6),
        ("Mars","MidHeaven",90,6,1.7),
        ("Uranus","MidHeaven",90,6,1.7),
        ("Mercury","Pluto",120,4,1.5),
        ("Uranus","Node",120,4,1.5),
        ("Venus","Uranus",135,3,1.3),
        ("Moon","Uranus",150,3,1.2),
        ("Pluto","Node",180,6,1.7),
        ("Uranus","Node",180,6,1.7),
    ],
    "bearish": [
        ("Venus","MidHeaven",0,6,1.0),
        ("Sun","Mercury",0,6,0.5),
        ("Venus","Jupiter",0,6,1.4),
        ("Venus","Mars",0,6,1.4),
        ("Mercury","Venus",45,3,1.2),
        ("Sun","Neptune",45,3,1.5),
        ("Moon","Jupiter",45,3,1.4),
        ("Venus","Neptune",60,4,1.4),
        ("Moon","Mars",60,4,1.4),
        ("Venus","Neptune",90,6,1.7),
        ("Saturn","Node",90,6,1.7),
        ("Mars","Sun",90,6,1.6),
        ("Uranus","Sun",90,6,1.6),
        ("Sun","Pluto",120,4,1.4),
        ("Venus","Saturn",135,3,1.4),
        ("Venus","Pluto",135,3,1.4),
        ("Saturn","Neptune",180,6,1.8),
        ("Mars","Uranus",180,6,1.7),
    ]
}

# Meridian's "bullish and bearish planetary combinations by transit".
# We model a "contact" with standard major aspects; applying contacts receive more weight.
MERIDIAN_TRANSIT_PAIRS = {
    "bullish": {
        ("Sun","Venus"), ("Jupiter","Sun"), ("Jupiter","Mercury"),
        ("Jupiter","Venus"), ("Jupiter","Jupiter"), ("Jupiter","Uranus"),
        ("Jupiter","Neptune"), ("Jupiter","Pluto"),
        ("Jupiter","Ascendant"), ("Jupiter","MidHeaven"),
        ("Uranus","Ascendant"), ("Uranus","MidHeaven"),
        ("Neptune","Sun"), ("Neptune","Jupiter")
    },
    "bearish": {
        ("Saturn","Sun"), ("Saturn","Mercury"), ("Saturn","Venus"),
        ("Saturn","Jupiter"), ("Saturn","Uranus"), ("Saturn","Neptune"),
        ("Saturn","Pluto"), ("Saturn","Ascendant"), ("Saturn","MidHeaven"),
        ("Neptune","Pluto"), ("Pluto","Ascendant"), ("Pluto","MidHeaven")
    }
}

MAJOR_CONTACTS = [(0,"conjunction"),(60,"sextile"),(90,"square"),(120,"trine"),(180,"opposition")]


def _sign_of(lon):
    return ZODIAC_SIGNS[int(_norm(lon)//30)]


def _chart_from_first_trade(first_trade=None, record=None):
    if record and isinstance(record, dict) and record.get("quality") in ("VERIFIED_TIME","VERIFIED_DATE"):
        try:
            return natal_chart_from_record(record)
        except Exception:
            pass
    if first_trade:
        try:
            return natal_longitudes(first_trade)
        except Exception:
            return None
    return None


def _chart_node(first_trade=None, record=None):
    """Add a geocentric True Node where possible; keeps Node rules separate from planets."""
    chart = _chart_from_first_trade(first_trade, record)
    if not chart:
        return None
    try:
        if record and isinstance(record,dict):
            dt=first_trade_datetime_utc(record)
        else:
            d=pd.Timestamp(first_trade)
            dt=datetime(d.year,d.month,d.day,12,0,tzinfo=timezone.utc)
        if dt:
            ecl,_ = swe.calc_ut(_jd(dt), swe.TRUE_NODE, swe.FLG_SWIEPH | swe.FLG_SPEED)
            chart["Node"]=_norm(float(ecl[0]))
    except Exception:
        pass
    return chart


def meridian_natal_signature(chart):
    if not chart:
        return {"score":0.0,"strength":0.0,"reasons":[],"bull":0.0,"bear":0.0}
    bull=bear=0.0; reasons=[]
    # sign frequencies are weak/moderate context, never a trade signal by themselves
    for p, rules in MERIDIAN_NATAL_SIGNS.items():
        if p not in chart: continue
        sign=_sign_of(chart[p])
        if sign in rules["bullish"] and sign not in rules["bearish"]:
            bull += 1.0; reasons.append(f"Meridian natal-sign context: {p} in {sign} appeared more often in his outperformer sample.")
        elif sign in rules["bearish"] and sign not in rules["bullish"]:
            bear += 1.0; reasons.append(f"Meridian natal-sign context: {p} in {sign} appeared more often in his underperformer sample.")
        elif sign in rules["bullish"] and sign in rules["bearish"]:
            reasons.append(f"Meridian natal-sign context: {p} in {sign} appeared on both sides of the sample and is treated as neutral.")

    for side, rows in MERIDIAN_NATAL_ASPECTS.items():
        for a,b,ang,orb,w in rows:
            if a not in chart or b not in chart: continue
            err=aspect_distance(chart[a],chart[b],ang)
            if err <= orb:
                # exactness taper prevents wide textbook orbs dominating
                val=w * max(0.25, 1.0-0.65*(err/orb))
                if side=="bullish": bull += val
                else: bear += val
                reasons.append(f"Meridian natal-aspect {side}: {a}–{b} near {ang:.0f}° (orb {err:.2f}°).")
    raw=bull-bear
    score=max(-30.0,min(30.0,raw*4.0))
    return {"score":round(score,1),"strength":round(min(30.0,(bull+bear)*3.0),1),"reasons":reasons[:14],"bull":round(bull,2),"bear":round(bear,2)}


def _snapshot_at(dt):
    return astro_snapshot_datetime(dt)


def _contact_state(transit_lon, natal_lon, angle, orb):
    err=aspect_distance(transit_lon,natal_lon,angle)
    return err, err<=orb


def meridian_transit_contacts(date, chart, applying_only=False, reference_hour_et=10):
    if not chart:
        return {"score":0.0,"strength":0.0,"reasons":[],"contacts":[]}
    et=ZoneInfo("America/New_York")
    d=pd.Timestamp(date).date()
    dt=datetime(d.year,d.month,d.day,reference_hour_et,0,tzinfo=et).astimezone(timezone.utc)
    now=_snapshot_at(dt); later=_snapshot_at(dt+timedelta(hours=6))
    bull=bear=0.0; contacts=[]
    for side,pairs in MERIDIAN_TRANSIT_PAIRS.items():
        for tp,np_ in pairs:
            if tp not in now or np_ not in chart: continue
            best=None
            for ang,label in MAJOR_CONTACTS:
                orb=2.0 if tp not in ("Moon",) else 1.5
                err=aspect_distance(now[tp]["lon"],chart[np_],ang)
                if err<=orb and (best is None or err<best[0]):
                    err_later=aspect_distance(later[tp]["lon"],chart[np_],ang)
                    applying=err_later < err
                    best=(err,label,ang,applying,err_later)
            if not best: continue
            err,label,ang,applying,err_later=best
            if applying_only and not applying: continue
            exact=max(0.25,1.0-err/2.2)
            w=2.0*exact*(1.25 if applying else 0.70)
            if side=="bullish": bull+=w
            else: bear+=w
            contacts.append({"side":side,"transit":tp,"natal":np_,"aspect":label,"orb":round(err,2),"applying":applying,"weight":round(w,2)})
    # cap repeated evidence; Meridian layer should not overpower price confirmation
    bull=min(bull,10.0); bear=min(bear,10.0)
    raw=bull-bear
    score=max(-50,min(50,raw*5.0))
    reasons=[f"Meridian transit {c['side']}: {c['transit']} {c['aspect']} natal {c['natal']} (orb {c['orb']:.2f}°, {'applying' if c['applying'] else 'separating'})." for c in sorted(contacts,key=lambda x:-x['weight'])[:12]]
    return {"score":round(score,1),"strength":round(min(50,(bull+bear)*5.0),1),"reasons":reasons,"contacts":contacts,"bull":round(bull,2),"bear":round(bear,2)}


def meridian_station_eclipse_timing(date):
    """Non-directional predictive-technique timing family: stations + eclipse geometry."""
    d=pd.Timestamp(date)
    snap=astro_snapshot(d)
    reasons=[]; points=0.0
    thresholds={"Mercury":0.35,"Venus":0.18,"Mars":0.10,"Jupiter":0.035,"Saturn":0.02,"Uranus":0.008,"Neptune":0.006,"Pluto":0.004}
    for p,t in thresholds.items():
        sp=abs(float(snap[p]["speed_lon"]))
        if sp<=t:
            closeness=max(0.2,1-sp/max(t,1e-6))
            pts=1.0+1.5*closeness
            points += pts
            reasons.append(f"Meridian predictive-technique timing: {p} is near-stationary (|speed| {sp:.4f}°/day).")
    try:
        jd=_jd(datetime(d.year,d.month,d.day,12,tzinfo=timezone.utc))
        node,_=swe.calc_ut(jd,swe.TRUE_NODE,swe.FLG_SWIEPH)
        node_lon=_norm(node[0])
        sun=snap['Sun']['lon']; moon=snap['Moon']['lon']
        phase=min(angular_distance(sun,moon), abs(180-angular_distance(sun,moon)))
        node_sep=min(angular_distance(moon,node_lon), angular_distance(moon,_norm(node_lon+180)))
        if phase<=12 and node_sep<=18:
            points += 3.0
            kind="new/full-moon eclipse geometry"
            reasons.append(f"Meridian eclipse-sensitive timing: {kind} is nearby (phase error {phase:.1f}°, node distance {node_sep:.1f}°).")
    except Exception:
        pass
    return {"score":round(min(10.0,points),1),"reasons":reasons[:8]}


def meridian_analysis(date, first_trade=None, record=None, applying_only=True):
    chart=_chart_node(first_trade,record)
    natal=meridian_natal_signature(chart)
    trans=meridian_transit_contacts(date,chart,applying_only=applying_only)
    timing=meridian_station_eclipse_timing(date)
    directional=max(-100,min(100,natal['score']+trans['score']))
    strength=min(100,abs(directional)+timing['score']*2.0)
    direction='BULLISH' if directional>=12 else 'BEARISH' if directional<=-12 else 'MIXED/NEUTRAL'
    return {"directional_score":round(directional,1),"strength":round(strength,1),"direction":direction,
            "natal":natal,"transits":trans,"timing":timing,"chart_available":bool(chart)}


def _best_aspect(lon1, lon2, aspect_defs=APPLYING_ASPECTS):
    """Return the nearest configured aspect while preserving BOTH actual and allowed orb.

    V8.0 accidentally overwrote the actual orb with the maximum allowed orb because
    both used the same dictionary key.  V8.1 keeps them separate.  This is essential
    for a genuine applying-only test.
    """
    best=None
    for label,cfg in aspect_defs.items():
        err=aspect_distance(lon1,lon2,cfg['angle'])
        if err<=cfg['orb'] and (best is None or err<best['current_orb']):
            best={
                "label":label,
                "angle":float(cfg['angle']),
                "max_orb":float(cfg['orb']),
                "base_score":float(cfg['score']),
                "current_orb":float(err),
            }
    return best


def _unordered_pair(a,b):
    return frozenset((str(a),str(b)))

# Exact Moon-related patterns that Meridian reports in the prominent natal-aspect table.
# These are not silently treated as proof that the same transit must have the same outcome;
# the report labels them as "natal-pattern support" when used for execution context.
MERIDIAN_MOON_ASPECT_PATTERNS = {
    (_unordered_pair('Moon','Ascendant'), 'sextile'): ('POSITIVE','DIRECT NATAL-PATTERN'),
    (_unordered_pair('Moon','Jupiter'), 'sextile'): ('POSITIVE','DIRECT NATAL-PATTERN'),
    (_unordered_pair('Moon','Jupiter'), 'semisquare'): ('NEGATIVE','DIRECT NATAL-PATTERN'),
    (_unordered_pair('Moon','Mars'), 'sextile'): ('NEGATIVE','DIRECT NATAL-PATTERN'),
    (_unordered_pair('Moon','Uranus'), 'quincunx'): ('POSITIVE','DIRECT NATAL-PATTERN'),
}


def _meridian_pair_direction(transiting, natal):
    """Direct directional classification from Meridian's transit-combination table."""
    pair=(str(transiting),str(natal))
    if pair in MERIDIAN_TRANSIT_PAIRS.get('bullish',set()): return 'POSITIVE'
    if pair in MERIDIAN_TRANSIT_PAIRS.get('bearish',set()): return 'NEGATIVE'
    return None


def meridian_execution_polarity(mover, aspect, target, relation):
    """Conservative polarity label for an applying execution contact.

    Priority:
      1) direct Meridian transit-pair classification for transit -> natal contacts;
      2) exact Moon-related natal aspect patterns reported by Meridian (clearly labelled);
      3) a limited Meridian-informed inference from his planet descriptions, never presented
         as a direct book rule; otherwise UNCLASSIFIED.

    The limited inference allows the requested examples without pretending Meridian published
    a complete Moon-aspect lookup table: soft Moon-Jupiter is positive; hard Moon-Saturn is
    negative. Uranus/Neptune/Pluto are generally left contextual unless the pair is direct.
    """
    mover=str(mover); target=str(target); aspect=str(aspect)
    soft={'trine','sextile','semisextile'}
    hard={'square','opposition','semisquare','sesquisquare','quincunx'}

    # Direct transit -> natal table from Meridian, printed p.21.
    if relation=='natal':
        direct=_meridian_pair_direction(mover,target)
        if direct:
            return {"direction":direct,"source":"DIRECT TRANSIT TABLE","confidence":1.0,
                    "note":f"Meridian lists {mover}-{target} among {'bullish' if direct=='POSITIVE' else 'bearish'} transit combinations."}

    # Exact Moon pair + aspect in Meridian's prominent natal-aspect statistics, printed p.23.
    key=(_unordered_pair(mover,target),aspect)
    if key in MERIDIAN_MOON_ASPECT_PATTERNS:
        direction,src=MERIDIAN_MOON_ASPECT_PATTERNS[key]
        return {"direction":direction,"source":src,"confidence":0.72,
                "note":"Same Moon pair/aspect appears in Meridian's prominent natal-aspect sample; used here only as contextual support."}

    # Limited source-informed inference from Meridian's planet descriptions (printed p.20).
    # Jupiter is described as beneficial/positive; Saturn as bearish/depressing.
    if mover=='Moon' or target=='Moon':
        other=target if mover=='Moon' else mover
        if other=='Jupiter' and aspect in soft|{'conjunction'}:
            return {"direction":"POSITIVE","source":"MERIDIAN-INFORMED INFERENCE","confidence":0.55,
                    "note":"Meridian describes Jupiter as beneficial/positive; harmonious Moon-Jupiter contact is treated as a positive timing bias, not a direct published transit rule."}
        if other=='Saturn' and aspect in hard|{'conjunction'}:
            return {"direction":"NEGATIVE","source":"MERIDIAN-INFORMED INFERENCE","confidence":0.55,
                    "note":"Meridian describes Saturn as bearish/depressing; hard Moon-Saturn contact is treated as a negative timing bias, not a direct published transit rule."}
        if other=='Mars' and aspect in hard:
            return {"direction":"NEGATIVE","source":"MERIDIAN-INFORMED INFERENCE","confidence":0.45,
                    "note":"Meridian describes Mars as adding energy and hard contacts as potentially disruptive; treated as negative/volatile rather than a direct directional rule."}
        if other=='Venus' and aspect in soft:
            return {"direction":"POSITIVE","source":"MERIDIAN-INFORMED INFERENCE","confidence":0.45,
                    "note":"Harmonious Venus contact is treated as mildly positive context; not a direct Meridian Moon-transit rule."}

    # Transit-to-transit: if the same ordered planetary pair is in Meridian's table,
    # preserve the direction but explicitly flag the extension beyond transit-to-natal.
    if relation=='transit':
        direct=_meridian_pair_direction(mover,target)
        if direct:
            return {"direction":direct,"source":"PAIR-SUPPORTED / T→T EXTENSION","confidence":0.50,
                    "note":"Planet pair is directional in Meridian's transit table; applying it transit-to-transit is an explicit model extension."}
        reverse=_meridian_pair_direction(target,mover)
        if reverse:
            return {"direction":reverse,"source":"PAIR-SUPPORTED / T→T EXTENSION","confidence":0.45,
                    "note":"Reverse planet pair is directional in Meridian's transit table; transit-to-transit use is an explicit model extension."}

    return {"direction":"UNCLASSIFIED","source":"NO DIRECT MERIDIAN RULE","confidence":0.0,
            "note":"No direct Meridian rule was found for this exact applying contact; it contributes timing strength but not directional polarity."}


def _execution_hit_weight(a, family_mult=1.0, polarity_conf=0.0):
    # exactness: 1.0 at exact aspect, fades toward zero at the configured outer orb.
    exactness=max(0.0,1.0-(a['current_orb']/max(a['max_orb'],1e-9)))
    # Retain a small floor so a newly-entered applying window is visible, but never full strength.
    base=0.30 + 0.70*exactness
    # Aspect importance uses the user's configured geometry, while direction is handled separately.
    aspect_mag=min(1.0,abs(a['base_score'])/5.0)
    return family_mult * (0.55 + 0.45*aspect_mag) * base * (1.0 + 0.10*polarity_conf)


def execution_timing_v8(date, first_trade=None, record=None, step_minutes=15):
    """Audited intraday applying-only scan.

    V8.1 fixes V8.0's orb overwrite bug, requires the aspect to become measurably
    more exact over the next interval, weights by exactness, caps independent families,
    and reports Meridian polarity separately from timing strength.
    """
    chart=_chart_node(first_trade,record)
    et=ZoneInfo('America/New_York')
    d=pd.Timestamp(date).date()
    start=datetime(d.year,d.month,d.day,9,30,tzinfo=et)
    end=datetime(d.year,d.month,d.day,16,0,tzinfo=et)
    rows=[]; dt=start

    while dt<=end:
        utc=dt.astimezone(timezone.utc)
        now=_snapshot_at(utc); nxt=_snapshot_at(utc+timedelta(minutes=step_minutes))
        hits=[]

        def add_hit(family,mover,target,relation,a,later_err,family_mult):
            # Genuine applying-only condition. Require a small improvement to avoid floating noise.
            if later_err >= a['current_orb'] - 1e-4:
                return
            pol=meridian_execution_polarity(mover,a['label'],target,relation)
            strength=_execution_hit_weight(a,family_mult,pol['confidence'])
            # Direction is deliberately independent from timing strength.
            sign=1 if pol['direction']=='POSITIVE' else -1 if pol['direction']=='NEGATIVE' else 0
            directional=strength*sign*pol['confidence']
            hits.append({
                'family':family,'mover':mover,'target':target,'relation':relation,
                'aspect':a['label'],'orb':a['current_orb'],'max_orb':a['max_orb'],
                'strength':strength,'directional':directional,'polarity':pol,
            })

        # 1) Moon -> natal
        if chart:
            for np_ in [x for x in ['Sun','Moon','Mercury','Venus','Mars','Jupiter','Saturn','Uranus','Neptune','Pluto','Ascendant','MidHeaven','Node'] if x in chart]:
                a=_best_aspect(now['Moon']['lon'],chart[np_])
                if a:
                    later=aspect_distance(nxt['Moon']['lon'],chart[np_],a['angle'])
                    add_hit('moon_natal','Moon',np_,'natal',a,later,1.35)

        # 2) Moon -> transiting planets
        for p in ['Sun','Mercury','Venus','Mars','Jupiter','Saturn','Uranus','Neptune','Pluto']:
            a=_best_aspect(now['Moon']['lon'],now[p]['lon'])
            if a:
                later=aspect_distance(nxt['Moon']['lon'],nxt[p]['lon'],a['angle'])
                add_hit('moon_transit','Moon',p,'transit',a,later,1.00)

        # 3) Inner planets -> outer planets
        for ip in ['Mercury','Venus','Mars']:
            for op in ['Jupiter','Saturn','Uranus','Neptune','Pluto']:
                a=_best_aspect(now[ip]['lon'],now[op]['lon'])
                if a:
                    later=aspect_distance(nxt[ip]['lon'],nxt[op]['lon'],a['angle'])
                    add_hit('inner_outer',ip,op,'transit',a,later,0.80)

        if hits:
            # Cap by independent family, not by raw number of aspects.
            family_caps={'moon_natal':4.5,'moon_transit':3.0,'inner_outer':2.5}
            fam_strength={}
            for fam,cap in family_caps.items():
                vals=sorted([h['strength'] for h in hits if h['family']==fam],reverse=True)
                # strongest contact dominates; second adds only modest confirmation
                val=(vals[0] if vals else 0.0) + (0.35*vals[1] if len(vals)>1 else 0.0)
                fam_strength[fam]=min(cap,val*3.0)
            strength=min(10.0,sum(fam_strength.values()))

            directional=sum(h['directional'] for h in hits)
            # Normalize polarity to a readable -10..+10 scale independent of strength.
            denom=max(1e-9,sum(abs(h['directional']) for h in hits))
            net=0.0 if denom==0 else 10.0*directional/denom
            positives=sum(1 for h in hits if h['polarity']['direction']=='POSITIVE')
            negatives=sum(1 for h in hits if h['polarity']['direction']=='NEGATIVE')
            unclassified=sum(1 for h in hits if h['polarity']['direction']=='UNCLASSIFIED')
            if net>=2.0: pol_label='POSITIVE'
            elif net<=-2.0: pol_label='NEGATIVE'
            else: pol_label='MIXED / NEUTRAL'

            hits_sorted=sorted(hits,key=lambda h:(-h['strength'],h['orb']))
            reasons=[]
            for h in hits_sorted[:7]:
                p=h['polarity']
                reasons.append(
                    f"{p['direction']} [{p['source']}] — {h['mover']} {h['aspect']} "
                    f"{'natal ' if h['relation']=='natal' else 'transiting '}{h['target']} "
                    f"(orb {h['orb']:.2f}° applying; limit {h['max_orb']:.1f}°). {p['note']}"
                )
            rows.append({
                'time_et':dt.strftime('%H:%M'),'strength':round(strength,1),
                'polarity_score':round(net,1),'polarity_label':pol_label,
                'positive_hits':positives,'negative_hits':negatives,'unclassified_hits':unclassified,
                'family_strength':{k:round(v,2) for k,v in fam_strength.items()},
                'reasons':reasons,
            })
        dt += timedelta(minutes=step_minutes)

    if not rows:
        return {'score':0.0,'label':'NO QUALIFYING WINDOW','best_time':None,'windows':[],
                'reasons':[],'polarity_score':0.0,'polarity_label':'UNCLASSIFIED'}

    # Highest strength first; if tied, prefer stronger absolute directional evidence, then later
    # timestamp rather than mechanically selecting the first row of a flat plateau.
    rows=sorted(rows,key=lambda r:(-r['strength'],-abs(r['polarity_score']),r['time_et']))
    best=rows[0]
    label='STRONG' if best['strength']>=7 else 'MODERATE' if best['strength']>=4 else 'LIGHT'
    return {
        'score':best['strength'],'label':label,'best_time':best['time_et'],'windows':rows[:8],
        'reasons':best['reasons'],'polarity_score':best['polarity_score'],
        'polarity_label':best['polarity_label'],'positive_hits':best['positive_hits'],
        'negative_hits':best['negative_hits'],'unclassified_hits':best['unclassified_hits'],
        'family_strength':best['family_strength'],
    }


def _apply_meridian_to_idea(idea, record=None):
    if not idea: return idea
    first=idea.get('first_trade')
    mer=meridian_analysis(idea['date'],first_trade=first,record=record or idea.get('first_trade_record'),applying_only=True)
    idea['meridian']=mer
    idea['meridian_score']=mer['directional_score']
    idea['meridian_strength']=mer['strength']
    idea['meridian_direction']=mer['direction']
    # alignment contribution is deliberately bounded to +/- 12 points
    tdir=idea.get('technical',{}).get('bias') or idea.get('bias')
    ms=mer['directional_score']
    if tdir=='LONG': align=max(-12,min(12,ms*0.16))
    elif tdir=='SHORT': align=max(-12,min(12,-ms*0.16))
    else: align=0
    idea['meridian_alignment']=round(align,1)
    idea['combined_score']=round(float(idea.get('combined_score',0))+align,1)
    idea['independent_families_confirmed']=sum([
        bool(idea.get('strong_primary')),
        float(idea.get('lunar_score',0))>=9,
        abs(float(idea.get('technical_score',0)))>=18,
        mer['strength']>=25,
        float(idea.get('quantum_score',0))>0,
    ])
    return idea

# Preserve V5 functions then wrap them so commodities automatically gain Meridian analysis.
_V5_score_idea = score_idea
_V5_scan_stocks = scan_stocks
_V5_scan_commodities = scan_commodities


def score_idea(name, ticker, date, kind="stock", first_trade=None, astro_events=None, long_only=False, mode="before_market"):
    idea=_V5_score_idea(name,ticker,date,kind,first_trade,astro_events,long_only,mode)
    return _apply_meridian_to_idea(idea)


def scan_commodities(date, long_only=False, top_n=5, mode="before_market", include_no_trade=True):
    ideas=_V5_scan_commodities(date,long_only,0,mode,include_no_trade)
    ideas=[_apply_meridian_to_idea(i) for i in ideas]
    ideas.sort(key=lambda x:(x['status']=='NO TRADE',-x['combined_score']))
    ideas=ideas[:top_n] if top_n else ideas
    for i in ideas:
        try: i['execution_v8']=execution_timing_v8(date,first_trade=i.get('first_trade'))
        except Exception as e: i['execution_v8']={"score":0,"label":"ERROR","best_time":None,"windows":[],"reasons":[f"Timing calculation error: {e}"]}
    return ideas


def scan_stocks(date, universe='liquid', long_only=False, top_n=5, max_symbols=120, mode='before_market', include_no_trade=True):
    ideas=_V5_scan_stocks(date,universe,long_only,0,max_symbols,mode,include_no_trade)
    # old scan already contains the verified record on each idea
    ideas=[_apply_meridian_to_idea(i, i.get('first_trade_record')) for i in ideas]
    ideas.sort(key=lambda x:(x['status']=='NO TRADE',-x['combined_score']))
    ideas=ideas[:top_n] if top_n else ideas
    for i in ideas:
        try:
            i['execution_v8']=execution_timing_v8(date,first_trade=i.get('first_trade'),record=i.get('first_trade_record'))
        except Exception as e:
            i['execution_v8']={"score":0,"label":"ERROR","best_time":None,"windows":[],"reasons":[f"Timing calculation error: {e}"]}
    return ideas


def summary_dataframe(ideas):
    rows=[]
    for i in ideas:
        ex=i.get('execution_v8',{})
        rows.append({
            'Market':i['name'],'Ticker':i['ticker'],'Status':i['status'],'Bias':i['bias'],'Quality':i['quality'],
            'Bucholtz':i['primary_score'],'Lunar':i.get('lunar_score',0),'Meridian':i.get('meridian_score',0),
            'Meridian strength':i.get('meridian_strength',0),'Execution':ex.get('score',0),'Best ET':ex.get('best_time') or '-',
            'Technical':i['technical_score'],'Combined':i['combined_score'],'Families':i.get('independent_families_confirmed',0),
            'Data through':i['data_asof']
        })
    return pd.DataFrame(rows)


def format_idea(idea):
    if not idea: return 'No idea.'
    mer=idea.get('meridian',{}); ex=idea.get('execution_v8',{}); rs=idea['reason_sections']
    lines=[
        f"{idea['name']} ({idea['ticker']}) — {idea['status']} — {idea['bias']} — Quality {idea['quality']}",
        f"Date: {idea['date']} | data through {idea['data_asof']}",
        f"Scores: Bucholtz primary {idea['primary_score']} | Lunar {idea.get('lunar_score',0)} | Meridian {idea.get('meridian_score',0)} ({idea.get('meridian_direction','')}, strength {idea.get('meridian_strength',0)}) | Technical {idea['technical_score']} | Combined {idea['combined_score']}",
        f"Independent families confirmed: {idea.get('independent_families_confirmed',0)}/5",
        f"First-trade/natal reference: {idea.get('first_trade') or 'not available'} ({idea.get('first_trade_note','')})",
        f"Research levels: price {idea['price']:.2f} | confirmation/entry {idea['entry']:.2f} | invalidation {idea['stop']:.2f} | initial objective {idea['target']:.2f}",
        '', 'A. PRIMARY BUCHOLTZ TIMING',
    ]
    lines += [f"  • {x}" for x in rs.get('primary',[])[:6]] or ['  • No qualifying primary Bucholtz event.']
    lines += ['', 'B. LUNAR CONFLUENCE']
    lines += [f"  • {x}" for x in idea.get('lunar_detail',{}).get('reasons',[])[:8]] or ['  • No material lunar confluence.']
    lines += ['', 'C. APPLYING-ONLY EXECUTION TIMING']
    lines.append(f"  • Best ET window: {ex.get('best_time') or 'none'} | strength {ex.get('score',0)}/10 | {ex.get('label','')} | Meridian polarity {ex.get('polarity_label','UNCLASSIFIED')} ({ex.get('polarity_score',0):+})")
    lines += [f"  • {x}" for x in ex.get('reasons',[])[:6]]
    lines += ['', 'D. BILL MERIDIAN — NATAL STOCK SIGNATURE']
    lines += [f"  • {x}" for x in mer.get('natal',{}).get('reasons',[])[:10]] or ['  • No usable first-trade natal signature.']
    lines += ['', 'E. BILL MERIDIAN — APPLYING TRANSITS']
    lines += [f"  • {x}" for x in mer.get('transits',{}).get('reasons',[])[:10]] or ['  • No qualifying applying Meridian transit combination.']
    lines += ['', 'F. MERIDIAN STATION / ECLIPSE TIMING']
    lines += [f"  • {x}" for x in mer.get('timing',{}).get('reasons',[])[:8]] or ['  • No station/eclipse timing concentration detected.']
    lines += ['', 'G. SECONDARY / QUANTUM / TECHNICAL']
    lines += [f"  • {x}" for x in (rs.get('secondary',[])[:4]+rs.get('quantum',[])[:2]+rs.get('technical',[])[:6])]
    lines += ['', 'HOW TO USE',
              '  • Planetary evidence defines a timing/context window; it does not override price.',
              '  • Applying-only execution windows are preferred to separating aspects.',
              '  • If Meridian direction conflicts materially with the technical direction, the combined score is reduced.',
              '  • Entry requires the stated price confirmation; invalidation and objective remain risk-planning levels, not guarantees.']
    return '\n'.join(lines)

print('V8 Meridian + applying-execution extension loaded.')
```

## V8.3 — Decision layer + exact applying-aspect timing

This patch **preserves the V8.2 engine and report logic** and adds a clearer decision layer above it.

Changes in V8.3:
- separates **direction** from **timing intensity**;
- renames the intraday polarity field to **Execution-window aspect polarity** so transit-to-transit extensions cannot be mistaken for direct Bill Meridian directional evidence;
- demotes transit-to-transit contacts relative to transit-to-natal contacts;
- splits lunar output into **Lunar Timing Strength**, **Lunar Direction**, and **Lunar Character**;
- estimates the **exact/perfection time** of the strongest applying aspects;
- creates a practical **execution window** around the strongest 15-minute timing cluster;
- adds a model-derived **Directional Bias (-100 to +100)**, **Timing Strength (0–10)**, and **Evidence Confidence (A/B/C)** without deleting any existing scores or details.

The directional and timing composites below are transparent **model summaries**, not claims that Bucholtz or Meridian published these exact formulas.


```python
# ================================================================
# V8.3 DECISION + EXACT-TIMING PATCH (non-destructive)
# ================================================================
from datetime import datetime, timedelta, timezone
from zoneinfo import ZoneInfo
import math

V8_VERSION = "8.3"
_V82_scan_commodities = scan_commodities
_V82_scan_stocks = scan_stocks

# --- helpers ------------------------------------------------------
def _clamp(x, lo, hi):
    return max(lo, min(hi, x))

def _aspect_error_at(dt_utc, mover, target, relation, angle, chart=None):
    snap = _snapshot_at(dt_utc)
    if mover not in snap:
        return None
    m_lon = snap[mover]['lon']
    if relation == 'natal':
        if not chart or target not in chart:
            return None
        t_lon = chart[target]
    else:
        if target not in snap:
            return None
        t_lon = snap[target]['lon']
    return aspect_distance(m_lon, t_lon, angle)

def _estimate_exact_time(start_et, hit, chart, sample_minutes=15, max_hours=24):
    """Estimate when an applying aspect becomes exact.

    We first estimate the closure rate from the next sample, then refine around the
    projected time at one-minute resolution. This avoids thousands of ephemeris calls
    while still giving a practical intraday perfection estimate.
    """
    try:
        et = ZoneInfo('America/New_York')
        if start_et.tzinfo is None:
            start_et = start_et.replace(tzinfo=et)
        start_utc = start_et.astimezone(timezone.utc)
        angle = APPLYING_ASPECTS[hit['aspect']]['angle']
        e0 = _aspect_error_at(start_utc, hit['mover'], hit['target'], hit['relation'], angle, chart)
        e1 = _aspect_error_at(start_utc + timedelta(minutes=sample_minutes), hit['mover'], hit['target'], hit['relation'], angle, chart)
        if e0 is None or e1 is None or e1 >= e0:
            return None
        rate = (e0 - e1) / sample_minutes
        if rate <= 1e-7:
            return None
        eta_min = e0 / rate
        eta_min = _clamp(eta_min, 0, max_hours * 60)
        guess = start_utc + timedelta(minutes=eta_min)
        best_dt, best_err = guess, float('inf')
        # refine ±30 minutes around the linear estimate
        for m in range(-30, 31):
            dt = guess + timedelta(minutes=m)
            err = _aspect_error_at(dt, hit['mover'], hit['target'], hit['relation'], angle, chart)
            if err is not None and err < best_err:
                best_dt, best_err = dt, err
        # second refinement in 10-second increments ±90 seconds
        fine_dt, fine_err = best_dt, best_err
        for sec in range(-90, 91, 10):
            dt = best_dt + timedelta(seconds=sec)
            err = _aspect_error_at(dt, hit['mover'], hit['target'], hit['relation'], angle, chart)
            if err is not None and err < fine_err:
                fine_dt, fine_err = dt, err
        exact_et = fine_dt.astimezone(et)
        minutes_to = (fine_dt - start_utc).total_seconds() / 60.0
        return {
            'exact_time_et': exact_et.strftime('%H:%M'),
            'exact_datetime_et': exact_et.isoformat(),
            'estimated_minutes_to_exact': round(minutes_to, 1),
            'estimated_exact_orb': round(float(fine_err), 4),
        }
    except Exception:
        return None

def _execution_window(rows, best_strength, tolerance=0.6, step_minutes=15):
    """Return a contiguous near-peak execution window around the best bucket."""
    if not rows:
        return None
    et = ZoneInfo('America/New_York')
    near = [r for r in rows if float(r.get('strength', 0)) >= best_strength - tolerance]
    if not near:
        return None
    times = sorted(datetime.strptime(r['time_et'], '%H:%M') for r in near)
    # choose the cluster containing the best-strength time; otherwise earliest strongest cluster
    best_rows = [r for r in rows if float(r.get('strength',0)) == best_strength]
    anchor = datetime.strptime(best_rows[0]['time_et'], '%H:%M') if best_rows else times[0]
    clusters=[]; cur=[times[0]]
    for t in times[1:]:
        if (t-cur[-1]).total_seconds()/60 <= step_minutes + 1:
            cur.append(t)
        else:
            clusters.append(cur); cur=[t]
    clusters.append(cur)
    cluster=min(clusters, key=lambda c: 0 if c[0] <= anchor <= c[-1] else min(abs((anchor-c[0]).total_seconds()),abs((anchor-c[-1]).total_seconds())))
    start=cluster[0]
    end=cluster[-1] + timedelta(minutes=step_minutes)
    return f"{start.strftime('%H:%M')}–{end.strftime('%H:%M')} ET"

# --- audited execution timing ------------------------------------
def execution_timing_v8(date, first_trade=None, record=None, step_minutes=15):
    """V8.3 applying-only timing.

    Transit->natal contacts retain priority. Transit->transit contacts remain useful
    timing context, but are deliberately demoted so that a general sky aspect cannot
    overpower stock/commodity-specific natal evidence.
    """
    chart=_chart_node(first_trade,record)
    et=ZoneInfo('America/New_York')
    d=pd.Timestamp(date).date()
    start=datetime(d.year,d.month,d.day,9,30,tzinfo=et)
    end=datetime(d.year,d.month,d.day,16,0,tzinfo=et)
    rows=[]; dt=start

    while dt<=end:
        utc=dt.astimezone(timezone.utc)
        now=_snapshot_at(utc); nxt=_snapshot_at(utc+timedelta(minutes=step_minutes))
        hits=[]

        def add_hit(family,mover,target,relation,a,later_err,family_mult):
            if later_err >= a['current_orb'] - 1e-4:
                return
            pol=meridian_execution_polarity(mover,a['label'],target,relation)
            strength=_execution_hit_weight(a,family_mult,pol['confidence'])
            sign=1 if pol['direction']=='POSITIVE' else -1 if pol['direction']=='NEGATIVE' else 0
            # Directional authority is reduced for transit->transit extensions.
            directional_multiplier = 1.0 if relation=='natal' else 0.45
            directional=strength*sign*pol['confidence']*directional_multiplier
            hits.append({
                'family':family,'mover':mover,'target':target,'relation':relation,
                'aspect':a['label'],'angle':a['angle'],'orb':a['current_orb'],'max_orb':a['max_orb'],
                'strength':strength,'directional':directional,'polarity':pol,
            })

        # 1) Moon -> natal: highest-priority intraday timing family
        if chart:
            for np_ in [x for x in ['Sun','Moon','Mercury','Venus','Mars','Jupiter','Saturn','Uranus','Neptune','Pluto','Ascendant','MidHeaven','Node'] if x in chart]:
                a=_best_aspect(now['Moon']['lon'],chart[np_])
                if a:
                    later=aspect_distance(nxt['Moon']['lon'],chart[np_],a['angle'])
                    add_hit('moon_natal','Moon',np_,'natal',a,later,1.45)

        # 2) Moon -> transiting planets: useful intraday context, but below Moon->natal
        for p in ['Sun','Mercury','Venus','Mars','Jupiter','Saturn','Uranus','Neptune','Pluto']:
            a=_best_aspect(now['Moon']['lon'],now[p]['lon'])
            if a:
                later=aspect_distance(nxt['Moon']['lon'],nxt[p]['lon'],a['angle'])
                add_hit('moon_transit','Moon',p,'transit',a,later,0.70)

        # 3) Inner planets -> outer planets: broader sky background
        for ip in ['Mercury','Venus','Mars']:
            for op in ['Jupiter','Saturn','Uranus','Neptune','Pluto']:
                a=_best_aspect(now[ip]['lon'],now[op]['lon'])
                if a:
                    later=aspect_distance(nxt[ip]['lon'],nxt[op]['lon'],a['angle'])
                    add_hit('inner_outer',ip,op,'transit',a,later,0.55)

        if hits:
            family_caps={'moon_natal':5.2,'moon_transit':2.4,'inner_outer':1.8}
            fam_strength={}
            for fam,cap in family_caps.items():
                vals=sorted([h['strength'] for h in hits if h['family']==fam],reverse=True)
                val=(vals[0] if vals else 0.0) + (0.30*vals[1] if len(vals)>1 else 0.0)
                fam_strength[fam]=min(cap,val*3.0)
            strength=min(10.0,sum(fam_strength.values()))

            directional=sum(h['directional'] for h in hits)
            denom=max(1e-9,sum(abs(h['directional']) for h in hits))
            net=0.0 if denom==0 else 10.0*directional/denom
            if net>=2.0: pol_label='POSITIVE'
            elif net<=-2.0: pol_label='NEGATIVE'
            else: pol_label='MIXED / NEUTRAL'

            hits_sorted=sorted(hits,key=lambda h:(0 if h['relation']=='natal' else 1,-h['strength'],h['orb']))
            reasons=[]
            for h in hits_sorted[:7]:
                p=h['polarity']
                reasons.append(
                    f"{p['direction']} [{p['source']}] — {h['mover']} {h['aspect']} "
                    f"{'natal ' if h['relation']=='natal' else 'transiting '}{h['target']} "
                    f"(orb {h['orb']:.2f}° applying; limit {h['max_orb']:.1f}°). {p['note']}"
                )
            rows.append({
                'time_et':dt.strftime('%H:%M'),'dt_et':dt,
                'strength':round(strength,1),'polarity_score':round(net,1),'polarity_label':pol_label,
                'positive_hits':sum(1 for h in hits if h['polarity']['direction']=='POSITIVE'),
                'negative_hits':sum(1 for h in hits if h['polarity']['direction']=='NEGATIVE'),
                'unclassified_hits':sum(1 for h in hits if h['polarity']['direction']=='UNCLASSIFIED'),
                'family_strength':{k:round(v,2) for k,v in fam_strength.items()},
                'reasons':reasons,'raw_hits':hits_sorted,
            })
        dt += timedelta(minutes=step_minutes)

    if not rows:
        return {'score':0.0,'label':'NO QUALIFYING WINDOW','best_time':None,'window':None,'windows':[],
                'reasons':[],'polarity_score':0.0,'polarity_label':'UNCLASSIFIED','exact_aspects':[]}

    rows=sorted(rows,key=lambda r:(-r['strength'],-abs(r['polarity_score']),r['time_et']))
    best=rows[0]
    label='STRONG' if best['strength']>=7 else 'MODERATE' if best['strength']>=4 else 'LIGHT'
    window=_execution_window(rows,best['strength'],tolerance=0.6,step_minutes=step_minutes)

    # Calculate perfection times only for the strongest contacts in the winning bucket.
    exact=[]
    for h in best.get('raw_hits',[])[:5]:
        ex=_estimate_exact_time(best['dt_et'],h,chart,step_minutes)
        if ex:
            exact.append({**{k:h[k] for k in ['mover','target','relation','aspect','orb','max_orb']},
                          'polarity':h['polarity']['direction'],'source':h['polarity']['source'],**ex})

    return {
        'score':best['strength'],'label':label,'best_time':best['time_et'],'window':window,
        'windows':[{k:v for k,v in r.items() if k not in ('dt_et','raw_hits')} for r in rows[:8]],
        'reasons':best['reasons'],'polarity_score':best['polarity_score'],
        'polarity_label':best['polarity_label'],'positive_hits':best['positive_hits'],
        'negative_hits':best['negative_hits'],'unclassified_hits':best['unclassified_hits'],
        'family_strength':best['family_strength'],'exact_aspects':exact,
    }

# --- transparent decision summaries ------------------------------
def _lunar_summary(idea):
    lunar=float(idea.get('lunar_score',0) or 0)
    reasons=idea.get('lunar_detail',{}).get('reasons',[]) or []
    txt=' '.join(str(x).lower() for x in reasons)
    pos=sum(txt.count(k) for k in ['supportive','trine','sextile'])
    neg=sum(txt.count(k) for k in ['tension','volatility','reversal','opposition','square'])
    if pos>neg: direction='POSITIVE'
    elif neg>pos: direction='NEGATIVE'
    else: direction='MIXED / NEUTRAL'
    env=str(idea.get('lunar_environment','') or '')
    if 'reversal' in env.lower(): character='Reversal-sensitive / volatile'
    elif 'support' in env.lower(): character='Supportive'
    elif 'mixed' in env.lower() or 'balanced' in env.lower(): character='Mixed / balanced'
    else: character=env or 'Neutral / contextual'
    return {'strength':round(lunar,1),'direction':direction,'character':character}

def _directional_bias(idea):
    """Model-derived direction score. Technical evidence is deliberately dominant."""
    tech=float(idea.get('technical_score',0) or 0)
    mer=float(idea.get('meridian_score',0) or 0)
    tech_scaled=_clamp((tech/45.0)*100.0,-100,100)
    mer_scaled=_clamp(mer*2.0,-100,100)
    score=_clamp(0.70*tech_scaled + 0.30*mer_scaled,-100,100)
    if score>=60: label='STRONGLY BULLISH'
    elif score>=25: label='BULLISH'
    elif score<=-60: label='STRONGLY BEARISH'
    elif score<=-25: label='BEARISH'
    else: label='NEUTRAL / CONFLICTED'
    return {'score':round(score,1),'label':label,'technical_component':round(tech_scaled,1),'meridian_component':round(mer_scaled,1)}

def _timing_strength(idea):
    """Model timing composite; not a published Bucholtz/Meridian formula."""
    ex=float(idea.get('execution_v8',{}).get('score',0) or 0)
    primary=_clamp(float(idea.get('primary_score',0) or 0)/25.0*10.0,0,10)
    lunar=_clamp(float(idea.get('lunar_score',0) or 0)/12.0*10.0,0,10)
    return round(_clamp(0.40*ex + 0.35*primary + 0.25*lunar,0,10),1)

def _evidence_confidence(idea):
    q=str(idea.get('natal_data_quality','') or idea.get('first_trade_note','')).upper()
    fam=int(idea.get('independent_families_confirmed',0) or 0)
    if 'VERIFIED_TIME' in q and fam>=3:
        return 'A'
    if ('VERIFIED_DATE' in q or 'PUBLISHED' in q or 'VERIFIED' in q) and fam>=3:
        return 'B'
    return 'C'

def _decision_layer(idea):
    ex=idea.get('execution_v8',{}) or {}
    lunar=_lunar_summary(idea)
    direct_mer=float(idea.get('meridian_score',0) or 0)
    mer_dir='POSITIVE' if direct_mer>=5 else 'NEGATIVE' if direct_mer<=-5 else 'MIXED / NEUTRAL'
    return {
        'directional_bias':_directional_bias(idea),
        'timing_strength':_timing_strength(idea),
        'execution_window':ex.get('window') or (f"{ex.get('best_time')} ET" if ex.get('best_time') else None),
        'best_time_et':ex.get('best_time'),
        'execution_aspect_polarity':ex.get('polarity_label','UNCLASSIFIED'),
        'execution_aspect_polarity_score':ex.get('polarity_score',0),
        'direct_meridian_direction':mer_dir,
        'direct_meridian_score':direct_mer,
        'lunar':lunar,
        'evidence_confidence':_evidence_confidence(idea),
        'exact_aspects':ex.get('exact_aspects',[]),
    }

def _enrich_v83(idea):
    if idea:
        idea['decision_v83']=_decision_layer(idea)
    return idea

# Re-wrap scans without removing any original behaviour or scores.
def scan_commodities(date, long_only=False, top_n=5, mode='before_market', include_no_trade=True):
    ideas=_V82_scan_commodities(date,long_only,top_n,mode,include_no_trade)
    return [_enrich_v83(i) for i in ideas]

def scan_stocks(date, universe='liquid', long_only=False, top_n=5, max_symbols=120, mode='before_market', include_no_trade=True):
    ideas=_V82_scan_stocks(date,universe,long_only,top_n,max_symbols,mode,include_no_trade)
    return [_enrich_v83(i) for i in ideas]

# Clearer interactive summary. Existing underlying scores remain visible.
def summary_dataframe(ideas):
    rows=[]
    for i in ideas:
        ex=i.get('execution_v8',{}); d=i.get('decision_v83') or _decision_layer(i); lunar=d['lunar']
        rows.append({
            'Market':i['name'],'Ticker':i['ticker'],'Status':i['status'],'Bias':i['bias'],'Quality':i['quality'],
            'Directional Bias':f"{d['directional_bias']['score']:+.1f} {d['directional_bias']['label']}",
            'Timing':d['timing_strength'],'Exec window':d['execution_window'] or '-',
            'Lunar strength':lunar['strength'],'Lunar direction':lunar['direction'],
            'Direct Meridian':f"{d['direct_meridian_score']:+.1f} {d['direct_meridian_direction']}",
            'Exec aspect polarity':f"{d['execution_aspect_polarity_score']:+.1f} {d['execution_aspect_polarity']}",
            'Confidence':d['evidence_confidence'],'Technical':i['technical_score'],'Combined (legacy)':i['combined_score'],
            'Data through':i['data_asof']
        })
    return pd.DataFrame(rows)

def format_idea(idea):
    if not idea: return 'No idea.'
    d=idea.get('decision_v83') or _decision_layer(idea); ex=idea.get('execution_v8',{}); rs=idea['reason_sections']; mer=idea.get('meridian',{})
    b=d['directional_bias']; l=d['lunar']
    lines=[
        f"{idea['name']} ({idea['ticker']}) — {idea['status']} — {idea['bias']} — Quality {idea['quality']}",
        f"Date: {idea['date']} | data through {idea['data_asof']}",
        '', 'V8.3 DECISION LAYER',
        f"  • Directional Bias: {b['score']:+.1f}/100 — {b['label']} (Technical component {b['technical_component']:+.1f}; direct Meridian component {b['meridian_component']:+.1f})",
        f"  • Timing Strength: {d['timing_strength']}/10",
        f"  • Execution Window: {d['execution_window'] or 'none'} | peak {d['best_time_et'] or 'none'} ET",
        f"  • Lunar: strength {l['strength']}/10 | direction {l['direction']} | character {l['character']}",
        f"  • Direct Bill Meridian direction: {d['direct_meridian_direction']} ({d['direct_meridian_score']:+.1f})",
        f"  • Execution-window aspect polarity: {d['execution_aspect_polarity']} ({d['execution_aspect_polarity_score']:+.1f})",
        f"  • Evidence Confidence: {d['evidence_confidence']}",
        '', 'EXACT / PERFECTION TIMES (estimated)',
    ]
    exact=d.get('exact_aspects',[])
    lines += [f"  • {x['mover']} {x['aspect']} {'natal ' if x['relation']=='natal' else 'transiting '}{x['target']} → approx. {x['exact_time_et']} ET (estimated minimum orb {x['estimated_exact_orb']:.3f}°; {x['polarity']})" for x in exact] or ['  • No qualifying exact-time estimate.']
    lines += [
        '', 'LEGACY / SOURCE-LAYER SCORES (preserved)',
        f"  • Bucholtz primary {idea['primary_score']} | Lunar {idea.get('lunar_score',0)} | Meridian {idea.get('meridian_score',0)} (strength {idea.get('meridian_strength',0)}) | Technical {idea['technical_score']} | Combined {idea['combined_score']}",
        f"  • First-trade/natal reference: {idea.get('first_trade') or 'not available'} ({idea.get('first_trade_note','')})",
        f"  • Research levels: price {idea['price']:.2f} | confirmation/entry {idea['entry']:.2f} | invalidation {idea['stop']:.2f} | initial objective {idea['target']:.2f}",
        '', 'A. PRIMARY BUCHOLTZ TIMING']
    lines += [f"  • {x}" for x in rs.get('primary',[])[:6]] or ['  • No qualifying primary Bucholtz event.']
    lines += ['', 'B. LUNAR CONFLUENCE']
    lines += [f"  • {x}" for x in idea.get('lunar_detail',{}).get('reasons',[])[:8]] or ['  • No material lunar confluence.']
    lines += ['', 'C. APPLYING-ONLY EXECUTION TIMING']
    lines.append(f"  • Peak ET: {ex.get('best_time') or 'none'} | window {ex.get('window') or 'none'} | strength {ex.get('score',0)}/10 | execution-window aspect polarity {ex.get('polarity_label','UNCLASSIFIED')} ({ex.get('polarity_score',0):+})")
    lines += [f"  • {x}" for x in ex.get('reasons',[])[:6]]
    lines += ['', 'D. BILL MERIDIAN — NATAL STOCK SIGNATURE']
    lines += [f"  • {x}" for x in mer.get('natal',{}).get('reasons',[])[:10]] or ['  • No usable first-trade natal signature.']
    lines += ['', 'E. BILL MERIDIAN — APPLYING TRANSITS']
    lines += [f"  • {x}" for x in mer.get('transits',{}).get('reasons',[])[:10]] or ['  • No qualifying applying Meridian transit combination.']
    lines += ['', 'F. SECONDARY / QUANTUM / TECHNICAL']
    lines += [f"  • {x}" for x in (rs.get('secondary',[])[:4]+rs.get('quantum',[])[:2]+rs.get('technical',[])[:6])]
    lines += ['', 'HOW TO USE',
              '  • Direction is led by price/technical confirmation and direct Meridian evidence; timing is a separate question.',
              '  • Transit-to-natal contacts outrank transit-to-transit extensions.',
              '  • Lunar strength means importance/intensity, not automatically bullishness.',
              '  • Entry still requires the stated price confirmation; invalidation and objective are risk-planning levels, not guarantees.']
    return '\n'.join(lines)

print('V8.3 decision + exact-timing patch loaded.')
```

```python
# V8.2 persistent scan-result store
# Do not reset this when report/export cells are re-run.
if "LAST_SCAN_RESULTS" not in globals():
    LAST_SCAN_RESULTS = {"date": None, "mode": None, "commodities": [], "stocks": []}


import ipywidgets as widgets
from IPython.display import display, clear_output
from datetime import date

date_widget = widgets.DatePicker(description="Date", value=date.today())
mode_widget = widgets.Dropdown(
    options=[("Before Market — build watchlist","before_market"),
             ("During Market — live/provisional","during_market"),
             ("After Market — confirm/reject","after_market")],
    value="before_market",
    description="Mode"
)
scan_widget = widgets.Dropdown(
    options=[("Both","both"),("Stocks","stocks"),("Commodities","commodities")],
    value="both",
    description="Scan"
)
universe_widget = widgets.Dropdown(
    options=[("Liquid US stocks (fast)","liquid"),("S&P 500 (slower)","sp500")],
    value="liquid",
    description="Stocks"
)
long_widget = widgets.Checkbox(value=False, description="Long ideas only")
top_widget = widgets.IntSlider(value=8, min=3, max=15, step=1, description="Show")
run_button = widgets.Button(description="RUN V8.2 SCAN", button_style="success")
out = widgets.Output()

def show_detail(i, rank):
    print("\n" + "="*105)
    print(f"#{rank}  {i['name']} ({i['ticker']}) — {i['status']} — {i['bias']} — Quality {i['quality']}")
    print("="*105)
    print(f"Date: {i['date']} | Price data through: {i['data_asof']} | {'⚠ PROVISIONAL' if i['provisional'] else 'date-confirmed data'}")
    print(f"Primary {i['primary_score']} | Lunar {i.get('lunar_score',0)} ({i.get('lunar_environment','')}) | Secondary {i['secondary_score']} | Quantum {i['quantum_score']} | Technical {i['technical_score']} | Combined {i['combined_score']}")
    print(f"Natal data quality: {i.get('natal_data_quality', 'PUBLISHED' if i.get('kind')=='commodity' else 'UNKNOWN')}")
    print(f"First-trade/natal reference: {i.get('first_trade') or 'not available'} — {i.get('first_trade_note','')}")
    print(f"Price {i['price']:.2f} | Confirmation/entry {i['entry']:.2f} | Invalidation {i['stop']:.2f} | Initial objective {i['target']:.2f}")
    rs=i["reason_sections"]

    print("\nA. PRIMARY BUCHOLTZ TIMING")
    for x in rs["primary"] or ["No qualifying primary timing event."]:
        print(" •",x)

    print("\nB. LUNAR CONFLUENCE")
    lunar=i.get("lunar_detail",{})
    print(f" • Score {i.get('lunar_score',0)}/20 — {i.get('lunar_label','')} — {i.get('lunar_environment','')}")
    for x in lunar.get("reasons",[])[:8] or ["No material lunar confluence."]:
        print(" •",x)

    print("\nC. SECONDARY ASTRONOMICAL CONFIRMATION")
    for x in rs["secondary"] or ["No secondary timing confluence."]:
        print(" •",x)

    print("\nD. QUANTUM / PRICE GEOMETRY")
    for x in rs["quantum"] or ["No nearby Quantum Line confluence."]:
        print(" •",x)

    print("\nE. TECHNICAL CONFIRMATION")
    for x in rs["technical"] or ["No directional technical confirmation."]:
        print(" •",x)

    print("\nDECISION")
    if i["status"]=="NO TRADE":
        print(" • Rejected: the setup did not satisfy the strong-primary + technical-confirmation gate.")
    elif i["mode"]=="before_market":
        print(" • This is a WATCHLIST candidate. Price must confirm during/after the requested session.")
    elif i["status"]=="PENDING CONFIRMATION":
        print(" • Requested date has not yet produced market data, so confirmation is pending.")
    else:
        print(" • Requested-date price data is available and the grouped timing + technical gate is satisfied.")

def _validated_market_mode(date_iso, requested_mode):
    """Enforce honest session labels in New York time."""
    req=pd.Timestamp(date_iso).date()
    now_et=datetime.now(ZoneInfo("America/New_York"))
    today=now_et.date()

    if req < today:
        return "after_market", "Historical date: complete-session mode enforced."
    if req > today:
        return "before_market", "Future date: watchlist mode enforced."

    mins=now_et.hour*60+now_et.minute
    if mins < 9*60+30:
        actual="before_market"
    elif mins < 16*60:
        actual="during_market"
    else:
        actual="after_market"

    if requested_mode != actual:
        return actual, f"Requested '{requested_mode}' changed to '{actual}' because current New York time is {now_et.strftime('%H:%M ET')}."
    return actual, f"Session mode verified at {now_et.strftime('%H:%M ET')}."

def run_scan(_):
    with out:
        clear_output()
        if date_widget.value is None:
            print("Choose a date.")
            return
        d=date_widget.value.isoformat()
        requested_mode=mode_widget.value
        effective_mode, mode_note=_validated_market_mode(d,requested_mode)
        LAST_SCAN_RESULTS["date"] = d
        LAST_SCAN_RESULTS["requested_mode"] = requested_mode
        LAST_SCAN_RESULTS["mode"] = effective_mode
        LAST_SCAN_RESULTS["mode_note"] = mode_note
        LAST_SCAN_RESULTS["commodities"] = []
        LAST_SCAN_RESULTS["stocks"] = []
        print(f"Bucholtz/Meridian/McWhirter scan: {d} | effective mode: {effective_mode}")
        print(mode_note)
        print("Scoring is grouped by independent evidence families to avoid over-counting.\n")

        if scan_widget.value in ("commodities","both"):
            ideas=scan_commodities(d,long_widget.value,top_widget.value,effective_mode,True)
            LAST_SCAN_RESULTS["commodities"] = ideas
            print("\nCOMMODITIES — SUMMARY")
            display(summary_dataframe(ideas))
            for n,i in enumerate(ideas,1):
                show_detail(i,n)

        if scan_widget.value in ("stocks","both"):
            print("\n\nSTOCKS — scanning verified first-trade database...")
            try:
                ideas=scan_stocks(
                    d, universe_widget.value, long_widget.value, top_widget.value,
                    120 if universe_widget.value=="sp500" else 60,
                    effective_mode, True
                )
                LAST_SCAN_RESULTS["stocks"] = ideas
                print("\nSTOCKS — SUMMARY")
                if ideas:
                    display(summary_dataframe(ideas))
                    for n,i in enumerate(ideas,1):
                        show_detail(i,n)
                else:
                    print("No stock rows were returned. This is not treated as a successful stock scan.")
            except Exception as e:
                LAST_SCAN_RESULTS["stocks"] = []
                print("STOCK SCAN ERROR:", type(e).__name__, "-", e)
                print("The commodity results above are still valid, but do not download the combined report until this stock error is resolved.")
                raise

run_button.on_click(run_scan)
display(widgets.VBox([
    widgets.HTML("<h3>Bucholtz + Meridian Trade Idea Agent — V8.2</h3>"),
    date_widget, mode_widget, scan_widget, universe_widget,
    long_widget, top_widget, run_button, out
]))
```

## How to read the Lunar score

The raw **Lunar Confluence score** is preserved for audit, while the V8.4 decision display is capped at **10.0/10**. It remains separate from the Primary Bucholtz score.

- **14–20 — STRONG:** several independent lunar factors align.
- **9–13.9 — MODERATE:** meaningful lunar timing support or tension.
- **5–8.9 — LIGHT:** some lunar relevance, but it should not drive the trade.
- **0–4.9 — QUIET:** little lunar timing significance.

The environment label describes the *character* of the Moon's relationships:
**Supportive / flowing**, **Volatility / tension**, **Tension / reversal-sensitive**, or **Mixed / balanced**.

This is not a long/short signal by itself. The **Technical score still supplies trade direction**.

```python

# ------------------------------------------------------------------
# PRINT / DOWNLOAD REPORT EXPORT
# ------------------------------------------------------------------
from pathlib import Path
from datetime import datetime
import html as _html
import pandas as pd
from reportlab.lib import colors
from reportlab.lib.enums import TA_LEFT
from reportlab.lib.pagesizes import A4, landscape
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import mm
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, PageBreak, KeepTogether
)
from IPython.display import display, HTML

try:
    from google.colab import files as colab_files
except Exception:
    colab_files = None

# V8.2: preserve an already-completed scan.
if "LAST_SCAN_RESULTS" not in globals():
    LAST_SCAN_RESULTS = {"date": None, "mode": None, "commodities": [], "stocks": []}

def _safe_filename(s):
    return re.sub(r"[^A-Za-z0-9_.-]+", "_", str(s)).strip("_")

def _report_title():
    d = LAST_SCAN_RESULTS.get("date") or "scan"
    mode = LAST_SCAN_RESULTS.get("mode") or "research"
    return f"Bucholtz_Meridian_V8_Report_{d}_{mode}"

def _status_order(status):
    order = {
        "CONFIRMED SETUP": 0,
        "WATCH DATE": 1,
        "WATCH / SETUP": 1,
        "PENDING CONFIRMATION": 2,
        "NO TRADE": 3,
    }
    return order.get(status, 9)

def _all_report_ideas():
    ideas = list(LAST_SCAN_RESULTS.get("commodities", [])) + list(LAST_SCAN_RESULTS.get("stocks", []))
    return sorted(ideas, key=lambda x: (_status_order(x.get("status","")), -float(x.get("combined_score",0))))

def _summary_rows(ideas):
    rows = [["Market","Ticker","Type","Status","Bias","Quality","Bucholtz","Lunar","Meridian","Exec","Technical","Combined","Data through"]]
    for i in ideas:
        rows.append([
            i["name"], i["ticker"], i["kind"].title(), i["status"], i["bias"], i["quality"],
            i['primary_score'], i.get('lunar_score',0), i.get('meridian_score',0), i.get('execution_v8',{}).get('score',0), i['technical_score'], i['combined_score'], i['data_asof']
        ])
    return rows

def make_html_report():
    ideas = _all_report_ideas()
    if not ideas:
        raise ValueError("No saved scan results are available. Run the main V8.2 scan once, then click the download button without re-running/resetting the notebook.")

    title = _report_title()
    path = Path("/content" if Path("/content").exists() else "/mnt/data") / f"{title}.html"

    def esc(x): return _html.escape(str(x))
    summary = _summary_rows(ideas)
    th = "".join(f"<th>{esc(x)}</th>" for x in summary[0])
    trs = []
    for row in summary[1:]:
        trs.append("<tr>" + "".join(f"<td>{esc(x)}</td>" for x in row) + "</tr>")

    detail_sections = []
    for rank, i in enumerate(ideas, 1):
        rs = i["reason_sections"]
        def lis(items, fallback):
            xs = items or [fallback]
            return "".join(f"<li>{esc(x)}</li>" for x in xs)
        detail_sections.append(f"""
        <section class="idea">
          <h2>#{rank} {esc(i['name'])} ({esc(i['ticker'])}) - {esc(i['status'])} - {esc(i['bias'])}</h2>
          <p><b>Date:</b> {esc(i['date'])} &nbsp; | &nbsp; <b>Data through:</b> {esc(i['data_asof'])}
             &nbsp; | &nbsp; <b>Quality:</b> {esc(i['quality'])}</p>
          <p><b>Scores:</b> Bucholtz {esc(i['primary_score'])} | Lunar {esc(i.get('lunar_score',0))} ({esc(i.get('lunar_environment',''))}) | Meridian {esc(i.get('meridian_score',0))} (strength {esc(i.get('meridian_strength',0))}) | Execution {esc(i.get('execution_v8',{}).get('score',0))}/10 | Secondary {esc(i['secondary_score'])}
             | Quantum {esc(i['quantum_score'])} | Technical {esc(i['technical_score'])}
             | Combined {esc(i['combined_score'])}</p>
          <p><b>First-trade/natal reference:</b> {esc(i.get('first_trade') or 'Not available')}
             ({esc(i.get('first_trade_note',''))})</p>
          <p><b>Research levels:</b> Price {i['price']:.2f} | Confirmation/entry {i['entry']:.2f}
             | Invalidation {i['stop']:.2f} | Initial objective {i['target']:.2f}</p>

          <h3>A. Primary Bucholtz timing</h3><ul>{lis(rs['primary'],'No qualifying primary timing event.')}</ul>
          <h3>B. Lunar confluence</h3><ul>{lis(i.get('lunar_detail',{}).get('reasons',[])[:8],'No material lunar confluence.')}</ul>
          <h3>C. Applying-only execution timing</h3><p>Best ET: {esc(i.get('execution_v8',{}).get('best_time') or 'none')} | Strength {esc(i.get('execution_v8',{}).get('score',0))}/10 | Meridian polarity {esc(i.get('execution_v8',{}).get('polarity_label','UNCLASSIFIED'))} ({esc(i.get('execution_v8',{}).get('polarity_score',0)):+})</p><ul>{lis(i.get('execution_v8',{}).get('reasons',[])[:6],'No qualifying applying timing window.')}</ul>
          <h3>D. Bill Meridian natal signature</h3><ul>{lis(i.get('meridian',{}).get('natal',{}).get('reasons',[])[:8],'No usable Meridian natal signature.')}</ul>
          <h3>E. Bill Meridian applying transits</h3><ul>{lis(i.get('meridian',{}).get('transits',{}).get('reasons',[])[:8],'No qualifying Meridian transit combination.')}</ul>
          <h3>F. Secondary astronomical confirmation</h3><ul>{lis(rs['secondary'],'No secondary timing confluence.')}</ul>
          <h3>D. Quantum / price geometry</h3><ul>{lis(rs['quantum'],'No nearby Quantum Line confluence.')}</ul>
          <h3>E. Technical confirmation</h3><ul>{lis(rs['technical'],'No directional technical confirmation.')}</ul>

          <h3>Decision</h3>
          <p>{esc(
              'Rejected: strong-primary + technical-confirmation gate was not satisfied.'
              if i['status']=='NO TRADE'
              else 'Watchlist candidate: price must confirm during/after the requested session.'
              if i['mode']=='before_market'
              else 'Confirmation pending because price data for the requested date is not yet available.'
              if i['status']=='PENDING CONFIRMATION'
              else 'Grouped timing and technical confirmation gate is satisfied for the requested date.'
          )}</p>
        </section>
        """)

    doc = f"""<!doctype html>
<html><head><meta charset="utf-8"><title>{esc(title)}</title>
<style>
@page {{ size: A4; margin: 15mm; }}
body {{ font-family: Arial, Helvetica, sans-serif; color:#222; line-height:1.35; font-size:10.5pt; }}
h1 {{ font-size:20pt; margin-bottom:4px; }}
h2 {{ font-size:14pt; margin-top:22px; border-bottom:1px solid #aaa; padding-bottom:4px; }}
h3 {{ font-size:11pt; margin-bottom:2px; }}
.note {{ background:#f3f3f3; padding:8px 10px; border-left:3px solid #777; }}
table {{ width:100%; border-collapse:collapse; margin:12px 0 18px; font-size:8.5pt; }}
th,td {{ border:1px solid #aaa; padding:4px; text-align:left; vertical-align:top; }}
th {{ background:#eee; }}
.idea {{ page-break-inside:avoid; }}
.small {{ font-size:8.5pt; color:#555; }}
@media print {{
  .no-print {{ display:none; }}
  a {{ color:#000; text-decoration:none; }}
}}
</style></head>
<body>
<h1>Bucholtz + Bill Meridian V8.1 Audited Trade Idea Report</h1>
<p><b>Requested date:</b> {esc(LAST_SCAN_RESULTS['date'])}<br>
<b>Mode:</b> {esc(LAST_SCAN_RESULTS['mode'])}<br>
<b>Generated:</b> {esc(datetime.now().strftime('%Y-%m-%d %H:%M'))}</p>

<div class="note">
<b>Research framework:</b> Primary Bucholtz timing must align with technical confirmation.
Related astronomical observations are grouped and capped to avoid score inflation.
Financial astrology is not scientifically established as a forecasting method. V8.1 audits applying aspects using actual orb contraction; polarity labels distinguish direct Meridian rules from clearly-labelled model inferences.
</div>

<h2>Executive Summary</h2>
<table><thead><tr>{th}</tr></thead><tbody>{''.join(trs)}</tbody></table>

{''.join(detail_sections)}

<p class="small">This report is for research and trade-planning purposes. Recheck market data and risk parameters before acting.</p>
</body></html>"""
    path.write_text(doc, encoding="utf-8")
    return str(path)

def make_pdf_report():
    ideas = _all_report_ideas()
    if not ideas:
        raise ValueError("No saved scan results are available. Run the main V8.2 scan once, then click the download button without re-running/resetting the notebook.")

    title = _report_title()
    path = Path("/content" if Path("/content").exists() else "/mnt/data") / f"{title}.pdf"

    styles = getSampleStyleSheet()
    styles.add(ParagraphStyle(name="SmallBody", parent=styles["BodyText"], fontSize=8.2, leading=10.2, spaceAfter=3))
    styles.add(ParagraphStyle(name="Tiny", parent=styles["BodyText"], fontSize=7.1, leading=8.6))
    styles.add(ParagraphStyle(name="Sec", parent=styles["Heading3"], fontSize=9.5, leading=11, spaceBefore=5, spaceAfter=2))
    styles.add(ParagraphStyle(name="IdeaHead", parent=styles["Heading2"], fontSize=12, leading=14, spaceBefore=6, spaceAfter=5))

    doc = SimpleDocTemplate(
        str(path), pagesize=A4,
        rightMargin=14*mm, leftMargin=14*mm, topMargin=14*mm, bottomMargin=14*mm
    )
    story = []
    story.append(Paragraph("Bucholtz + Bill Meridian V8.1 Audited Trade Idea Report", styles["Title"]))
    story.append(Paragraph(
        f"<b>Requested date:</b> {LAST_SCAN_RESULTS['date']}<br/>"
        f"<b>Mode:</b> {LAST_SCAN_RESULTS['mode']}<br/>"
        f"<b>Generated:</b> {datetime.now().strftime('%Y-%m-%d %H:%M')}",
        styles["BodyText"]
    ))
    story.append(Spacer(1, 5))
    story.append(Paragraph(
        "<b>Research framework:</b> Primary Bucholtz timing must align with technical confirmation. "
        "Related astronomical observations are grouped and capped to avoid score inflation. "
        "Financial astrology is not scientifically established as a forecasting method. V8.1 audits applying aspects using actual orb contraction; polarity labels distinguish direct Meridian rules from clearly-labelled model inferences.",
        styles["SmallBody"]
    ))
    story.append(Spacer(1, 7))
    story.append(Paragraph("Executive Summary", styles["Heading2"]))

    summary = _summary_rows(ideas)
    pdf_rows = []
    for r, row in enumerate(summary):
        pdf_rows.append([Paragraph(str(x), styles["Tiny"]) for x in row])
    col_widths = [20*mm,12*mm,12*mm,20*mm,11*mm,12*mm,11*mm,10*mm,11*mm,10*mm,11*mm,13*mm,17*mm]
    table = Table(pdf_rows, colWidths=col_widths, repeatRows=1)
    table.setStyle(TableStyle([
        ("BACKGROUND",(0,0),(-1,0),colors.lightgrey),
        ("GRID",(0,0),(-1,-1),0.35,colors.grey),
        ("VALIGN",(0,0),(-1,-1),"TOP"),
        ("LEFTPADDING",(0,0),(-1,-1),2),
        ("RIGHTPADDING",(0,0),(-1,-1),2),
        ("TOPPADDING",(0,0),(-1,-1),2),
        ("BOTTOMPADDING",(0,0),(-1,-1),2),
    ]))
    story.append(table)

    for rank, i in enumerate(ideas, 1):
        rs = i["reason_sections"]
        story.append(PageBreak())
        story.append(Paragraph(
            f"#{rank} {i['name']} ({i['ticker']}) - {i['status']} - {i['bias']} - Quality {i['quality']}",
            styles["IdeaHead"]
        ))
        story.append(Paragraph(
            f"<b>Date:</b> {i['date']} | <b>Data through:</b> {i['data_asof']}<br/>"
            f"<b>Scores:</b> Bucholtz {i['primary_score']} | Lunar {i.get('lunar_score',0)} ({i.get('lunar_environment','')}) | Meridian {i.get('meridian_score',0)} (strength {i.get('meridian_strength',0)}) | Execution {i.get('execution_v8',{}).get('score',0)}/10 | Secondary {i['secondary_score']} | "
            f"Quantum {i['quantum_score']} | Technical {i['technical_score']} | Combined {i['combined_score']}<br/>"
            f"<b>Natal data quality:</b> {i.get('natal_data_quality', 'PUBLISHED' if i.get('kind')=='commodity' else 'UNKNOWN')}<br/>"
            f"<b>First-trade/natal reference:</b> {i.get('first_trade') or 'Not available'} "
            f"({i.get('first_trade_note','')})<br/>"
            f"<b>Research levels:</b> Price {i['price']:.2f} | Confirmation/entry {i['entry']:.2f} | "
            f"Invalidation {i['stop']:.2f} | Initial objective {i['target']:.2f}",
            styles["SmallBody"]
        ))

        for heading, items, fallback in [
            ("A. Primary Bucholtz timing", rs["primary"], "No qualifying primary timing event."),
            ("B. Lunar confluence", i.get("lunar_detail",{}).get("reasons",[])[:8], "No material lunar confluence."),
            ("C. Applying-only execution timing", [f"Best ET: {i.get('execution_v8',{}).get('best_time') or 'none'} | Strength {i.get('execution_v8',{}).get('score',0)}/10 | Meridian polarity {i.get('execution_v8',{}).get('polarity_label','UNCLASSIFIED')} ({i.get('execution_v8',{}).get('polarity_score',0):+})"] + i.get('execution_v8',{}).get('reasons',[])[:7], "No qualifying applying timing window."),
            ("D. Bill Meridian natal signature", i.get('meridian',{}).get('natal',{}).get('reasons',[])[:8], "No usable Meridian natal signature."),
            ("E. Bill Meridian applying transits", i.get('meridian',{}).get('transits',{}).get('reasons',[])[:8], "No qualifying Meridian transit combination."),
            ("F. Secondary astronomical confirmation", rs["secondary"], "No secondary timing confluence."),
            ("G. Quantum / price geometry", rs["quantum"], "No nearby Quantum Line confluence."),
            ("H. Technical confirmation", rs["technical"], "No directional technical confirmation."),
        ]:
            story.append(Paragraph(heading, styles["Sec"]))
            for x in (items or [fallback]):
                story.append(Paragraph(f"- {_html.escape(str(x))}", styles["SmallBody"]))

        decision = (
            "Rejected: strong-primary + technical-confirmation gate was not satisfied."
            if i["status"] == "NO TRADE"
            else "Watchlist candidate: price must confirm during/after the requested session."
            if i["mode"] == "before_market"
            else "Confirmation pending because price data for the requested date is not yet available."
            if i["status"] == "PENDING CONFIRMATION"
            else "Grouped timing and technical confirmation gate is satisfied for the requested date."
        )
        story.append(Paragraph("Decision", styles["Sec"]))
        story.append(Paragraph(decision, styles["SmallBody"]))

    story.append(PageBreak())
    story.append(Paragraph("Research Note", styles["Heading2"]))
    story.append(Paragraph(
        "This report is designed as a structured research aid. Recheck market data, position sizing, "
        "liquidity and risk parameters before acting. A NO TRADE result is a valid outcome.",
        styles["BodyText"]
    ))
    doc.build(story)
    return str(path)

def make_csv_report():
    ideas = _all_report_ideas()
    if not ideas:
        raise ValueError("No saved scan results are available. Run the main V8.2 scan once, then click the download button without re-running/resetting the notebook.")
    path = Path("/content" if Path("/content").exists() else "/mnt/data") / f"{_report_title()}_summary.csv"
    summary_dataframe(ideas).to_csv(path, index=False)
    return str(path)

def _download(path):
    print(f"Created: {path}")
    if colab_files is not None:
        colab_files.download(path)
    else:
        display(HTML(f'<a href="file://{path}" download>Download {Path(path).name}</a>'))

pdf_button = widgets.Button(description="DOWNLOAD PDF REPORT", button_style="info")
html_button = widgets.Button(description="DOWNLOAD HTML REPORT")
csv_button = widgets.Button(description="DOWNLOAD CSV SUMMARY")
export_out = widgets.Output()

def on_pdf(_):
    with export_out:
        clear_output()
        try:
            _download(make_pdf_report())
        except Exception as e:
            print("PDF export error:", e)

def on_html(_):
    with export_out:
        clear_output()
        try:
            _download(make_html_report())
        except Exception as e:
            print("HTML export error:", e)

def on_csv(_):
    with export_out:
        clear_output()
        try:
            _download(make_csv_report())
        except Exception as e:
            print("CSV export error:", e)

pdf_button.on_click(on_pdf)
html_button.on_click(on_html)
csv_button.on_click(on_csv)

display(widgets.HTML("<h3>Print / Download Report</h3>"))
display(widgets.HBox([pdf_button, html_button, csv_button]))
display(export_out)
```

```python
# ================================================================
# V8.3 REPORT PRESENTATION OVERRIDE
# ================================================================

def _summary_rows(ideas):
    rows=[["Market","Ticker","Bias","Direction","Timing","Exec window","Lunar","Direct Meridian","Confidence","Technical","Legacy combined","Data through"]]
    for i in ideas:
        d=i.get('decision_v83') or _decision_layer(i); b=d['directional_bias']; l=d['lunar']
        rows.append([
            i['name'],i['ticker'],i['bias'],f"{b['score']:+.0f} {b['label']}",d['timing_strength'],d['execution_window'] or '-',
            f"{l['strength']:.1f} {l['direction']}",f"{d['direct_meridian_score']:+.1f} {d['direct_meridian_direction']}",
            d['evidence_confidence'],i['technical_score'],i['combined_score'],i['data_asof']
        ])
    return rows

def make_html_report():
    ideas=_all_report_ideas()
    if not ideas:
        raise ValueError('No saved scan results are available. Run the main scan once, then click the download button.')
    for i in ideas: _enrich_v83(i)
    title=_report_title().replace('V8_','V8_3_') if 'V8_' in _report_title() else _report_title()+'_V8_3'
    path=Path('/content' if Path('/content').exists() else '/mnt/data')/f"{title}.html"
    def esc(x): return _html.escape(str(x))
    summary=_summary_rows(ideas)
    th=''.join(f'<th>{esc(x)}</th>' for x in summary[0])
    trs=['<tr>'+''.join(f'<td>{esc(x)}</td>' for x in row)+'</tr>' for row in summary[1:]]
    details=[]
    for rank,i in enumerate(ideas,1):
        d=i['decision_v83']; b=d['directional_bias']; l=d['lunar']; ex=i.get('execution_v8',{}); rs=i['reason_sections']
        exact=''.join(f"<li>{esc(x['mover'])} {esc(x['aspect'])} {esc('natal ' if x['relation']=='natal' else 'transiting ')}{esc(x['target'])}: approx. <b>{esc(x['exact_time_et'])} ET</b>, min orb {esc(x['estimated_exact_orb'])}°, {esc(x['polarity'])}</li>" for x in d.get('exact_aspects',[])) or '<li>No qualifying exact-time estimate.</li>'
        def lis(xs,fb): return ''.join(f'<li>{esc(x)}</li>' for x in (xs or [fb]))
        details.append(f"""
<section class='idea'>
<h2>#{rank} {esc(i['name'])} ({esc(i['ticker'])}) — {esc(i['status'])} — {esc(i['bias'])}</h2>
<div class='decision'><b>V8.3 decision layer</b><br>
Directional Bias: <b>{b['score']:+.1f}/100 — {esc(b['label'])}</b><br>
Timing Strength: <b>{d['timing_strength']}/10</b><br>
Execution Window: <b>{esc(d['execution_window'] or 'none')}</b> (peak {esc(d['best_time_et'] or 'none')} ET)<br>
Lunar: <b>{l['strength']}/10</b> | {esc(l['direction'])} | {esc(l['character'])}<br>
Direct Bill Meridian: <b>{d['direct_meridian_score']:+.1f} {esc(d['direct_meridian_direction'])}</b><br>
Execution-window aspect polarity: {d['execution_aspect_polarity_score']:+.1f} {esc(d['execution_aspect_polarity'])}<br>
Evidence Confidence: <b>{esc(d['evidence_confidence'])}</b></div>
<p><b>Research levels:</b> Price {i['price']:.2f} | Confirmation/entry {i['entry']:.2f} | Invalidation {i['stop']:.2f} | Initial objective {i['target']:.2f}</p>
<h3>Estimated aspect perfection times</h3><ul>{exact}</ul>
<h3>Legacy/source-layer scores (preserved)</h3><p>Bucholtz {i['primary_score']} | Lunar {i.get('lunar_score',0)} | Meridian {i.get('meridian_score',0)} (strength {i.get('meridian_strength',0)}) | Execution {ex.get('score',0)}/10 | Secondary {i['secondary_score']} | Quantum {i['quantum_score']} | Technical {i['technical_score']} | Combined {i['combined_score']}</p>
<h3>A. Primary Bucholtz timing</h3><ul>{lis(rs.get('primary',[]),'No qualifying primary timing event.')}</ul>
<h3>B. Lunar confluence</h3><ul>{lis(i.get('lunar_detail',{}).get('reasons',[])[:8],'No material lunar confluence.')}</ul>
<h3>C. Applying-only execution timing</h3><p>Peak ET: {esc(ex.get('best_time') or 'none')} | Window: {esc(ex.get('window') or 'none')} | Strength {ex.get('score',0)}/10 | Execution-window aspect polarity {esc(ex.get('polarity_label','UNCLASSIFIED'))} ({ex.get('polarity_score',0):+})</p><ul>{lis(ex.get('reasons',[])[:7],'No qualifying applying timing window.')}</ul>
<h3>D. Bill Meridian natal signature</h3><ul>{lis(i.get('meridian',{}).get('natal',{}).get('reasons',[])[:8],'No usable Meridian natal signature.')}</ul>
<h3>E. Bill Meridian applying transits</h3><ul>{lis(i.get('meridian',{}).get('transits',{}).get('reasons',[])[:8],'No qualifying direct Meridian transit combination.')}</ul>
<h3>F. Secondary / Quantum / Technical</h3><ul>{lis(rs.get('secondary',[])[:4]+rs.get('quantum',[])[:2]+rs.get('technical',[])[:6],'No additional confirmation.')}</ul>
</section>""")
    doc=f"""<!doctype html><html><head><meta charset='utf-8'><title>{esc(title)}</title><style>
@page{{size:A4;margin:15mm}}body{{font-family:Arial,sans-serif;color:#222;line-height:1.35;font-size:10.5pt}}h1{{font-size:20pt}}h2{{font-size:14pt;border-bottom:1px solid #aaa;padding-bottom:4px;margin-top:22px}}h3{{font-size:11pt;margin-bottom:2px}}table{{width:100%;border-collapse:collapse;font-size:8pt}}th,td{{border:1px solid #aaa;padding:4px;vertical-align:top}}th{{background:#eee}}.decision{{background:#f1f1f1;border-left:4px solid #555;padding:9px;margin:8px 0}}.idea{{page-break-before:always}}</style></head><body>
<h1>Bucholtz + Bill Meridian V8.3 Decision & Timing Report</h1>
<p><b>Requested date:</b> {esc(LAST_SCAN_RESULTS['date'])}<br><b>Mode:</b> {esc(LAST_SCAN_RESULTS['mode'])}<br><b>Generated:</b> {esc(datetime.now().strftime('%Y-%m-%d %H:%M'))}</p>
<p><b>Interpretation:</b> Direction and timing are separated. Direction is led by technical confirmation plus direct Bill Meridian evidence. Lunar score is timing intensity, not automatically bullishness. Transit-to-transit extensions are secondary. Exact-time values are ephemeris-based estimates of aspect perfection.</p>
<h2>Executive Summary</h2><table><thead><tr>{th}</tr></thead><tbody>{''.join(trs)}</tbody></table>{''.join(details)}
<p><small>Research tool only. Financial astrology is not scientifically established as a forecasting method. Recheck live market data, liquidity, position sizing and risk before acting.</small></p></body></html>"""
    path.write_text(doc,encoding='utf-8')
    return str(path)

def make_pdf_report():
    ideas=_all_report_ideas()
    if not ideas:
        raise ValueError('No saved scan results are available. Run the main scan once, then click the download button.')
    for i in ideas: _enrich_v83(i)
    title=_report_title().replace('V8_','V8_3_') if 'V8_' in _report_title() else _report_title()+'_V8_3'
    path=Path('/content' if Path('/content').exists() else '/mnt/data')/f"{title}.pdf"
    styles=getSampleStyleSheet()
    styles.add(ParagraphStyle(name='V83Small',parent=styles['BodyText'],fontSize=8.0,leading=10,spaceAfter=3))
    styles.add(ParagraphStyle(name='V83Tiny',parent=styles['BodyText'],fontSize=6.6,leading=8.0))
    styles.add(ParagraphStyle(name='V83Sec',parent=styles['Heading3'],fontSize=9.4,leading=11,spaceBefore=5,spaceAfter=2))
    styles.add(ParagraphStyle(name='V83Head',parent=styles['Heading2'],fontSize=12,leading=14,spaceBefore=6,spaceAfter=5))
    doc=SimpleDocTemplate(str(path),pagesize=A4,rightMargin=13*mm,leftMargin=13*mm,topMargin=13*mm,bottomMargin=13*mm)
    story=[Paragraph('Bucholtz + Bill Meridian V8.3 Decision & Timing Report',styles['Title']),
           Paragraph(f"<b>Requested date:</b> {LAST_SCAN_RESULTS['date']}<br/><b>Mode:</b> {LAST_SCAN_RESULTS['mode']}<br/><b>Generated:</b> {datetime.now().strftime('%Y-%m-%d %H:%M')}",styles['BodyText']),Spacer(1,5),
           Paragraph('<b>Interpretation:</b> Direction and timing are separated. Direction is led by technical confirmation plus direct Bill Meridian evidence. Lunar score is timing intensity, not automatically bullishness. Transit-to-transit extensions are secondary. Exact-time values are ephemeris-based estimates of aspect perfection.',styles['V83Small']),Spacer(1,6),Paragraph('Executive Summary',styles['Heading2'])]
    summary=_summary_rows(ideas)
    pdf_rows=[[Paragraph(str(x),styles['V83Tiny']) for x in row] for row in summary]
    widths=[25*mm,11*mm,10*mm,22*mm,10*mm,20*mm,18*mm,21*mm,10*mm,12*mm,13*mm,17*mm]
    t=Table(pdf_rows,colWidths=widths,repeatRows=1)
    t.setStyle(TableStyle([('BACKGROUND',(0,0),(-1,0),colors.lightgrey),('GRID',(0,0),(-1,-1),0.3,colors.grey),('VALIGN',(0,0),(-1,-1),'TOP'),('LEFTPADDING',(0,0),(-1,-1),1.5),('RIGHTPADDING',(0,0),(-1,-1),1.5),('TOPPADDING',(0,0),(-1,-1),2),('BOTTOMPADDING',(0,0),(-1,-1),2)]))
    story.append(t)
    def add_list(title,items,fallback):
        story.append(Paragraph(title,styles['V83Sec']))
        for x in (items or [fallback]): story.append(Paragraph('• '+str(x),styles['V83Small']))
    for rank,i in enumerate(ideas,1):
        d=i['decision_v83']; b=d['directional_bias']; l=d['lunar']; ex=i.get('execution_v8',{}); rs=i['reason_sections']
        story += [PageBreak(),Paragraph(f"#{rank} {i['name']} ({i['ticker']}) — {i['status']} — {i['bias']} — Quality {i['quality']}",styles['V83Head']),
                  Paragraph(f"<b>V8.3 Decision Layer</b><br/><b>Directional Bias:</b> {b['score']:+.1f}/100 — {b['label']}<br/><b>Timing Strength:</b> {d['timing_strength']}/10<br/><b>Execution Window:</b> {d['execution_window'] or 'none'} (peak {d['best_time_et'] or 'none'} ET)<br/><b>Lunar:</b> {l['strength']}/10 | {l['direction']} | {l['character']}<br/><b>Direct Bill Meridian:</b> {d['direct_meridian_score']:+.1f} {d['direct_meridian_direction']}<br/><b>Execution-window aspect polarity:</b> {d['execution_aspect_polarity_score']:+.1f} {d['execution_aspect_polarity']}<br/><b>Evidence Confidence:</b> {d['evidence_confidence']}",styles['V83Small']),
                  Paragraph(f"<b>Research levels:</b> Price {i['price']:.2f} | Confirmation/entry {i['entry']:.2f} | Invalidation {i['stop']:.2f} | Initial objective {i['target']:.2f}",styles['V83Small'])]
        exact=[f"{x['mover']} {x['aspect']} {'natal ' if x['relation']=='natal' else 'transiting '}{x['target']} → approx. {x['exact_time_et']} ET (min orb {x['estimated_exact_orb']:.3f}°; {x['polarity']})" for x in d.get('exact_aspects',[])]
        add_list('Estimated aspect perfection times',exact,'No qualifying exact-time estimate.')
        story.append(Paragraph(f"<b>Legacy/source-layer scores (preserved):</b> Bucholtz {i['primary_score']} | Lunar {i.get('lunar_score',0)} | Meridian {i.get('meridian_score',0)} (strength {i.get('meridian_strength',0)}) | Execution {ex.get('score',0)}/10 | Secondary {i['secondary_score']} | Quantum {i['quantum_score']} | Technical {i['technical_score']} | Combined {i['combined_score']}",styles['V83Small']))
        add_list('A. Primary Bucholtz timing',rs.get('primary',[])[:6],'No qualifying primary timing event.')
        add_list('B. Lunar confluence',i.get('lunar_detail',{}).get('reasons',[])[:8],'No material lunar confluence.')
        add_list('C. Applying-only execution timing',[f"Peak ET {ex.get('best_time') or 'none'} | Window {ex.get('window') or 'none'} | Strength {ex.get('score',0)}/10 | Execution-window aspect polarity {ex.get('polarity_label','UNCLASSIFIED')} ({ex.get('polarity_score',0):+})"]+ex.get('reasons',[])[:7],'No qualifying applying timing window.')
        add_list('D. Bill Meridian natal signature',i.get('meridian',{}).get('natal',{}).get('reasons',[])[:8],'No usable Meridian natal signature.')
        add_list('E. Bill Meridian applying transits',i.get('meridian',{}).get('transits',{}).get('reasons',[])[:8],'No qualifying direct Meridian transit combination.')
        add_list('F. Secondary / Quantum / Technical',rs.get('secondary',[])[:4]+rs.get('quantum',[])[:2]+rs.get('technical',[])[:6],'No additional confirmation.')
    story.append(Spacer(1,8)); story.append(Paragraph('Research tool only. Financial astrology is not scientifically established as a forecasting method. Recheck live market data, liquidity, position sizing and risk before acting.',styles['V83Small']))
    doc.build(story)
    return str(path)

print('V8.3 PDF/HTML presentation override loaded.')
```

## V8.4 FINAL — execution-window discipline + capped lunar display

This final patch preserves all V8.3 logic and data, while tightening the presentation and execution decision layer:

- **Lunar decision score is capped at 10.0/10**. The uncapped legacy/raw lunar score is still preserved for audit.
- **Execution-window polarity uses only classified aspects whose estimated perfection falls inside the selected execution window, with a ±30 minute buffer.**
- Aspects outside that buffer are shown separately as **Later / outside execution-window aspects** and do not influence the immediate execution polarity.
- Adds **Astrological Direction Agreement**: CONFIRMS, CONFLICTS, MIXED, or UNCLASSIFIED relative to the proposed LONG/SHORT bias.
- Keeps Bucholtz timing, Bill Meridian direct evidence, lunar timing, Quantum Lines, technical confirmation, first-trade data, exact-time estimates, PDF/HTML/CSV exports, and all existing scan behaviour.


```python

# ================================================================
# V8.4 FINAL PATCH — EXECUTION-WINDOW DISCIPLINE
# ================================================================
# Non-destructive: all V8.3 source layers remain available.
# The changes below affect only final execution-window polarity,
# capped lunar presentation, agreement labelling, and report layout.

V84_WINDOW_BUFFER_MINUTES = 30

def _v84_parse_window(window_text, base_dt):
    """Return ET datetimes for an execution window such as '09:30–10:15 ET'."""
    if not window_text or not base_dt:
        return None, None
    try:
        txt=str(window_text).replace('ET','').strip().replace('—','-').replace('–','-')
        if '-' not in txt:
            return None, None
        a,b=[x.strip() for x in txt.split('-',1)]
        et=ZoneInfo('America/New_York')
        d=base_dt.astimezone(et).date() if getattr(base_dt,'tzinfo',None) else base_dt.date()
        ah,am=map(int,a.split(':')); bh,bm=map(int,b.split(':'))
        return (datetime(d.year,d.month,d.day,ah,am,tzinfo=et),
                datetime(d.year,d.month,d.day,bh,bm,tzinfo=et))
    except Exception:
        return None, None

def _v84_aspect_sentence(x):
    rel='natal ' if x.get('relation')=='natal' else 'transiting '
    pol=x.get('polarity','UNCLASSIFIED')
    src=x.get('source','')
    return (f"{pol} [{src}] — {x.get('mover')} {x.get('aspect')} {rel}{x.get('target')} "
            f"→ approx. {x.get('exact_time_et')} ET "
            f"(min orb {float(x.get('estimated_exact_orb',0)):.3f}°).")

def _v84_polarity_from_exact(exact_items):
    """Directional polarity from immediate classified exact contacts only."""
    classified=[x for x in exact_items if x.get('polarity') in ('POSITIVE','NEGATIVE')]
    if not classified:
        return 0.0,'UNCLASSIFIED',0,0
    pos=sum(float(x.get('_directional_abs',1.0)) for x in classified if x.get('polarity')=='POSITIVE')
    neg=sum(float(x.get('_directional_abs',1.0)) for x in classified if x.get('polarity')=='NEGATIVE')
    denom=pos+neg
    net=0.0 if denom<=1e-12 else 10.0*(pos-neg)/denom
    if net>=2.0: lab='POSITIVE'
    elif net<=-2.0: lab='NEGATIVE'
    else: lab='MIXED / NEUTRAL'
    return round(net,1),lab,len([x for x in classified if x.get('polarity')=='POSITIVE']),len([x for x in classified if x.get('polarity')=='NEGATIVE'])

def execution_timing_v8(date, first_trade=None, record=None, step_minutes=15):
    """V8.4 applying-only timing with strict immediate-window polarity.

    Timing strength still identifies the strongest applying-aspect cluster.
    Directional polarity is calculated only from classified aspects estimated
    to perfect inside that execution window, with a ±30 minute buffer.
    """
    chart=_chart_node(first_trade,record)
    et=ZoneInfo('America/New_York')
    d=pd.Timestamp(date).date()
    start=datetime(d.year,d.month,d.day,9,30,tzinfo=et)
    end=datetime(d.year,d.month,d.day,16,0,tzinfo=et)
    rows=[]; dt=start

    while dt<=end:
        utc=dt.astimezone(timezone.utc)
        now=_snapshot_at(utc); nxt=_snapshot_at(utc+timedelta(minutes=step_minutes))
        hits=[]

        def add_hit(family,mover,target,relation,a,later_err,family_mult):
            if later_err >= a['current_orb'] - 1e-4:
                return
            pol=meridian_execution_polarity(mover,a['label'],target,relation)
            strength=_execution_hit_weight(a,family_mult,pol['confidence'])
            sign=1 if pol['direction']=='POSITIVE' else -1 if pol['direction']=='NEGATIVE' else 0
            directional_multiplier=1.0 if relation=='natal' else 0.45
            directional=strength*sign*pol['confidence']*directional_multiplier
            hits.append({
                'family':family,'mover':mover,'target':target,'relation':relation,
                'aspect':a['label'],'angle':a['angle'],'orb':a['current_orb'],'max_orb':a['max_orb'],
                'strength':strength,'directional':directional,'polarity':pol,
            })

        if chart:
            for np_ in [x for x in ['Sun','Moon','Mercury','Venus','Mars','Jupiter','Saturn','Uranus','Neptune','Pluto','Ascendant','MidHeaven','Node'] if x in chart]:
                a=_best_aspect(now['Moon']['lon'],chart[np_])
                if a:
                    later=aspect_distance(nxt['Moon']['lon'],chart[np_],a['angle'])
                    add_hit('moon_natal','Moon',np_,'natal',a,later,1.45)

        for p in ['Sun','Mercury','Venus','Mars','Jupiter','Saturn','Uranus','Neptune','Pluto']:
            a=_best_aspect(now['Moon']['lon'],now[p]['lon'])
            if a:
                later=aspect_distance(nxt['Moon']['lon'],nxt[p]['lon'],a['angle'])
                add_hit('moon_transit','Moon',p,'transit',a,later,0.70)

        for ip in ['Mercury','Venus','Mars']:
            for op in ['Jupiter','Saturn','Uranus','Neptune','Pluto']:
                a=_best_aspect(now[ip]['lon'],now[op]['lon'])
                if a:
                    later=aspect_distance(nxt[ip]['lon'],nxt[op]['lon'],a['angle'])
                    add_hit('inner_outer',ip,op,'transit',a,later,0.55)

        if hits:
            family_caps={'moon_natal':5.2,'moon_transit':2.4,'inner_outer':1.8}
            fam_strength={}
            for fam,cap in family_caps.items():
                vals=sorted([h['strength'] for h in hits if h['family']==fam],reverse=True)
                val=(vals[0] if vals else 0.0)+(0.30*vals[1] if len(vals)>1 else 0.0)
                fam_strength[fam]=min(cap,val*3.0)
            strength=min(10.0,sum(fam_strength.values()))
            hits_sorted=sorted(hits,key=lambda h:(0 if h['relation']=='natal' else 1,-h['strength'],h['orb']))
            rows.append({'time_et':dt.strftime('%H:%M'),'dt_et':dt,'strength':round(strength,1),
                         'family_strength':{k:round(v,2) for k,v in fam_strength.items()},
                         'raw_hits':hits_sorted})
        dt += timedelta(minutes=step_minutes)

    if not rows:
        return {'score':0.0,'label':'NO QUALIFYING WINDOW','best_time':None,'window':None,'windows':[],
                'reasons':[],'polarity_score':0.0,'polarity_label':'UNCLASSIFIED','exact_aspects':[],
                'immediate_aspects':[],'later_aspects':[],'immediate_reasons':[],'later_reasons':[],
                'window_buffer_minutes':V84_WINDOW_BUFFER_MINUTES}

    rows=sorted(rows,key=lambda r:(-r['strength'],r['time_et']))
    best=rows[0]
    label='STRONG' if best['strength']>=7 else 'MODERATE' if best['strength']>=4 else 'LIGHT'
    window=_execution_window(rows,best['strength'],tolerance=0.6,step_minutes=step_minutes)

    # Estimate exact times for a wider set of the winning cluster.
    exact=[]
    for h in best.get('raw_hits',[])[:10]:
        ex=_estimate_exact_time(best['dt_et'],h,chart,step_minutes)
        if ex:
            exact.append({**{k:h[k] for k in ['mover','target','relation','aspect','orb','max_orb']},
                          'polarity':h['polarity']['direction'],'source':h['polarity']['source'],
                          '_directional_abs':abs(float(h.get('directional',0) or 0)),
                          **ex})

    w0,w1=_v84_parse_window(window,best['dt_et'])
    lo=w0-timedelta(minutes=V84_WINDOW_BUFFER_MINUTES) if w0 else None
    hi=w1+timedelta(minutes=V84_WINDOW_BUFFER_MINUTES) if w1 else None
    immediate=[]; later=[]
    for x in exact:
        try:
            xdt=datetime.fromisoformat(x['exact_datetime_et'])
        except Exception:
            xdt=None
        if xdt is not None and lo is not None and hi is not None and lo<=xdt<=hi:
            immediate.append(x)
        else:
            later.append(x)

    pol_score,pol_label,pos_n,neg_n=_v84_polarity_from_exact(immediate)
    immediate_reasons=[_v84_aspect_sentence(x) for x in immediate]
    later_reasons=[_v84_aspect_sentence(x) for x in later]

    # Keep compact window diagnostics, but do not attach misleading bucket polarity.
    window_rows=[]
    for r in rows[:8]:
        window_rows.append({'time_et':r['time_et'],'strength':r['strength'],'family_strength':r['family_strength']})

    return {
        'score':best['strength'],'label':label,'best_time':best['time_et'],'window':window,
        'windows':window_rows,
        'reasons':immediate_reasons,
        'polarity_score':pol_score,'polarity_label':pol_label,
        'positive_hits':pos_n,'negative_hits':neg_n,
        'unclassified_hits':sum(1 for x in immediate if x.get('polarity')=='UNCLASSIFIED'),
        'family_strength':best['family_strength'],
        'exact_aspects':exact,
        'immediate_aspects':immediate,
        'later_aspects':later,
        'immediate_reasons':immediate_reasons,
        'later_reasons':later_reasons,
        'window_buffer_minutes':V84_WINDOW_BUFFER_MINUTES,
    }

def _lunar_summary(idea):
    raw=float(idea.get('lunar_score',0) or 0)
    capped=min(10.0,max(0.0,raw))
    reasons=idea.get('lunar_detail',{}).get('reasons',[]) or []
    txt=' '.join(str(x).lower() for x in reasons)
    pos=sum(txt.count(k) for k in ['supportive','trine','sextile'])
    neg=sum(txt.count(k) for k in ['tension','volatility','reversal','opposition','square'])
    if pos>neg: direction='POSITIVE'
    elif neg>pos: direction='NEGATIVE'
    else: direction='MIXED / NEUTRAL'
    env=str(idea.get('lunar_environment','') or '')
    if 'reversal' in env.lower(): character='Reversal-sensitive / volatile'
    elif 'support' in env.lower(): character='Supportive'
    elif 'mixed' in env.lower() or 'balanced' in env.lower(): character='Mixed / balanced'
    else: character=env or 'Neutral / contextual'
    return {'strength':round(capped,1),'raw_strength':round(raw,1),'direction':direction,'character':character}

def _v84_astrological_agreement(idea, direct_meridian_score, execution_polarity_score):
    """Compare directional astro evidence with the proposed LONG/SHORT bias.

    Bucholtz primary and lunar intensity are timing evidence, not directional votes.
    Directional votes come from direct Meridian evidence and immediate classified
    execution-window aspects.
    """
    bias=str(idea.get('bias','')).upper()
    expected=1 if bias=='LONG' else -1 if bias=='SHORT' else 0
    votes=[]
    if abs(float(direct_meridian_score or 0))>=5:
        votes.append(1 if direct_meridian_score>0 else -1)
    if abs(float(execution_polarity_score or 0))>=2:
        votes.append(1 if execution_polarity_score>0 else -1)
    if expected==0 or not votes:
        return 'UNCLASSIFIED'
    matches=sum(1 for v in votes if v==expected)
    opposes=sum(1 for v in votes if v==-expected)
    if matches and not opposes:
        return 'CONFIRMS'
    if opposes and not matches:
        return 'CONFLICTS'
    return 'MIXED'

def _decision_layer(idea):
    ex=idea.get('execution_v8',{}) or {}
    lunar=_lunar_summary(idea)
    direct_mer=float(idea.get('meridian_score',0) or 0)
    mer_dir='POSITIVE' if direct_mer>=5 else 'NEGATIVE' if direct_mer<=-5 else 'MIXED / NEUTRAL'
    ex_pol=float(ex.get('polarity_score',0) or 0)
    return {
        'directional_bias':_directional_bias(idea),
        'timing_strength':_timing_strength(idea),
        'execution_window':ex.get('window') or (f"{ex.get('best_time')} ET" if ex.get('best_time') else None),
        'best_time_et':ex.get('best_time'),
        'execution_aspect_polarity':ex.get('polarity_label','UNCLASSIFIED'),
        'execution_aspect_polarity_score':ex_pol,
        'direct_meridian_direction':mer_dir,
        'direct_meridian_score':direct_mer,
        'lunar':lunar,
        'evidence_confidence':_evidence_confidence(idea),
        'exact_aspects':ex.get('exact_aspects',[]),
        'immediate_aspects':ex.get('immediate_aspects',[]),
        'later_aspects':ex.get('later_aspects',[]),
        'astrological_direction_agreement':_v84_astrological_agreement(idea,direct_mer,ex_pol),
    }

def _enrich_v83(idea):
    # Keep the historical key name for backwards compatibility with the existing UI/export cells.
    if idea:
        idea['decision_v83']=_decision_layer(idea)
        idea['decision_v84']=idea['decision_v83']
    return idea

# Re-wrap the scan functions so newly generated ideas use the V8.4 execution logic.
_scan_commodities_v83=scan_commodities
_scan_stocks_v83=scan_stocks

def scan_commodities(date, long_only=False, top_n=5, mode='before_market', include_no_trade=True):
    ideas=_scan_commodities_v83(date,long_only=long_only,top_n=top_n,mode=mode,include_no_trade=include_no_trade)
    # The inherited scan calls score_idea; refresh only the execution layer here so
    # no Bucholtz/Meridian/technical source logic is lost.
    for i in ideas:
        try:
            rec=i.get('first_trade_record') or i.get('record')
            ft=i.get('first_trade') or i.get('first_trade_date')
            i['execution_v8']=execution_timing_v8(date,ft,rec)
        except Exception:
            pass
        _enrich_v83(i)
    return ideas

def scan_stocks(date, universe='liquid', long_only=False, top_n=5, max_symbols=120, mode='before_market', include_no_trade=True):
    ideas=_scan_stocks_v83(date,universe=universe,long_only=long_only,top_n=top_n,max_symbols=max_symbols,mode=mode,include_no_trade=include_no_trade)
    for i in ideas:
        try:
            rec=i.get('first_trade_record') or i.get('record')
            ft=i.get('first_trade') or i.get('first_trade_date')
            i['execution_v8']=execution_timing_v8(date,ft,rec)
        except Exception:
            pass
        _enrich_v83(i)
    return ideas

def _summary_rows(ideas):
    rows=[["Market","Ticker","Bias","Direction","Timing","Exec window","Lunar","Direct Meridian","Immediate polarity","Astro agreement","Confidence","Technical","Legacy combined","Data through"]]
    for i in ideas:
        d=i.get('decision_v84') or i.get('decision_v83') or _decision_layer(i)
        b=d['directional_bias']; l=d['lunar']
        lunar_txt=f"{l['strength']:.1f} {l['direction']}"
        if l.get('raw_strength',l['strength'])>10:
            lunar_txt+=f" (raw {l['raw_strength']:.1f})"
        rows.append([
            i['name'],i['ticker'],i['bias'],f"{b['score']:+.0f} {b['label']}",d['timing_strength'],d['execution_window'] or '-',
            lunar_txt,f"{d['direct_meridian_score']:+.1f} {d['direct_meridian_direction']}",
            f"{d['execution_aspect_polarity_score']:+.1f} {d['execution_aspect_polarity']}",
            d['astrological_direction_agreement'],d['evidence_confidence'],
            i['technical_score'],i['combined_score'],i['data_asof']
        ])
    return rows

def make_html_report():
    ideas=_all_report_ideas()
    if not ideas:
        raise ValueError('No saved scan results are available. Run the main scan once, then click the download button.')
    for i in ideas: _enrich_v83(i)
    title=_report_title().replace('V8_','V8_4_') if 'V8_' in _report_title() else _report_title()+'_V8_4'
    path=Path('/content' if Path('/content').exists() else '/mnt/data')/f"{title}.html"
    def esc(x): return _html.escape(str(x))
    summary=_summary_rows(ideas)
    th=''.join(f'<th>{esc(x)}</th>' for x in summary[0])
    trs=['<tr>'+''.join(f'<td>{esc(x)}</td>' for x in row)+'</tr>' for row in summary[1:]]
    def lis(xs,fb): return ''.join(f'<li>{esc(x)}</li>' for x in (xs or [fb]))
    details=[]
    for rank,i in enumerate(ideas,1):
        d=i.get('decision_v84') or i['decision_v83']; b=d['directional_bias']; l=d['lunar']; ex=i.get('execution_v8',{}); rs=i['reason_sections']
        immediate=[_v84_aspect_sentence(x) for x in d.get('immediate_aspects',[])]
        later=[_v84_aspect_sentence(x) for x in d.get('later_aspects',[])]
        raw_note=f" (raw legacy score {l['raw_strength']:.1f})" if l.get('raw_strength',0)>10 else ''
        details.append(f"""
<section class='idea'>
<h2>#{rank} {esc(i['name'])} ({esc(i['ticker'])}) — {esc(i['status'])} — {esc(i['bias'])}</h2>
<div class='decision'><b>V8.4 FINAL Decision Layer</b><br>
Directional Bias: <b>{b['score']:+.1f}/100 — {esc(b['label'])}</b><br>
Timing Strength: <b>{d['timing_strength']}/10</b><br>
Execution Window: <b>{esc(d['execution_window'] or 'none')}</b> (peak {esc(d['best_time_et'] or 'none')} ET)<br>
Lunar: <b>{l['strength']}/10</b>{esc(raw_note)} | {esc(l['direction'])} | {esc(l['character'])}<br>
Direct Bill Meridian: <b>{d['direct_meridian_score']:+.1f} {esc(d['direct_meridian_direction'])}</b><br>
Immediate execution polarity: <b>{d['execution_aspect_polarity_score']:+.1f} {esc(d['execution_aspect_polarity'])}</b><br>
Astrological Direction Agreement: <b>{esc(d['astrological_direction_agreement'])}</b><br>
Evidence Confidence: <b>{esc(d['evidence_confidence'])}</b></div>
<p><b>Research levels:</b> Price {i['price']:.2f} | Confirmation/entry {i['entry']:.2f} | Invalidation {i['stop']:.2f} | Initial objective {i['target']:.2f}</p>
<h3>A. Primary Bucholtz timing</h3><ul>{lis(rs.get('primary',[])[:6],'No qualifying primary timing event.')}</ul>
<h3>B. Lunar confluence</h3><ul>{lis(i.get('lunar_detail',{}).get('reasons',[])[:8],'No material lunar confluence.')}</ul>
<h3>C1. Immediate execution-window aspects (window ±{V84_WINDOW_BUFFER_MINUTES} min)</h3>
<p>Peak ET {esc(ex.get('best_time') or 'none')} | Window {esc(ex.get('window') or 'none')} | Timing strength {ex.get('score',0)}/10 | Immediate polarity {ex.get('polarity_label','UNCLASSIFIED')} ({ex.get('polarity_score',0):+})</p>
<ul>{lis(immediate,'No classified or unclassified aspect was estimated to perfect inside the execution window buffer.')}</ul>
<h3>C2. Later / outside execution-window aspects</h3><ul>{lis(later,'No later exact-time aspect from the winning timing cluster.')}</ul>
<h3>D. Bill Meridian natal signature</h3><ul>{lis(i.get('meridian',{}).get('natal',{}).get('reasons',[])[:8],'No usable Meridian natal signature.')}</ul>
<h3>E. Bill Meridian applying transits</h3><ul>{lis(i.get('meridian',{}).get('transits',{}).get('reasons',[])[:8],'No qualifying direct Meridian transit combination.')}</ul>
<h3>F. Secondary / Quantum / Technical</h3><ul>{lis(rs.get('secondary',[])[:4]+rs.get('quantum',[])[:2]+rs.get('technical',[])[:6],'No additional confirmation.')}</ul>
<h3>Legacy/source-layer scores (preserved)</h3>
<p>Bucholtz {i['primary_score']} | Lunar raw {i.get('lunar_score',0)} | Meridian {i.get('meridian_score',0)} (strength {i.get('meridian_strength',0)}) | Execution timing {ex.get('score',0)}/10 | Secondary {i['secondary_score']} | Quantum {i['quantum_score']} | Technical {i['technical_score']} | Combined {i['combined_score']}</p>
</section>""")
    doc=f"""<!doctype html><html><head><meta charset='utf-8'><title>{esc(title)}</title><style>
@page{{size:A4;margin:15mm}}body{{font-family:Arial,sans-serif;color:#222;line-height:1.35;font-size:10.5pt}}h1{{font-size:20pt}}h2{{font-size:14pt;border-bottom:1px solid #aaa;padding-bottom:4px;margin-top:22px}}h3{{font-size:11pt;margin-bottom:2px}}table{{width:100%;border-collapse:collapse;font-size:7.5pt}}th,td{{border:1px solid #aaa;padding:3px;vertical-align:top}}th{{background:#eee}}.decision{{background:#f1f1f1;border-left:4px solid #555;padding:9px;margin:8px 0}}.idea{{page-break-before:always}}</style></head><body>
<h1>Bucholtz + Bill Meridian V8.4 FINAL Decision & Timing Report</h1>
<p><b>Requested date:</b> {esc(LAST_SCAN_RESULTS['date'])}<br><b>Mode:</b> {esc(LAST_SCAN_RESULTS['mode'])}<br><b>Generated:</b> {esc(datetime.now().strftime('%Y-%m-%d %H:%M'))}</p>
<p><b>Interpretation:</b> Direction and timing are separated. Technical confirmation plus direct Bill Meridian evidence lead direction. Bucholtz and lunar scores are timing evidence. Lunar display is capped at 10/10 while the raw score remains auditable. Immediate execution polarity uses only classified aspects estimated to perfect inside the execution window ±{V84_WINDOW_BUFFER_MINUTES} minutes; later aspects are shown separately and cannot alter that immediate polarity.</p>
<h2>Executive Summary</h2><table><thead><tr>{th}</tr></thead><tbody>{''.join(trs)}</tbody></table>{''.join(details)}
<p><small>Research tool only. Financial astrology is not scientifically established as a forecasting method. Recheck live market data, liquidity, position sizing and risk before acting.</small></p></body></html>"""
    path.write_text(doc,encoding='utf-8')
    return str(path)

def make_pdf_report():
    ideas=_all_report_ideas()
    if not ideas:
        raise ValueError('No saved scan results are available. Run the main scan once, then click the download button.')
    for i in ideas: _enrich_v83(i)
    title=_report_title().replace('V8_','V8_4_') if 'V8_' in _report_title() else _report_title()+'_V8_4'
    path=Path('/content' if Path('/content').exists() else '/mnt/data')/f"{title}.pdf"
    styles=getSampleStyleSheet()
    styles.add(ParagraphStyle(name='V84Small',parent=styles['BodyText'],fontSize=8.0,leading=10,spaceAfter=3))
    styles.add(ParagraphStyle(name='V84Tiny',parent=styles['BodyText'],fontSize=6.2,leading=7.4))
    styles.add(ParagraphStyle(name='V84Sec',parent=styles['Heading3'],fontSize=9.4,leading=11,spaceBefore=5,spaceAfter=2))
    styles.add(ParagraphStyle(name='V84Head',parent=styles['Heading2'],fontSize=12,leading=14,spaceBefore=6,spaceAfter=5))
    doc=SimpleDocTemplate(str(path),pagesize=A4,rightMargin=11*mm,leftMargin=11*mm,topMargin=12*mm,bottomMargin=12*mm)
    story=[Paragraph('Bucholtz + Bill Meridian V8.4 FINAL Decision & Timing Report',styles['Title']),
           Paragraph(f"<b>Requested date:</b> {LAST_SCAN_RESULTS['date']}<br/><b>Mode:</b> {LAST_SCAN_RESULTS['mode']}<br/><b>Generated:</b> {datetime.now().strftime('%Y-%m-%d %H:%M')}",styles['BodyText']),Spacer(1,5),
           Paragraph(f'<b>Interpretation:</b> Direction and timing are separated. Technical confirmation plus direct Bill Meridian evidence lead direction. Bucholtz and lunar scores are timing evidence. Lunar display is capped at 10/10 while the raw score remains auditable. Immediate execution polarity uses only classified aspects estimated to perfect inside the execution window ±{V84_WINDOW_BUFFER_MINUTES} minutes; later aspects are shown separately and cannot alter that immediate polarity.',styles['V84Small']),Spacer(1,6),Paragraph('Executive Summary',styles['Heading2'])]
    summary=_summary_rows(ideas)
    pdf_rows=[[Paragraph(str(x),styles['V84Tiny']) for x in row] for row in summary]
    widths=[18*mm,8*mm,8*mm,17*mm,8*mm,15*mm,14*mm,16*mm,16*mm,14*mm,8*mm,9*mm,10*mm,15*mm]
    t=Table(pdf_rows,colWidths=widths,repeatRows=1)
    t.setStyle(TableStyle([('BACKGROUND',(0,0),(-1,0),colors.lightgrey),('GRID',(0,0),(-1,-1),0.3,colors.grey),
                           ('VALIGN',(0,0),(-1,-1),'TOP'),('LEFTPADDING',(0,0),(-1,-1),1.2),
                           ('RIGHTPADDING',(0,0),(-1,-1),1.2),('TOPPADDING',(0,0),(-1,-1),1.7),('BOTTOMPADDING',(0,0),(-1,-1),1.7)]))
    story.append(t)
    def add_list(title,items,fallback):
        story.append(Paragraph(title,styles['V84Sec']))
        for x in (items or [fallback]):
            story.append(Paragraph('• '+str(x),styles['V84Small']))
    for rank,i in enumerate(ideas,1):
        d=i.get('decision_v84') or i['decision_v83']; b=d['directional_bias']; l=d['lunar']; ex=i.get('execution_v8',{}); rs=i['reason_sections']
        raw_note=f" (raw legacy {l['raw_strength']:.1f})" if l.get('raw_strength',0)>10 else ''
        story += [PageBreak(),Paragraph(f"#{rank} {i['name']} ({i['ticker']}) — {i['status']} — {i['bias']} — Quality {i['quality']}",styles['V84Head']),
                  Paragraph(f"<b>V8.4 FINAL Decision Layer</b><br/><b>Directional Bias:</b> {b['score']:+.1f}/100 — {b['label']}<br/><b>Timing Strength:</b> {d['timing_strength']}/10<br/><b>Execution Window:</b> {d['execution_window'] or 'none'} (peak {d['best_time_et'] or 'none'} ET)<br/><b>Lunar:</b> {l['strength']}/10{raw_note} | {l['direction']} | {l['character']}<br/><b>Direct Bill Meridian:</b> {d['direct_meridian_score']:+.1f} {d['direct_meridian_direction']}<br/><b>Immediate execution polarity:</b> {d['execution_aspect_polarity_score']:+.1f} {d['execution_aspect_polarity']}<br/><b>Astrological Direction Agreement:</b> {d['astrological_direction_agreement']}<br/><b>Evidence Confidence:</b> {d['evidence_confidence']}",styles['V84Small']),
                  Paragraph(f"<b>Research levels:</b> Price {i['price']:.2f} | Confirmation/entry {i['entry']:.2f} | Invalidation {i['stop']:.2f} | Initial objective {i['target']:.2f}",styles['V84Small'])]
        add_list('A. Primary Bucholtz timing',rs.get('primary',[])[:6],'No qualifying primary timing event.')
        add_list('B. Lunar confluence',i.get('lunar_detail',{}).get('reasons',[])[:8],'No material lunar confluence.')
        immediate=[_v84_aspect_sentence(x) for x in d.get('immediate_aspects',[])]
        later=[_v84_aspect_sentence(x) for x in d.get('later_aspects',[])]
        add_list(f'C1. Immediate execution-window aspects (window ±{V84_WINDOW_BUFFER_MINUTES} min)',
                 [f"Peak ET {ex.get('best_time') or 'none'} | Window {ex.get('window') or 'none'} | Timing strength {ex.get('score',0)}/10 | Immediate polarity {ex.get('polarity_label','UNCLASSIFIED')} ({ex.get('polarity_score',0):+})"]+immediate,
                 'No immediate exact-time aspect.')
        add_list('C2. Later / outside execution-window aspects',later,'No later exact-time aspect from the winning timing cluster.')
        add_list('D. Bill Meridian natal signature',i.get('meridian',{}).get('natal',{}).get('reasons',[])[:8],'No usable Meridian natal signature.')
        add_list('E. Bill Meridian applying transits',i.get('meridian',{}).get('transits',{}).get('reasons',[])[:8],'No qualifying direct Meridian transit combination.')
        add_list('F. Secondary / Quantum / Technical',rs.get('secondary',[])[:4]+rs.get('quantum',[])[:2]+rs.get('technical',[])[:6],'No additional confirmation.')
        story.append(Paragraph(f"<b>Legacy/source-layer scores (preserved):</b> Bucholtz {i['primary_score']} | Lunar raw {i.get('lunar_score',0)} | Meridian {i.get('meridian_score',0)} (strength {i.get('meridian_strength',0)}) | Execution timing {ex.get('score',0)}/10 | Secondary {i['secondary_score']} | Quantum {i['quantum_score']} | Technical {i['technical_score']} | Combined {i['combined_score']}",styles['V84Small']))
    story.append(Spacer(1,8))
    story.append(Paragraph('Research tool only. Financial astrology is not scientifically established as a forecasting method. Recheck live market data, liquidity, position sizing and risk before acting.',styles['V84Small']))
    doc.build(story)
    return str(path)

print('V8.4 FINAL patch loaded: lunar cap, strict execution-window polarity, immediate/later split, astrological agreement, updated PDF/HTML.')
```

## Reading the two modes

**Before Market**  
Use this to build a watchlist for a future/current session. Even a strong candidate remains a **WATCH DATE** until price confirms.

**After Market**  
Use this after the requested session has market data. The app can then label a candidate **CONFIRMED SETUP** or reject it as **NO TRADE**.

The scoring gate requires a strong Primary Bucholtz event plus technical direction. Secondary aspects and Quantum Lines can strengthen a setup, but cannot create one by themselves.


## V8 score interpretation

- **Bucholtz Primary:** timing intensity. It is not a directional forecast by itself.
- **Lunar:** 0–20 timing environment score.
- **Meridian directional:** -100 to +100. Positive values favour the bullish side of Meridian's historical sample rules; negative values favour the bearish side. It is intentionally bounded and cannot override technical confirmation.
- **Meridian strength:** 0–100, combining natal/transit magnitude with station/eclipse timing concentration.
- **Execution:** 0–10 from applying-only intraday Moon→natal, Moon→transit and inner→outer contacts. The app reports the strongest ET window rather than inventing a time when none qualifies.
- **Technical:** remains the trade-direction gate.
- **Independent families:** a quick confluence count across Bucholtz primary, lunar, technical, Meridian and quantum evidence.

The practical workflow is: **date → shortlist → independent timing confluence → applying execution window → price confirmation → risk levels**.


# V9 — Louise McWhirter Terrain + Individual Stock Engine

This extension runs **alongside V8.4**. It does not replace Bucholtz, Bill Meridian, lunar timing, technical confirmation, Quantum Lines or the V8.4 execution-window discipline.

**Source basis:** Louise McWhirter, *Astrology and Stock Market Forecasting*, supplied by the user. The book is implemented chapter-by-chapter:

- **Chapter I — Theory and Application of Forecasting Stock Market Trends:** North Node Business Curve plus McWhirter's listed favourable/unfavourable secondary factors.
- **Chapter II — Business Cycles and Stock Market Trends 1850–1950:** historical-cycle examples are used as an **analogue/terrain-strength layer**, not converted into invented direct rules.
- **Chapter III — How to Forecast Monthly and Daily Trends on the New York Stock Exchange:** lunation, Saturn/Uranus, Mercury/Uranus/Neptune/Saturn and NYSE-chart contacts.
- **Chapter IV — How to Forecast Trends of Individual Stocks:** incorporation-chart Sun, Moon, sign ruler and Jupiter/Saturn/Uranus/Mars/outer-planet relationships.
- **Chapter V — Date and Place of Incorporation:** treated as a data chapter. Exact incorporation data are preferred; the engine refuses to silently substitute first-trade data for incorporation data.
- **Chapter VI — Simplified Ephemeris:** modern Swiss Ephemeris calculations supply the planetary positions while preserving McWhirter's rule structure.
- **Chapter VII — Glossary of Terms:** aspect geometry and favourable/unfavourable classifications are encoded explicitly.
- **Appendix — Signs Rulership of the Countries:** retained as optional mundane context and given **zero ranking weight by default** to avoid over-fitting.

### Important interpretation rule
McWhirter's historical examples are not scientifically validated forecasting evidence. Scores below are a transparent **implementation model**, not formulas published by McWhirter. The report separates **source-stated rules**, **historical analogues**, and **model-derived scoring**.


```python
# ================================================================
# V9 McWHIRTER TERRAIN + INDIVIDUAL-STOCK ENGINE
# ================================================================
from pathlib import Path
from datetime import datetime, timedelta, timezone
from zoneinfo import ZoneInfo
import math, re, pandas as pd, numpy as np

V9_VERSION = "9.5"
_V84_scan_stocks = scan_stocks
_V84_scan_commodities = scan_commodities
_V84_summary_dataframe = summary_dataframe
_V84_show_detail = globals().get('show_detail')
_V84_make_pdf_report = globals().get('make_pdf_report')
_V84_make_csv_report = globals().get('make_csv_report')

# ---------- Source-rule manifest ---------------------------------
MCWHIRTER_CHAPTERS = pd.DataFrame([
    {"chapter":"I","title":"Theory and Application of Forecasting Stock Market Trends","role":"Primary market terrain","pdf_pages":"12–16","scored":True},
    {"chapter":"II","title":"Business Cycles and Stock Market Trends 1850–1950","role":"Historical analogue / regime strength","pdf_pages":"16–58","scored":True},
    {"chapter":"III","title":"How to Forecast Monthly and Daily Trends on the New York Stock Exchange","role":"Monthly/daily market conditions","pdf_pages":"58–65","scored":True},
    {"chapter":"IV","title":"How to Forecast Trends of Individual Stocks","role":"User-adapted company-specific FIRST-TRADE chart factors (McWhirter source text uses incorporation charts)","pdf_pages":"66–76","scored":True},
    {"chapter":"V","title":"Date and Place of Incorporation of Stocks Listed on the New York Stock Exchange","role":"Source-data chapter; V9.2 adapts its company-chart slot to FIRST-TRADE/IPO database quality","pdf_pages":"77–101","scored":True},
    {"chapter":"VI","title":"Simplified Ephemeris of the Planets 1938–1950","role":"Astronomical calculation support","pdf_pages":"101–103","scored":True},
    {"chapter":"VII","title":"Glossary of Terms","role":"Aspect/rulership interpretation audit","pdf_pages":"104–109","scored":True},
    {"chapter":"Appendix","title":"Signs Rulership of the Countries","role":"Optional mundane context; no default ranking weight","pdf_pages":"109–110","scored":False},
])

MCW_ASPECTS = [
    (0.0,'conjunction',3.0),(30.0,'semisextile',3.0),(45.0,'semisquare',3.0),
    (60.0,'sextile',3.0),(90.0,'square',3.0),(120.0,'trine',3.0),
    (135.0,'sesquisquare',3.0),(150.0,'quincunx',3.0),(180.0,'opposition',3.0)
]
MCW_FAV = {'semisextile','sextile','trine'}
MCW_UNFAV = {'semisquare','square','sesquisquare','opposition'}
SIGNS=['Aries','Taurus','Gemini','Cancer','Leo','Virgo','Libra','Scorpio','Sagittarius','Capricorn','Aquarius','Pisces']

# Smooth numerical rendering of Chapter I's stated Business Curve.
# This interpolation is model-derived; sign-phase order and peaks/troughs follow the text.
MCW_NODE_CURVE = {
    'Leo':100,'Cancer':75,'Gemini':40,'Taurus':0,'Aries':-40,'Pisces':-75,
    'Aquarius':-100,'Capricorn':-75,'Sagittarius':-40,'Scorpio':0,'Libra':40,'Virgo':75,
}

# Existing V8/user research anchors for the NYSE chart. Planetary positions are additionally
# calculated from the 17-May-1792 anchor; angle values are kept configurable and clearly
# distinguished from McWhirter's rule text.
MCW_NYSE = {
    'date':'1792-05-17', 'time_utc':'15:00',
    'Ascendant':103.85,       # 13°51 Cancer
    'MidHeaven':353.35,       # 23°21 Pisces
    'Mars':168.7167,          # 18°43 Virgo
    'Neptune':207.70,         # 27°42 Libra
}

MCW_STATE_TZ={
    'NY':'America/New_York','NJ':'America/New_York','DE':'America/New_York','PA':'America/New_York',
    'MA':'America/New_York','CT':'America/New_York','MD':'America/New_York','VA':'America/New_York',
    'IL':'America/Chicago','MO':'America/Chicago','MN':'America/Chicago','WI':'America/Chicago','TX':'America/Chicago',
    'CO':'America/Denver','AZ':'America/Phoenix','CA':'America/Los_Angeles','WA':'America/Los_Angeles','OR':'America/Los_Angeles'
}

# Chapter V incorporation database. The book contains a long historical NYSE list; the engine
# supports a full transcription CSV. A source-faithful starter record for U.S. Steel is bundled.
MCW_INCORP_DB=None
MCW_INCORP_PATH=None

def load_mcwhirter_incorporations(path=None):
    global MCW_INCORP_DB, MCW_INCORP_PATH
    candidates=[]
    if path: candidates.append(Path(path))
    candidates += [Path('/content/mcwhirter_incorporations_verified_v9_1.csv'),Path('mcwhirter_incorporations_verified_v9_1.csv'),Path('/mnt/data/mcwhirter_incorporations_verified_v9_1.csv'),Path('/content/mcwhirter_incorporations.csv'),Path('mcwhirter_incorporations.csv'),Path('/mnt/data/mcwhirter_incorporations.csv')]
    p=next((x for x in candidates if x.exists()),None)
    if p is None:
        MCW_INCORP_DB=pd.DataFrame(columns=['Ticker','Company','IncorporationDate','Place','Source','Quality'])
        return MCW_INCORP_DB
    df=pd.read_csv(p)
    for c in ['Ticker','Company','IncorporationDate','Place','Source','Quality','DataQualityScore','SourceURL','EntityChoice','Notes','Precision','Status']:
        if c not in df.columns: df[c]=''
    df['Ticker']=df['Ticker'].astype(str).str.upper().str.strip()
    MCW_INCORP_DB=df
    MCW_INCORP_PATH=str(p)
    return df

# V10.2.3: legacy incorporation DB autoload disabled; McWhirter uses first-trade adaptation.

def _mcw_clamp(x,lo=-100,hi=100): return max(lo,min(hi,float(x)))
def _mcw_sign(lon): return SIGNS[int((float(lon)%360)//30)]
def _mcw_signed_for_bias(score,bias):
    return float(score) if str(bias).upper()=='LONG' else -float(score)

def _mcw_best_aspect(a,b,orb=3.0):
    best=None
    for ang,name,default_orb in MCW_ASPECTS:
        err=aspect_distance(a,b,ang)
        lim=min(float(orb),default_orb)
        if err<=lim and (best is None or err<best['orb']):
            best={'angle':ang,'name':name,'orb':float(err),'limit':lim}
    return best

def _mcw_soft_hard(aspect_name):
    if aspect_name in MCW_FAV: return 1
    if aspect_name in MCW_UNFAV: return -1
    return 0

def _mcw_node_snapshot(date):
    dt=datetime.combine(pd.Timestamp(date).date(), datetime.min.time(), tzinfo=timezone.utc)+timedelta(hours=12)
    jd=_jd(dt)
    node=_planet(jd,swe.MEAN_NODE)
    snap=astro_snapshot_datetime(dt)
    snap['Node']=node
    return snap

# ---------- Chapter I: Business Curve + secondary factors --------
def mcw_chapter1_terrain(date):
    s=_mcw_node_snapshot(date); node=s['Node']['lon']; node_sign=_mcw_sign(node)
    base=MCW_NODE_CURVE[node_sign]
    reasons=[f"North Node in {node_sign}: Business-Curve phase rendered {base:+.0f}/100 (model interpolation of McWhirter's stated sign phases)."]
    adj=0.0; fav=0; unfav=0
    def add(points,text,side):
        nonlocal adj,fav,unfav
        adj+=points; reasons.append(text)
        if side>0: fav+=1
        elif side<0: unfav+=1
    a=_mcw_best_aspect(s['Jupiter']['lon'],node,3)
    if a and a['name']=='conjunction': add(18,"Favourable Ch.I factor: Jupiter conjunct North Node.",1)
    a=_mcw_best_aspect(s['Saturn']['lon'],s['Uranus']['lon'],3)
    if a and a['name'] in {'trine','sextile','semisextile'}: add(14,f"Favourable Ch.I factor: Saturn {a['name']} Uranus.",1)
    if a and a['name'] in {'conjunction','square','opposition','semisquare'}: add(-16,f"Unfavourable Ch.I factor: Saturn {a['name']} Uranus.",-1)
    if _mcw_sign(s['Jupiter']['lon']) in {'Gemini','Cancer'}: add(10,f"Favourable Ch.I factor: Jupiter in {_mcw_sign(s['Jupiter']['lon'])}.",1)
    if _mcw_sign(node)=='Gemini': add(10,"Favourable Ch.I factor: North Node in Gemini.",1)
    if _mcw_sign(s['Saturn']['lon'])=='Gemini': add(-10,"Unfavourable Ch.I factor: Saturn in Gemini.",-1)
    if _mcw_sign(s['Uranus']['lon'])=='Gemini': add(-10,"Unfavourable Ch.I factor: Uranus in Gemini.",-1)
    a=_mcw_best_aspect(s['Saturn']['lon'],node,3)
    if a and a['name'] in {'conjunction','square','opposition'}: add(-18,f"Unfavourable Ch.I factor: Saturn {a['name']} North Node.",-1)
    a=_mcw_best_aspect(s['Uranus']['lon'],node,3)
    if a and a['name'] in {'conjunction','square','opposition'}: add(-16,f"Unfavourable Ch.I factor: Uranus {a['name']} North Node.",-1)
    # McWhirter explicitly lists favourable/unfavourable aspects to Pluto without specifying a unique planet.
    # This implementation checks Jupiter, Saturn and Uranus and caps the contribution to avoid over-counting.
    pl=0
    for p in ['Jupiter','Saturn','Uranus']:
        ap=_mcw_best_aspect(s[p]['lon'],s['Pluto']['lon'],3)
        if not ap: continue
        sh=_mcw_soft_hard(ap['name'])
        if sh>0: pl+=4; reasons.append(f"Ch.I Pluto support: {p} {ap['name']} Pluto.")
        elif sh<0: pl-=4; reasons.append(f"Ch.I Pluto pressure: {p} {ap['name']} Pluto.")
    pl=max(-8,min(8,pl)); adj+=pl
    # Compound factor from item 4: Jupiter favourable to Saturn/Uranus while Saturn and Uranus are mutually in aspect.
    js=_mcw_best_aspect(s['Jupiter']['lon'],s['Saturn']['lon'],3); ju=_mcw_best_aspect(s['Jupiter']['lon'],s['Uranus']['lon'],3)
    su=_mcw_best_aspect(s['Saturn']['lon'],s['Uranus']['lon'],3)
    if su and ((js and js['name'] in {'conjunction','sextile','trine'}) or (ju and ju['name'] in {'conjunction','sextile','trine'})):
        add(10,"Favourable Ch.I compound factor: Jupiter favourably linked with Saturn/Uranus while Saturn-Uranus are in aspect.",1)
    adj=max(-40,min(40,adj))
    score=_mcw_clamp(0.72*base+adj)
    if score>=60: regime='HIGH / PROSPERITY TERRAIN'
    elif score>=20: regime='ABOVE NORMAL / EXPANSION'
    elif score>-20: regime='TRANSITION / NORMAL'
    elif score>-60: regime='BELOW NORMAL / CONTRACTION'
    else: regime='LOW / DEPRESSION TERRAIN'
    return {'score':round(score,1),'base_curve':base,'node_sign':node_sign,'secondary_adjustment':round(adj,1),
            'favourable_factors':fav,'unfavourable_factors':unfav,'regime':regime,'reasons':reasons}

# ---------- Chapter II: historical-example analogue ---------------
def mcw_chapter2_analogue(ch1):
    # Chapter II is mainly a sequence of historical examples. Do not invent a direct prediction formula.
    n=ch1.get('favourable_factors',0)+ch1.get('unfavourable_factors',0)
    strength=min(100.0,35.0+0.45*abs(float(ch1['score']))+min(20.0,3.0*n))
    return {'score':round(strength,1),'regime':ch1['regime'],
            'directional_reference':ch1['score'],
            'reasons':["Chapter II is treated as historical analogue evidence, not a separate published scoring formula.",
                       f"Current terrain resembles the book's {ch1['regime'].lower()} examples with analogue strength {strength:.1f}/100."]}

# ---------- Chapter III: NYSE monthly / daily conditions ----------
def _mcw_nyse_chart():
    d=pd.Timestamp(MCW_NYSE['date']).date(); hh,mm=map(int,MCW_NYSE['time_utc'].split(':'))
    dt=datetime(d.year,d.month,d.day,hh,mm,tzinfo=timezone.utc)
    snap=astro_snapshot_datetime(dt)
    chart={p:snap[p]['lon'] for p in PLANETS}
    chart.update({k:v for k,v in MCW_NYSE.items() if k in ('Ascendant','MidHeaven','Mars','Neptune')})
    return chart

def _mcw_previous_new_moon(date):
    # Search previous 35 days for minimum Sun-Moon separation, then refine hourly ±36h.
    d=pd.Timestamp(date).normalize(); best=None
    for dd in range(0,36):
        x=d-pd.Timedelta(days=dd); sn=astro_snapshot(x)
        err=angular_distance(sn['Sun']['lon'],sn['Moon']['lon'])
        if best is None or err<best[0]: best=(err,x)
    center=best[1].to_pydatetime().replace(tzinfo=timezone.utc)+timedelta(hours=12)
    hb=None
    for h in range(-36,37):
        dt=center+timedelta(hours=h); sn=astro_snapshot_datetime(dt)
        err=angular_distance(sn['Sun']['lon'],sn['Moon']['lon'])
        if hb is None or err<hb[0]: hb=(err,dt,sn)
    return hb[1],hb[2]

def mcw_chapter3_market(date):
    chart=_mcw_nyse_chart(); dt_lun,lun=_mcw_previous_new_moon(date); now=_mcw_node_snapshot(date)
    score=0.0; strength=0.0; reasons=[]
    lun_lon=(lun['Sun']['lon']+lun['Moon']['lon'])/2.0
    # Lunation to key NYSE points: soft favourable, hard unfavourable; conjunction is contextual/timing only.
    for key,w in [('Neptune',18),('Mars',14),('MidHeaven',12),('Saturn',10),('Uranus',10)]:
        if key not in chart: continue
        a=_mcw_best_aspect(lun_lon,chart[key],3)
        if not a: continue
        sh=_mcw_soft_hard(a['name']); strength+=w
        if sh: score+=sh*w; reasons.append(f"Lunation {a['name']} NYSE natal {key} ({a['orb']:.2f}°): {'favourable' if sh>0 else 'unfavourable'} by Ch.VII aspect class.")
        else: reasons.append(f"Lunation conjunction/neutral-class contact to NYSE natal {key} ({a['orb']:.2f}°): timing emphasis, no automatic direction.")
    # Saturn-Uranus relationship is a stated major-trend factor in Chapters I/III.
    su=_mcw_best_aspect(now['Saturn']['lon'],now['Uranus']['lon'],3)
    if su:
        if su['name'] in {'trine','sextile','semisextile'}: score+=22; strength+=22; reasons.append(f"Saturn {su['name']} Uranus: favourable major-trend factor.")
        elif su['name'] in {'conjunction','square','opposition','semisquare'}: score-=24; strength+=24; reasons.append(f"Saturn {su['name']} Uranus: unfavourable major-trend factor.")
    # Upturn cluster described in Chapter III: Mercury conjunct Uranus, with favourable Neptune/Saturn support.
    mu=_mcw_best_aspect(now['Mercury']['lon'],now['Uranus']['lon'],3)
    if mu and mu['name']=='conjunction':
        score+=8; strength+=8; reasons.append("Mercury conjunct Uranus: Chapter III upturn-cluster component.")
        mn=_mcw_best_aspect(now['Mercury']['lon'],now['Neptune']['lon'],3)
        ms=_mcw_best_aspect(now['Mercury']['lon'],now['Saturn']['lon'],3)
        if mn and mn['name'] in MCW_FAV: score+=8; strength+=8; reasons.append(f"Mercury {mn['name']} Neptune adds favourable Chapter III support.")
        if ms and ms['name'] in MCW_FAV: score+=6; strength+=6; reasons.append(f"Mercury {ms['name']} Saturn adds favourable Chapter III support.")
    # Midheaven pressure/activity examples.
    for p,w,dirn in [('Saturn',16,-1),('Mars',8,0)]:
        dist=angular_distance(now[p]['lon'],chart['MidHeaven'])
        if dist<=10:
            strength+=w
            if dirn: score+=dirn*w
            reasons.append(f"{p} is within {dist:.1f}° of the NYSE Midheaven: {'depressing pressure' if p=='Saturn' else 'activity/volatility'} in Chapter III examples.")
    score=_mcw_clamp(score); strength=min(100,35+strength)
    return {'score':round(score,1),'strength':round(strength,1),'lunation':dt_lun.astimezone(ZoneInfo('America/New_York')).strftime('%Y-%m-%d %H:%M ET'),'reasons':reasons or ['No strong Chapter III configuration was detected.']}

# ---------- Chapter IV/V: individual stock via incorporation chart -
MCW_RULERS={
    'Aries':['Mars'],'Taurus':['Venus'],'Gemini':['Mercury'],'Cancer':['Moon'],'Leo':['Sun'],'Virgo':['Mercury'],
    'Libra':['Venus'],'Scorpio':['Mars','Pluto'],'Sagittarius':['Jupiter'],'Capricorn':['Saturn'],
    'Aquarius':['Saturn','Uranus'],'Pisces':['Jupiter','Neptune']
}

def _mcw_first_trade_record(ticker, resolved_record=None):
    """Use the SAME first-trade record already resolved by the V8.4 stock engine.

    If V8.4 supplies a VERIFIED_TIME/VERIFIED_DATE record, Chapter IV consumes that
    exact dictionary. Only if no resolved record is supplied do we perform a fallback lookup.
    """
    if isinstance(resolved_record,dict) and resolved_record.get("quality") in ("VERIFIED_TIME","VERIFIED_DATE"):
        return resolved_record
    rec=lookup_first_trade_record(ticker)
    if not rec or rec.get("quality") not in ("VERIFIED_TIME","VERIFIED_DATE"):
        return None
    return rec

def _mcw_first_trade_chart(rec):
    # Reuse the already-audited V8 first-trade chart engine, including exact time when present.
    return natal_chart_from_record(rec)

def mcw_chapter4_stock(date,ticker,name=None,resolved_record=None):
    rec=_mcw_first_trade_record(ticker,resolved_record)
    if rec is None:
        return {'score':None,'strength':0.0,'data_quality':0,'record':None,
                'chart_basis':'FIRST_TRADE_ADAPTATION',
                'reasons':['No unambiguous first-trade/IPO record is loaded for this ticker. Chapter IV is left unscored rather than guessing a date.']}

    natal=_mcw_first_trade_chart(rec)
    if not natal:
        return {'score':None,'strength':0.0,'data_quality':0,'record':rec,
                'chart_basis':'FIRST_TRADE_ADAPTATION',
                'reasons':['First-trade record exists but could not be converted into a chart.']}

    now=_mcw_node_snapshot(date)
    sun_sign=_mcw_sign(natal['Sun'])
    rulers=MCW_RULERS[sun_sign]
    quality=rec.get('quality','')
    data_quality=100.0 if quality=='VERIFIED_TIME' else 85.0 if quality=='VERIFIED_DATE' else 0.0

    score=0.0
    strength=0.0
    time_text = rec.get('time') or 'NOON DATE-ONLY MODEL'
    reasons=[
        f"FIRST-TRADE chart loaded: {rec.get('date')} {time_text} | {rec.get('place') or ''} | Sun in {sun_sign}; ruler(s): {', '.join(rulers)}.",
        f"Source: {rec.get('source','user-supplied first_trades_IPOs.csv')} | quality {quality}.",
        f"V8.4/McWhirter shared record: {rec.get('ticker',ticker)} {rec.get('date')} {rec.get('time') or 'date-only'}.",
        "Method note: McWhirter's Chapter IV company rules are being ADAPTED to the first-trade chart at the user's instruction; the supplied book itself labels these as incorporation-chart rules."
    ]

    # Natal baseline: same McWhirter-derived company relationships, now evaluated on first trade.
    for akey,bkey,w in [('Sun','Moon',8),('Jupiter','Sun',10),('Jupiter','Moon',10)]:
        a=_mcw_best_aspect(natal[akey],natal[bkey],3)
        if a:
            sh=_mcw_soft_hard(a['name'])
            strength+=w
            if sh:
                score+=sh*w
                reasons.append(f"First-trade baseline: {akey} {a['name']} {bkey} ({'supportive' if sh>0 else 'adverse'}).")

    # Jupiter/Uranus to first-trade Sun/Moon.
    for tp in ['Jupiter','Uranus']:
        for np_ in ['Sun','Moon']:
            a=_mcw_best_aspect(now[tp]['lon'],natal[np_],3)
            if not a:
                continue
            sh=_mcw_soft_hard(a['name'])
            w=14
            if a['name']=='conjunction':
                sh=1 if tp=='Jupiter' else 0
            strength+=w
            if sh:
                score+=sh*w
            reasons.append(f"{tp} {a['name']} FIRST-TRADE {np_}: {'favourable' if sh>0 else 'unfavourable' if sh<0 else 'active/ambiguous'} adapted Chapter IV contact.")

    # Saturn to first-trade Sun/Moon.
    for np_ in ['Sun','Moon']:
        a=_mcw_best_aspect(now['Saturn']['lon'],natal[np_],3)
        if not a:
            continue
        sh=_mcw_soft_hard(a['name'])
        w=18
        if a['name']=='conjunction':
            sh=-1
        strength+=w
        score+=sh*w
        reasons.append(f"Saturn {a['name']} FIRST-TRADE {np_}: {'supportive' if sh>0 else 'depressing' if sh<0 else 'timing only'}.")

    # Mars = activity rather than automatic direction.
    for np_ in ['Sun','Moon','Jupiter']:
        a=_mcw_best_aspect(now['Mars']['lon'],natal[np_],3)
        if a:
            strength+=5
            reasons.append(f"Mars {a['name']} FIRST-TRADE {np_}: activity/volatility factor, not automatic direction.")

    # Sign ruler contacts.
    for ruler in rulers:
        if ruler not in natal:
            continue
        for tp in ['Jupiter','Saturn','Uranus','Neptune','Pluto']:
            a=_mcw_best_aspect(now[tp]['lon'],natal[ruler],3)
            if not a:
                continue
            sh=_mcw_soft_hard(a['name'])
            w=7
            if tp=='Saturn' and a['name']=='conjunction':
                sh=-1
            strength+=w
            if sh:
                score+=sh*w
            reasons.append(f"{tp} {a['name']} FIRST-TRADE ruler {ruler}: {'supportive' if sh>0 else 'adverse' if sh<0 else 'contextual'}.")

    # If an exact first-trade time exists, explicitly retain the time-resolved chart evidence.
    if natal.get('_time_resolved'):
        reasons.append("Exact first-trade time available: time-resolved chart retained for the broader V8/Bucholtz/Meridian engine.")
    else:
        reasons.append("Date-only first-trade record: noon local modelling assumption used; angles are not treated as verified.")

    score=_mcw_clamp(score)
    strength=min(100.0,strength)
    return {
        'score':round(score,1),
        'strength':round(strength,1),
        'data_quality':data_quality,
        'record':rec,
        'sun_sign':sun_sign,
        'rulers':rulers,
        'chart_basis':'FIRST_TRADE_ADAPTATION',
        'reasons':reasons
    }

# ---------- Chapters V/VI/VII + Appendix audit --------------------
def mcw_reference_scores(ch4):
    ch5=float(ch4.get('data_quality',0))
    try:
        _=_mcw_node_snapshot(pd.Timestamp.today()); ch6=100.0
    except Exception: ch6=0.0
    ch7=100.0  # encoded glossary aspect geometry is loaded in MCW_ASPECTS
    return {'chapter5_data':ch5,'chapter6_ephemeris':ch6,'chapter7_glossary':ch7}

def mcw_appendix_context(date):
    # The Appendix places the United States under Gemini. Keep this contextual only by default.
    snap=_mcw_node_snapshot(date); in_gem=[p for p in ['Jupiter','Saturn','Uranus','Neptune','Pluto','Mars','Venus','Mercury'] if _mcw_sign(snap[p]['lon'])=='Gemini']
    return {'country_sign':'Gemini','planets_in_country_sign':in_gem,'weight_in_ranking':0.0,
            'note':'Appendix mundane context retained; zero default ranking weight to avoid adding a non-stock-specific directional rule.'}

# ---------- Full McWhirter chapter stack --------------------------
def mcwhirter_stack(date,ticker=None,name=None,bias='LONG',resolved_record=None):
    c1=mcw_chapter1_terrain(date); c2=mcw_chapter2_analogue(c1); c3=mcw_chapter3_market(date)
    c4=mcw_chapter4_stock(date,ticker,name,resolved_record=resolved_record) if ticker else {'score':None,'strength':0,'data_quality':0,'reasons':['Not applicable to non-stock market.']}
    refs=mcw_reference_scores(c4); app=mcw_appendix_context(date)
    # Market terrain combines Chapter I and III direction; Chapter II modulates confidence only.
    market=_mcw_clamp(0.62*c1['score']+0.38*c3['score'])
    # If Chapter IV is unavailable, keep company-specific FIRST-TRADE score N/A rather than inventing a proxy.
    individual=c4.get('score')
    available=[market]
    weights=[0.70]
    if individual is not None:
        available.append(float(individual)); weights.append(0.30)
    mcw_dir=sum(v*w for v,w in zip(available,weights))/sum(weights)
    coverage=(1 + 1 + 1 + (1 if individual is not None else 0) + (1 if refs['chapter5_data']>0 else 0) + 1 + 1)/7*100
    return {'chapter1':c1,'chapter2':c2,'chapter3':c3,'chapter4':c4,**refs,'appendix':app,
            'market_terrain_score':round(market,1),'directional_score':round(mcw_dir,1),'coverage':round(coverage,1)}

# ---------- V8.4 + McWhirter selection layer ----------------------
def enrich_v9(idea):
    if not idea: return idea
    d=idea.get('decision_v84') or idea.get('decision_v83') or _decision_layer(idea)
    bias=str(idea.get('bias','LONG')).upper(); sign=1 if bias=='LONG' else -1
    ticker=idea.get('ticker')
    is_stock = str(idea.get('kind','')).lower()=='stock' or (ticker and '=F' not in str(ticker))
    shared_record=idea.get('first_trade_record') if is_stock else None
    if is_stock and (not isinstance(shared_record,dict) or shared_record.get('quality') not in ('VERIFIED_TIME','VERIFIED_DATE')):
        shared_record=lookup_first_trade_record(ticker)
        if isinstance(shared_record,dict):
            idea['first_trade_record']=shared_record
            if shared_record.get('quality') in ('VERIFIED_TIME','VERIFIED_DATE'):
                idea['first_trade']=shared_record.get('date')
                idea['first_trade_time']=shared_record.get('time')
                idea['natal_data_quality']=shared_record.get('quality')
    mcw=mcwhirter_stack(idea['date'],ticker if is_stock else None,idea.get('name'),bias,resolved_record=shared_record)
    v8bias=float(d.get('directional_bias',{}).get('score',0) if isinstance(d.get('directional_bias'),dict) else d.get('directional_bias',0) or 0)
    timing=float(d.get('timing_strength',0) or 0)
    market_align=max(0,min(100,(sign*mcw['market_terrain_score']+100)/2))
    stock_score=mcw['chapter4'].get('score')
    stock_align=50.0 if stock_score is None else max(0,min(100,(sign*float(stock_score)+100)/2))
    v8_align=max(0,min(100,(sign*v8bias+100)/2))
    # Transparent model-derived ranking. V8.4 remains dominant; McWhirter runs alongside it.
    selection=0.40*v8_align+0.20*(timing*10)+0.22*market_align+0.13*stock_align+0.05*mcw['coverage']
    astro_agree='CONFIRMS' if sign*mcw['directional_score']>=20 else 'CONFLICTS' if sign*mcw['directional_score']<=-20 else 'MIXED / NEUTRAL'
    idea['mcwhirter']=mcw
    idea['v9_selection_score']=round(selection,1)
    idea['mcwhirter_agreement']=astro_agree
    return idea

def scan_commodities(date,long_only=False,top_n=5,mode='before_market',include_no_trade=True):
    ideas=_V84_scan_commodities(date,long_only,0,mode,include_no_trade)
    ideas=[enrich_v9(i) for i in ideas]
    ideas.sort(key=lambda x:(x.get('status')=='NO TRADE',-float(x.get('v9_selection_score',0))))
    return ideas[:top_n] if top_n else ideas

def scan_stocks(date,universe='liquid',long_only=False,top_n=5,max_symbols=120,mode='before_market',include_no_trade=True):
    ideas=_V84_scan_stocks(date,universe,long_only,0,max_symbols,mode,include_no_trade)
    ideas=[enrich_v9(i) for i in ideas]
    ideas.sort(key=lambda x:(x.get('status')=='NO TRADE',-float(x.get('v9_selection_score',0))))
    return ideas[:top_n] if top_n else ideas

def _market_confluence_target(i):
    """Choose a market-derived target from structure/Quantum levels.

    Sources:
    - prior 20/50/100-session swing highs/lows,
    - 50/200-day moving averages when directionally relevant,
    - nearby Bucholtz Quantum Lines already calculated by the engine,
    - a 20-session range projection as a market-derived fallback.

    The existing ATR target is preserved separately as the risk-based objective.
    """
    tech=i.get('technical',{}) or {}
    entry=float(i.get('entry')) if i.get('entry') is not None else None
    stop=float(i.get('stop')) if i.get('stop') is not None else None
    atr=float(i.get('atr') or tech.get('atr') or 0)
    bias=str(i.get('bias','')).upper()
    if entry is None or stop is None or atr<=0 or bias not in ('LONG','SHORT'):
        return {'level':None,'sources':[],'zone':None,'market_rr':None}

    risk=abs(entry-stop)
    candidates=[]

    def add(level,label,weight):
        if level is None:
            return
        try: level=float(level)
        except: return
        if not math.isfinite(level):
            return
        directional=(level>entry) if bias=='LONG' else (level<entry)
        if not directional:
            return
        dist=abs(level-entry)
        # Ignore noise immediately beside entry and extreme levels unlikely to be useful as an initial exit.
        if dist < max(0.35*atr,0.002*abs(entry)):
            return
        if dist > max(6*atr,4*risk):
            return
        candidates.append({'level':level,'label':label,'weight':weight,'dist':dist})

    if bias=='LONG':
        add(tech.get('high20'),'Prior 20-session high',3.0)
        add(tech.get('high50'),'Prior 50-session high',3.2)
        add(tech.get('high100'),'Prior 100-session high',3.4)
    else:
        add(tech.get('low20'),'Prior 20-session low',3.0)
        add(tech.get('low50'),'Prior 50-session low',3.2)
        add(tech.get('low100'),'Prior 100-session low',3.4)

    add(tech.get('ma50'),'50-day moving average',2.0)
    add(tech.get('ma200'),'200-day moving average',2.2)

    for e in i.get('astro_events',[]) or []:
        if e.get('type')=='quantum_line' and e.get('level') is not None:
            add(e.get('level'),f"{e.get('planet','')} Quantum Line",2.5)

    # 20-session range projection gives a directional market-structure fallback.
    h20=tech.get('high20'); l20=tech.get('low20')
    if h20 is not None and l20 is not None:
        try:
            rng=abs(float(h20)-float(l20))
            if rng>0:
                proj=entry+rng if bias=='LONG' else entry-rng
                add(proj,'20-session range projection',2.4)
        except:
            pass

    if not candidates:
        return {'level':None,'sources':[],'zone':None,'market_rr':None}

    # Cluster nearby market levels; confluence beats isolated levels.
    tol=max(0.35*atr,0.004*abs(entry))
    candidates=sorted(candidates,key=lambda x:x['level'])
    clusters=[]
    for cand in candidates:
        placed=False
        for cl in clusters:
            centre=sum(x['level']*x['weight'] for x in cl)/sum(x['weight'] for x in cl)
            if abs(cand['level']-centre)<=tol:
                cl.append(cand); placed=True; break
        if not placed:
            clusters.append([cand])

    ranked=[]
    for cl in clusters:
        w=sum(x['weight'] for x in cl)
        centre=sum(x['level']*x['weight'] for x in cl)/w
        rr=abs(centre-entry)/risk if risk else 0
        # Prefer genuine confluence and an initial-exit distance roughly 0.75R–3R.
        distance_quality=max(0,1-abs(rr-1.6)/2.0)
        score=w + 2.0*(len(cl)-1) + distance_quality
        ranked.append((score,centre,rr,cl))

    ranked.sort(key=lambda x:(-x[0],abs(x[2]-1.6)))
    _,level,rr,cl=ranked[0]
    sources=[x['label'] for x in sorted(cl,key=lambda x:-x['weight'])]

    # Exit zone around the chosen market cluster.
    half=max(0.18*atr,0.0015*abs(entry))
    zone=(level-half,level+half)

    return {
        'level':round(level,4),
        'sources':sources,
        'zone':(round(zone[0],4),round(zone[1],4)),
        'market_rr':round(rr,2)
    }

def _trade_plan_metrics(i):
    """Trader-facing plan with both market-derived and risk-based objectives."""
    entry=float(i.get('entry')) if i.get('entry') is not None else None
    invalid=float(i.get('stop')) if i.get('stop') is not None else None
    risk_target=float(i.get('target')) if i.get('target') is not None else None
    bias=str(i.get('bias','')).upper()
    risk=abs(entry-invalid) if entry is not None and invalid is not None else None
    reward=abs(risk_target-entry) if risk_target is not None and entry is not None else None
    risk_rr=(reward/risk) if risk not in (None,0) and reward is not None else None

    market=_market_confluence_target(i)
    market_target=market.get('level')
    zone=market.get('zone')

    # If the market-derived target and 1.67R target are close, combine them into a preferred exit zone.
    preferred_zone=zone
    if market_target is not None and risk_target is not None and i.get('atr'):
        if abs(market_target-risk_target) <= float(i.get('atr')):
            preferred_zone=(min(market_target,risk_target),max(market_target,risk_target))

    if bias=='LONG':
        trigger=f"LONG at/above {entry:.2f}" if entry is not None else "N/A"
        invalid_text=f"Below {invalid:.2f}" if invalid is not None else "N/A"
    elif bias=='SHORT':
        trigger=f"SHORT at/below {entry:.2f}" if entry is not None else "N/A"
        invalid_text=f"Above {invalid:.2f}" if invalid is not None else "N/A"
    else:
        trigger=f"Entry {entry:.2f}" if entry is not None else "N/A"
        invalid_text=f"{invalid:.2f}" if invalid is not None else "N/A"

    return {
        'entry':entry,'invalidation':invalid,
        'market_target':market_target,
        'market_target_sources':market.get('sources',[]),
        'market_rr':market.get('market_rr'),
        'risk_target':risk_target,'risk_rr':risk_rr,
        'preferred_zone':preferred_zone,
        'trigger':trigger,'invalid_text':invalid_text
    }

def summary_dataframe(ideas):
    rows=[]
    for i in ideas:
        if 'mcwhirter' not in i: enrich_v9(i)
        m=i['mcwhirter']; d=i.get('decision_v84') or i.get('decision_v83') or _decision_layer(i)
        b=d.get('directional_bias'); bscore=b.get('score') if isinstance(b,dict) else b
        tp=_trade_plan_metrics(i)
        rows.append({
            'Market':i['name'],'Ticker':i['ticker'],'Bias':i['bias'],'V9 Selection':i['v9_selection_score'],
            'Entry / Confirmation':tp['entry'],'Invalidation':tp['invalidation'],
            'Technical / Confluence Target':tp['market_target'],
            'Market R:R':tp['market_rr'],
            '1.67R Risk Target':tp['risk_target'],
            'Preferred Exit Zone':f"{tp['preferred_zone'][0]:.2f}–{tp['preferred_zone'][1]:.2f}" if tp['preferred_zone'] else 'N/A',
            'Execution Window':d.get('execution_window',''),'Peak ET':d.get('best_time_et',''),
            'V8.4 Direction':round(float(bscore or 0),1),'Timing':round(float(d.get('timing_strength',0) or 0),1),
            'McW Terrain':m['market_terrain_score'],'McW Agreement':i['mcwhirter_agreement'],
            'Ch I Terrain':m['chapter1']['score'],'Ch II Analogue':m['chapter2']['score'],'Ch III NYSE':m['chapter3']['score'],
            'Ch IV Individual':m['chapter4'].get('score') if m['chapter4'].get('score') is not None else 'N/A',
            'Ch V Data':m['chapter5_data'],'Ch VI Eph':m['chapter6_ephemeris'],'Ch VII Glossary':m['chapter7_glossary'],
            'Coverage %':m['coverage'],'Technical':i['technical_score'],'Data through':i['data_asof']
        })
    return pd.DataFrame(rows)

# Existing UI callback resolves this global name at click time.
def show_detail(i,rank):
    if 'mcwhirter' not in i: enrich_v9(i)
    m=i['mcwhirter']
    print('\n'+'='*105)
    print(f"#{rank} {i['name']} ({i['ticker']}) — V9 SELECTION {i['v9_selection_score']}/100 — {i['bias']} — McWhirter {i['mcwhirter_agreement']}")
    print('='*105)
    print(f"McWhirter market terrain {m['market_terrain_score']:+.1f} | coverage {m['coverage']:.0f}% | regime {m['chapter1']['regime']}")
    print(f"Ch I {m['chapter1']['score']:+.1f} | Ch II {m['chapter2']['score']:.1f} | Ch III {m['chapter3']['score']:+.1f} | Ch IV {m['chapter4'].get('score') if m['chapter4'].get('score') is not None else 'N/A'} | Ch V {m['chapter5_data']:.0f} | Ch VI {m['chapter6_ephemeris']:.0f} | Ch VII {m['chapter7_glossary']:.0f}")
    print('\nMcWhirter Chapter I — Terrain')
    for x in m['chapter1']['reasons'][:10]: print(' •',x)
    print('\nMcWhirter Chapter II — Historical analogue')
    for x in m['chapter2']['reasons']: print(' •',x)
    print('\nMcWhirter Chapter III — NYSE monthly/daily')
    print(' • Governing lunation:',m['chapter3']['lunation'])
    for x in m['chapter3']['reasons'][:10]: print(' •',x)
    print('\nMcWhirter Chapter IV/V — Individual stock / FIRST-TRADE adaptation')
    for x in m['chapter4']['reasons'][:12]: print(' •',x)
    tp=_trade_plan_metrics(i)
    d=i.get('decision_v84') or i.get('decision_v83') or _decision_layer(i)
    print('\n--- TRADE PLAN / RESEARCH LEVELS ---')
    print(' Direction:',i.get('bias'))
    print(' Entry / confirmation:',f"{tp['entry']:.2f}" if tp['entry'] is not None else 'N/A')
    print(' Invalidation:',f"{tp['invalidation']:.2f}" if tp['invalidation'] is not None else 'N/A')
    print(' Technical / confluence target:',f"{tp['market_target']:.2f}" if tp['market_target'] is not None else 'N/A')
    print(' Market-derived R:R:',f"{tp['market_rr']:.2f}:1" if tp['market_rr'] is not None else 'N/A')
    print(' 1.67R risk target:',f"{tp['risk_target']:.2f}" if tp['risk_target'] is not None else 'N/A')
    print(' Preferred exit zone:',f"{tp['preferred_zone'][0]:.2f}–{tp['preferred_zone'][1]:.2f}" if tp['preferred_zone'] else 'N/A')
    print(' Execution window:',d.get('execution_window','N/A'),'| Peak:',d.get('best_time_et','N/A'))
    print(' Rule: price confirmation activates the setup; invalidation reached first cancels it.')
    print('\n--- Existing V8.4 evidence ---')
    if callable(_V84_show_detail): _V84_show_detail(i,rank)

print('V9 McWhirter engine loaded. Chapter coverage:')
if 'display' in globals(): display(MCWHIRTER_CHAPTERS)
else: print(MCWHIRTER_CHAPTERS.to_string(index=False))
if MCW_INCORP_PATH:
    print('Legacy incorporation database (not used by V9.2 Chapter IV):',MCW_INCORP_PATH,'rows:',len(MCW_INCORP_DB))
else:
    print('V9.2 Chapter IV uses the first_trades_IPOs database, not the incorporation CSV.')
```

## V9.2 — McWhirter FIRST-TRADE Adaptation

At the user's direction, McWhirter's Chapter IV company-specific rules now run on the **same first-trade/IPO chart used by the existing V8.4/Bucholtz/Meridian engine**.

Source-integrity note: the supplied McWhirter book literally describes incorporation charts and Chapter V is titled as an incorporation-date list. V9.2 therefore labels this explicitly as a **FIRST-TRADE ADAPTATION** rather than claiming McWhirter herself used IPO dates.

Practical effect:
- `first_trades_IPOs.csv` is now the McWhirter Chapter IV data source.
- Exact `ipo_time` records produce time-resolved first-trade charts.
- Date-only records use a noon modelling assumption and do not claim verified angles.
- Ambiguous or missing first-trade records remain N/A.
- The separate incorporation database is no longer used for McWhirter scoring.


```python
# V9.2 FIRST-TRADE coverage audit for McWhirter Chapter IV adaptation
def mcwhirter_first_trade_coverage(tickers=None):
    if FIRST_TRADE_DB is None:
        load_first_trade_database()
    tickers = tickers or list(DEFAULT_STOCKS)
    rows=[]
    for t in tickers:
        r=lookup_first_trade_record(t)
        rows.append({
            'Ticker':t,
            'Quality':r.get('quality'),
            'FirstTradeDate':r.get('date'),
            'FirstTradeTime':r.get('time'),
            'Place':r.get('place'),
            'Company':r.get('company'),
            'ChapterIVReady':r.get('quality') in ('VERIFIED_TIME','VERIFIED_DATE'),
            'TimeResolved':r.get('quality')=='VERIFIED_TIME',
            'Source':r.get('source','')
        })
    return pd.DataFrame(rows)


# V10.2.1: audit is available on demand after ticker selection; not run during startup.
```

```python
# ================================================================
# V9 REPORT EXPORT — CHAPTER-BY-CHAPTER McWHIRTER AUDIT
# ================================================================
from reportlab.lib import colors
from reportlab.lib.pagesizes import A4, landscape
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import mm
from reportlab.platypus import SimpleDocTemplate,Paragraph,Spacer,Table,TableStyle,PageBreak
import html as _html


def _report_title():
    d=LAST_SCAN_RESULTS.get('date') or 'scan'; mode=LAST_SCAN_RESULTS.get('mode') or 'research'
    return f"Bucholtz_Meridian_McWhirter_V9_Report_{d}_{mode}"

def _all_report_ideas():
    ideas=list(LAST_SCAN_RESULTS.get('commodities',[]))+list(LAST_SCAN_RESULTS.get('stocks',[]))
    for i in ideas:
        if 'mcwhirter' not in i: enrich_v9(i)
    return sorted(ideas,key=lambda x:(x.get('status')=='NO TRADE',-float(x.get('v9_selection_score',0))))

def make_csv_report():
    ideas=_all_report_ideas()
    if not ideas: raise ValueError('No saved scan results. Run the main scan first.')
    path=Path('/content' if Path('/content').exists() else '/mnt/data')/f"{_report_title()}_summary.csv"
    summary_dataframe(ideas).to_csv(path,index=False)
    return str(path)

def make_pdf_report():
    ideas=_all_report_ideas()
    if not ideas: raise ValueError('No saved scan results. Run the main scan first.')
    path=Path('/content' if Path('/content').exists() else '/mnt/data')/f"{_report_title()}.pdf"
    styles=getSampleStyleSheet()
    styles.add(ParagraphStyle(name='V9Small',parent=styles['BodyText'],fontSize=7.8,leading=9.5,spaceAfter=2))
    styles.add(ParagraphStyle(name='V9Tiny',parent=styles['BodyText'],fontSize=6.1,leading=7.3))
    styles.add(ParagraphStyle(name='V9Sec',parent=styles['Heading3'],fontSize=9.3,leading=11,spaceBefore=5,spaceAfter=2))
    styles.add(ParagraphStyle(name='TradePlan',parent=styles['BodyText'],fontSize=9.0,leading=11.5,spaceAfter=3))
    doc=SimpleDocTemplate(str(path),pagesize=A4,rightMargin=10*mm,leftMargin=10*mm,topMargin=10*mm,bottomMargin=10*mm)
    now_et=datetime.now(ZoneInfo('America/New_York')).strftime('%Y-%m-%d %H:%M ET')
    data_cut=max([str(i.get('data_asof','')) for i in ideas] or [''])
    story=[
        Paragraph('Bucholtz + Bill Meridian + Louise McWhirter V10.2 Timing Continuum Report',styles['Title']),
        Paragraph(f"<b>Requested date:</b> {LAST_SCAN_RESULTS.get('date')}<br/><b>Effective mode:</b> {LAST_SCAN_RESULTS.get('mode')}<br/><b>Session validation:</b> {LAST_SCAN_RESULTS.get('mode_note','')}<br/><b>Generated:</b> {now_et}<br/><b>Latest market-data date in report:</b> {data_cut}",styles['V9Small']),
        Paragraph("<b>How to use:</b> The entry level is a confirmation trigger, not an automatic order. The setup remains valid only while price stays on the correct side of the invalidation level. The initial objective is the first planned exit/target. The execution window is a timing window, not a substitute for price confirmation.",styles['V9Small']),
        Spacer(1,4),Paragraph('Executive Trade Plan',styles['Heading2'])
    ]

    # Trader-first executive summary.
    rows=[['Market','Bias','V9','Entry','Invalid.','Mkt Target','Mkt R:R','1.67R Target','Exec window','McW IV','McW agree']]
    for i in ideas:
        m=i['mcwhirter']; d=i.get('decision_v84') or i.get('decision_v83') or _decision_layer(i); tp=_trade_plan_metrics(i)
        rows.append([
            str(i['name'])[:17],i['bias'],i['v9_selection_score'],
            f"{tp['entry']:.2f}" if tp['entry'] is not None else 'N/A',
            f"{tp['invalidation']:.2f}" if tp['invalidation'] is not None else 'N/A',
            f"{tp['market_target']:.2f}" if tp['market_target'] is not None else 'N/A',
            f"{tp['market_rr']:.2f}:1" if tp['market_rr'] is not None else 'N/A',
            f"{tp['risk_target']:.2f}" if tp['risk_target'] is not None else 'N/A',
            str(d.get('execution_window','')),
            m['chapter4'].get('score') if m['chapter4'].get('score') is not None else 'N/A',
            i['mcwhirter_agreement']
        ])
    t=Table(rows,repeatRows=1,colWidths=[25*mm,10*mm,9*mm,14*mm,14*mm,15*mm,12*mm,15*mm,21*mm,12*mm,17*mm])
    t.setStyle(TableStyle([
        ('BACKGROUND',(0,0),(-1,0),colors.lightgrey),
        ('GRID',(0,0),(-1,-1),0.25,colors.grey),
        ('FONTSIZE',(0,0),(-1,-1),5.6),
        ('VALIGN',(0,0),(-1,-1),'TOP')
    ]))
    story += [t,Spacer(1,5),
              Paragraph('McWhirter / V8.4 context summary',styles['Heading2'])]

    rows2=[['Market','V8 dir','Timing','McW terrain','Ch I','Ch II','Ch III','Ch IV','Coverage']]
    for i in ideas:
        m=i['mcwhirter']; d=i.get('decision_v84') or i.get('decision_v83') or _decision_layer(i)
        b=d.get('directional_bias'); bs=b.get('score') if isinstance(b,dict) else b
        rows2.append([str(i['name'])[:19],round(float(bs or 0),0),round(float(d.get('timing_strength',0) or 0),1),
                      m['market_terrain_score'],m['chapter1']['score'],m['chapter2']['score'],m['chapter3']['score'],
                      m['chapter4'].get('score') if m['chapter4'].get('score') is not None else 'N/A',m['coverage']])
    t2=Table(rows2,repeatRows=1,colWidths=[35*mm,16*mm,15*mm,22*mm,14*mm,14*mm,14*mm,14*mm,16*mm])
    t2.setStyle(TableStyle([('BACKGROUND',(0,0),(-1,0),colors.lightgrey),('GRID',(0,0),(-1,-1),0.25,colors.grey),('FONTSIZE',(0,0),(-1,-1),6),('VALIGN',(0,0),(-1,-1),'TOP')]))
    story += [t2,PageBreak()]

    for rank,i in enumerate(ideas,1):
        m=i['mcwhirter']; d=i.get('decision_v84') or i.get('decision_v83') or _decision_layer(i)
        b=d.get('directional_bias'); bs=b.get('score') if isinstance(b,dict) else b
        tp=_trade_plan_metrics(i)

        story.append(Paragraph(f"#{rank} {_html.escape(str(i['name']))} ({i['ticker']}) — {i['bias']} — V9 {i['v9_selection_score']}/100",styles['Heading2']))

        # Prominent execution card.
        plan_rows=[
            [Paragraph('<b>TRADE PLAN / RESEARCH LEVELS</b>',styles['TradePlan']),''],
            ['Direction',i['bias']],
            ['Entry / confirmation trigger',f"{tp['entry']:.2f}" if tp['entry'] is not None else 'N/A'],
            ['Invalidation / setup cancelled',f"{tp['invalidation']:.2f}" if tp['invalidation'] is not None else 'N/A'],
            ['Technical / confluence target',f"{tp['market_target']:.2f}" if tp['market_target'] is not None else 'N/A'],
            ['Market-derived R:R',f"{tp['market_rr']:.2f} : 1" if tp['market_rr'] is not None else 'N/A'],
            ['Market target basis',', '.join(tp['market_target_sources'][:3]) if tp['market_target_sources'] else 'No qualifying market level'],
            ['1.67R risk-based target',f"{tp['risk_target']:.2f}" if tp['risk_target'] is not None else 'N/A'],
            ['Preferred exit zone',f"{tp['preferred_zone'][0]:.2f}–{tp['preferred_zone'][1]:.2f}" if tp['preferred_zone'] else 'N/A'],
            ['Preferred timing window',
             (i.get('v102_timing_plan',{}).get('preferred_window') or
              f"{d.get('execution_window','N/A')} (legacy peak {d.get('best_time_et','N/A')} ET)")],
        ]
        pt=Table(plan_rows,colWidths=[64*mm,116*mm])
        pt.setStyle(TableStyle([
            ('SPAN',(0,0),(1,0)),
            ('BACKGROUND',(0,0),(1,0),colors.lightgrey),
            ('GRID',(0,0),(-1,-1),0.5,colors.grey),
            ('FONTSIZE',(0,0),(-1,-1),8.2),
            ('VALIGN',(0,0),(-1,-1),'MIDDLE'),
            ('TOPPADDING',(0,0),(-1,-1),4),('BOTTOMPADDING',(0,0),(-1,-1),4)
        ]))
        story += [pt,Spacer(1,4)]
        story.append(Paragraph(
            f"<b>Activation rule:</b> {'Price must confirm at/above the entry level for a LONG.' if i['bias']=='LONG' else 'Price must confirm at/below the entry level for a SHORT.'} "
            f"If the invalidation level is reached first, the setup is cancelled. The technical/confluence target is market-derived from price structure/Quantum confluence; the 1.67R level is shown separately as a risk-based objective. Neither is guaranteed.",
            styles['V9Small']
        ))

        _tp2=i.get('v102_timing_plan',{})
        if _tp2:
            story.append(Paragraph(
                f"<b>V10.2 timing conclusion:</b> {_html.escape(str(_tp2.get('classification','')))} — "
                f"{_html.escape(str(_tp2.get('reason','')))}",
                styles['V9Small']
            ))

        story.append(Paragraph(
            f"<b>V8.4 direction:</b> {float(bs or 0):+.1f} | <b>Timing:</b> {float(d.get('timing_strength',0) or 0):.1f}/10 | "
            f"<b>McWhirter terrain:</b> {m['market_terrain_score']:+.1f} | <b>McWhirter agreement:</b> {i['mcwhirter_agreement']} | "
            f"<b>Coverage:</b> {m['coverage']:.0f}%",
            styles['V9Small']
        ))
        story.append(Paragraph(
            f"<b>Chapter scores:</b> I {m['chapter1']['score']:+.1f} | II {m['chapter2']['score']:.1f}/100 analogue | "
            f"III {m['chapter3']['score']:+.1f} | IV {m['chapter4'].get('score') if m['chapter4'].get('score') is not None else 'N/A'} | "
            f"V data {m['chapter5_data']:.0f} | VI ephemeris {m['chapter6_ephemeris']:.0f} | VII glossary {m['chapter7_glossary']:.0f}",
            styles['V9Small']
        ))

        for heading,key,maxn in [
            ('McWhirter Ch.I — Business Curve terrain','chapter1',10),
            ('McWhirter Ch.II — Historical analogue','chapter2',5),
            ('McWhirter Ch.III — NYSE monthly/daily','chapter3',10),
            ('McWhirter Ch.IV/V — Individual stock & FIRST-TRADE adaptation','chapter4',12)
        ]:
            story.append(Paragraph(heading,styles['V9Sec']))
            if key=='chapter3':
                story.append(Paragraph(f"Governing lunation: {m[key].get('lunation','')}",styles['V9Small']))
            for x in m[key].get('reasons',[])[:maxn]:
                story.append(Paragraph('• '+_html.escape(str(x)),styles['V9Small']))

        # FULL V8.4 EVIDENCE SECTIONS A–F — preserved exactly in spirit from the earlier report.
        rs=i.get('reason_sections',{})
        ex=i.get('execution_v8',{})

        def add_v84_list(title,items,fallback):
            story.append(Paragraph(title,styles['V9Sec']))
            vals=items or [fallback]
            for x in vals:
                story.append(Paragraph('• '+_html.escape(str(x)),styles['V9Small']))

        add_v84_list(
            'A. Primary Bucholtz timing',
            rs.get('primary',[])[:6],
            'No qualifying primary timing event.'
        )

        add_v84_list(
            'B. Lunar confluence',
            i.get('lunar_detail',{}).get('reasons',[])[:8],
            'No material lunar confluence.'
        )

        # V10.2 previous-close → overnight → premarket → session continuum.
        if 'v102_timing_continuum' not in i:
            compute_timing_continuum_for_selected_results()
        v102_add_report_sections(story,styles,i)

        immediate=[_v84_aspect_sentence(x) for x in d.get('immediate_aspects',[])]
        later=[_v84_aspect_sentence(x) for x in d.get('later_aspects',[])]

        add_v84_list(
            f'C1. Immediate execution-window aspects (window ±{V84_WINDOW_BUFFER_MINUTES} min)',
            [f"Peak ET {ex.get('best_time') or 'none'} | Window {ex.get('window') or 'none'} | "
             f"Timing strength {ex.get('score',0)}/10 | Immediate polarity "
             f"{ex.get('polarity_label','UNCLASSIFIED')} ({ex.get('polarity_score',0):+})"] + immediate,
            'No immediate exact-time aspect.'
        )

        add_v84_list(
            'C2. Later / outside execution-window aspects',
            later,
            'No later exact-time aspect from the winning timing cluster.'
        )

        add_v84_list(
            'D. Bill Meridian natal signature',
            i.get('meridian',{}).get('natal',{}).get('reasons',[])[:8],
            'No usable Meridian natal signature.'
        )

        add_v84_list(
            'E. Bill Meridian applying transits',
            i.get('meridian',{}).get('transits',{}).get('reasons',[])[:8],
            'No qualifying direct Meridian transit combination.'
        )

        add_v84_list(
            'F. Secondary / Quantum / Technical',
            rs.get('secondary',[])[:4] + rs.get('quantum',[])[:2] + rs.get('technical',[])[:6],
            'No additional confirmation.'
        )

        _bctx=v10_bucholtz_calendar_context(i,5)
        if _bctx:
            story.append(Paragraph('G. Bucholtz Source-Rule Future Trigger Context',styles['V9Sec']))
            story.append(Paragraph(
                'These are <b>model-derived extensions</b> of trigger rules stated in Bucholtz sources; '
                'they are not dates published by Bucholtz and are reversal/pivot watches, not directional calls.',
                styles['V9Small']))
            _rows=[['Date','ET','Trigger','Planet','Rule']]
            for _r in _bctx:
                _rows.append([_r.get('Date',''),_r.get('Exact_ET',''),_r.get('Trigger_Type',''),
                              _r.get('Planet',''),_r.get('Rule','')])
            _tb=Table(_rows,repeatRows=1,colWidths=[24*mm,16*mm,31*mm,20*mm,89*mm])
            _tb.setStyle(TableStyle([
                ('BACKGROUND',(0,0),(-1,0),colors.lightgrey),('GRID',(0,0),(-1,-1),0.25,colors.grey),
                ('FONTSIZE',(0,0),(-1,-1),5.8),('VALIGN',(0,0),(-1,-1),'TOP')
            ]))
            story.append(_tb)

        story.append(Paragraph(
            f"<b>Legacy/source-layer scores (preserved):</b> "
            f"Bucholtz {i.get('primary_score',0)} | "
            f"Lunar raw {i.get('lunar_score',0)} | "
            f"Meridian {i.get('meridian_score',0)} (strength {i.get('meridian_strength',0)}) | "
            f"Execution timing {ex.get('score',0)}/10 | "
            f"Secondary {i.get('secondary_score',0)} | "
            f"Quantum {i.get('quantum_score',0)} | "
            f"Technical {i.get('technical_score',0)} | "
            f"Combined {i.get('combined_score',0)}",
            styles['V9Small']
        ))

        story.append(PageBreak())

    story.append(Paragraph('Source & Method Note',styles['Heading2']))
    story.append(Paragraph(
        "McWhirter chapter rules were extracted from the supplied scanned edition. Chapter IV/V company-chart rules are intentionally adapted to first-trade/IPO charts at the user's direction. "
        "The entry and invalidation levels come from the existing V8.4 technical research-level engine. V9.5 adds an independent market-derived technical/confluence target from swing structure, moving averages, Quantum Lines and a 20-session range projection, while preserving the original 1.67R ATR target separately. It also enforces honest before/during/after-market session labelling. "
        "V10 adds a complete lunar applying→exact→separating NYSE-session clock and model-derived future Bucholtz trigger dates. Those future dates extend source-stated rules with Swiss Ephemeris calculations and are not dates published by Bucholtz. Numerical chapter/composite scores are model implementations, not formulas published by McWhirter. Financial astrology is not scientifically established as a forecasting method.",
        styles['BodyText']
    ))
    doc.build(story)
    return str(path)

print('V10.2 timing-continuum report exporters loaded. Existing DOWNLOAD PDF/CSV buttons will now call the V9 exporters.')
```

```python

# ================================================================
# V10 — FULL LUNAR APPLY → EXACT → SEPARATE CLOCK + BUCHOLTZ CALENDAR
# ================================================================
V10_VERSION="10.0"

import io,base64,zlib,math
from datetime import datetime,timedelta,timezone
from zoneinfo import ZoneInfo

V10_LUNAR_ASPECTS = {
    "conjunction": {"angle":0.0,   "orb":5.0},
    "opposition":  {"angle":180.0, "orb":5.0},
    "trine":       {"angle":120.0, "orb":3.0},
    "sextile":     {"angle":60.0,  "orb":3.0},
    "square":      {"angle":90.0,  "orb":3.0},
    "quincunx":    {"angle":150.0, "orb":1.5},
    "semisquare":  {"angle":45.0,  "orb":1.5},
    "sesquisquare":{"angle":135.0, "orb":1.5},
}
V10_MOON_TRANSIT_TARGETS=["Sun","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
V10_MOON_NATAL_TARGETS=["Sun","Moon","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
V10_NYSE_TZ=ZoneInfo("America/New_York")

_EMBEDDED_V10_FIRST_TRADES_ZLIB_B64 = ""
_EMBEDDED_V10_CALENDAR_ZLIB_B64 = "eNrlXV13JLdxfc+vmHcNLXw2GpOThxV3LSvRavfsruxELzoUdywxYcgNOZTj/Kr8hvyyoApNbqFnBo19INBTfY6Pbdn6uLcHjQaq7r318mK3Xb/674vL3c+vPqxfX9z9x3a3/nB1+R/bu/WHu6tff93e/fzh75+267fXFzfwf13c/Rr+493D9Xb9/vbh7nL78/vdxe7hfv3dzW579+luG/7o6vbmH5RQ3ZmQ4V9rYTbGrn8If9bdxfXq24v79Q/f/tMf1z+9efndi/Of3754/x7+wfdr1es/CPF//7u+udiFP/H11cezP20vft/erF+/efnq+7OXr9599+dXL1d/fPfm9eqbH8//9Ob7Dz+t3r/58d35q9W7H79/tX63/X17d39x/fWnq99vd6u/Xewuf/vH1U34rx+v7raXACz8jX/5++pqd7+9/usfPqNUa6k2Qq/Pb//61+12/S/nhwGaGQC8u1n/VALvxf3l9ubjxc2uIro3F7v7AGym6P7y2/Zit/7pLzODp+ENkV3+DXn/cNP0BTFroTfGH31BIj4zA3yH3o8D6Gr+wo/gDr8eMwF35O1ojM6tZbexcv3q4e529cd/XXevDr+9fSuA/Vrg5nIUYHyArfBJvZZmo1T4Wv/9l+3FTVh/7wPAl6/Ov//uhxcfvnvzw8/vXr199eLD+s/bm/AZP1PqDwHo6uP28vrqBr/lA2j8/5P/vRT8y89/zepu+ykss0cC3zxc/nZ7vfuf1e4u/K/3q91v2/vt6uJ+df+3q5tfB7K7q/8Mf7C6vH7Y3ifElAZi09umT1bve/jzb6o8+Qjw/cOvF3fr99/koIU/WN3/18PF3XY1bJ5hWa+UWX1/9cvdRQWwdi3lRhQd0xo9zn6t5EaK9Td3V7ur+99Wb28fbj6uu8OP1VOIP9x+3FYEeH5xc/Hx6uJm9fL2+jr88t35XCAqCRBVP/UMh9+5JcLJh9gQo8Ut/8C7cnxXlf40dlXVraXb2InPmRYueezbT7uHm2d/8govk2FLFROngVbw4FsbFoZZv3i434WFQZfvi0OPUdkmp4KANBz63Ea6QxtBZhHLU1jEhN3BTYQRv+Q1OH1e4cWWFl7vz5e446T6E+P0+e7HhxO5Mp4+KbWWAr7qUvzb6kP4y+8f7v4eThS77fqnHw5t3rpRsSNC7TYqrVYe/wHcifwAYVcTG9uXfD3jZ77Z5zOckoTcmKNQ37368O7Nt+9evHz18/sPL959WL/e3l2G5bS+3w0POv5xeHi7u9tf7y4+buM//QugP/2Fv8Cn+yL8zQb891iWf/wByD/h0/bu6vbjPfwan8KyvtldBXbhT7v5eHb528XNr9uvf7+9Dviur3bhb3Z18/H2b/cHOB84MrDme/AQwZvx7eXtxfqcPc+/fPhudX73EP6Gb66u1+ffc+dLClHM1+/j8ZE1zSMHFdacP5+ieb+pSbGe9yb8+Q7Bj6eGkhVcK1zYlHa7QOH8w+goGw/p2g63iXgov7y9+feHGzyBPvYDbsN/VWb19iocc++f+4gbcIffp99Yl234atMlrYCHmwrAHALrJiqUJikB4n2hHrT09Z3AVqeBguigxdOVXHDj/arVDVdDhUsqaEofwXrgsouQz46V2bGdNpOrbqAHP8KR8uvearHDO/bp4Zfr8OduP+JecIYoVq/Pw6Z09fx3XoK5qL01G9RurcTG6HxJMa6d/jTWziOjbEHxJBnly4knRQlO5xstiouJNtGHxDPB878cfi060GYeO5s8QuvNIYlClc99rDVrVVBqe/XDSw4nxifG+UIbN7ZTZTZ2fA8W2bixzJfYuLE9XGDjt3IPlNe4kZwurnFjfLC0xu4NPVZYY7fxHiyrcWEZ3s8ehFZZu06L6pTSUGSRtlgCVr9KFfvsup8S0bWooD1hO15BOwSuTgnNYDHXQg3kyK97VIPxlTwJVT0yVBreLE2/f9/chgvITwdVwk+Fa7yAeyC5ux1p2L/dhn/UVYWfB6uGWhTJm+dRoSKgyxTPs4HtUAQtyqvJDUoc5lGS68tFRbU3vYixhxWQM4oQn8jtp0+391e0N1TTK2LQK+IB77Tzpmsi0DJQrFZiEzazrF3RNUIXDmlKbcLPfuB2vHdAkOMFuboM96/t3XPDtOgMdBuhsg/RmAZSe4vvjJw0zUaPSJtFaLGJ1W+snzCVu1bwLKhIhS1pQllb+SRN4BX1mxoADG9x+AJKP/UWx89Ku9c4XJaU+oJeRKcSqMNt8dlhBoDCbpTP9/U7X//c3+EZTWyk/IJzP/7qX0l7Au0q+BeckqyaEC64ZCeFY2eFRx+g6Y1x+U20yVeogzaf8Bvlps3Vw3LoT2I5QF9QgWapcMdoZG5HoOG2Kk22gdmT/uW+rqrCp6LDT4WGqlDBga/fc7RW+lJ0YL4VZqOKLv8A1YvPl5Lk4v/PD5+udtu7lViFj/dlJehwWihzNlc/LBB8Zbf8FghRHqZUaeXSN/gYhNNB+BhouxzfDeG8CN8N4bsQ3w1lzNl3Q3guwndD+LL23STrl6/vhtBcjO+GcObsu6FvKnPfDd2EOftuIs9uXDc4ejFX4gQu5m4wE1lT3HRrUssLOC0quqeMMFI0Mpg4aF9Cn7vIgSFFruX64v6yTs+VgC4qic8GtpRQJRNuQnDSorSL4MIrpTxXSYfDNpTdKFuY9TRsiOokNsQnbqVJT6fJbirn6bRYobNUyZzcQcpZ5GK6RyOPn9Y6SFm9IOawrCihMTWhl3MtsKGsxbqlOHcI4wU4dwjbRTh3KF++zh3CcgHOHcKWsXMnWblcnTuE5EKcO4QxX+cOfUNZO3foxsvXueOgy6xEdp4DnqXTZMca59UexsQIC2WKUutDT50PDY0PPQ6hMCBFLK0LNlLsBKRYlg3XlhLRZDMpTI+KHbPRHaeI1cjKQV221A1RX+TRg15V9HDlZpMCTTixSYEmnPikQPfQNhJqYwzHYHzCjmUwPuHHKhjfY9+vh2jsiTJfs4aaB4k89KZskQBzJq0pgrpMljkf3D2KNU3WmiW75Ov9491FBculxyOegpzBieZv12JIi8cxTQreJXbtMMKNYTuMsGPUDvOY2KEg8oDdNCmPhgI7OaMvbq3122eID9JSuqx/RLqGAXgRoz8SB7cPNbkxv71+2N1WRFgmS2mDUaEh3JZaW2RPvC2+qbmFgC/0Yz2Br5ZiIAVovoQEIc3EZ9+rFqUnABidhGZChlS/9gjYYnJ0z0etGEhJh6Pf8wNQBRF6HEg9r6j2CIhjelZYIty+xcANM7Os5ORVBVo4vlqqxVjCKOclWMIo32VYwhLGjC1hlOcSLGGUL2dLWLp+2VrCKM2lWMIoZ8aWsORN5W0JSzZhxpYw4GmhGmVFSQqHkmrfjfI1Fiu+xlFMVYQIANrBjAZRnN6kZNICGooVz41T4mh2C0GLn3f8fWi+SXeKovu8ac0RHXkBZwQv+resYNgTp+w49sQpP049ceClobqUcxXiazMK3qv52mg06+qF+F8oY/7+F8p2Cf6XhC9b/wtlyd//Qtny9b+kK5ep/4WSXIb/hTJm639J3lDO/pdk42Xrf4ksIRmFZ5Jw4KdQBaB9cbnC6OoaRanA6wPBsV1p2KWyja4RCuSUsGLMhK5CpeM1KtV91BkWv4/MXDn1SyZhx7I4QPixKg4okJ5I9wWZ9yqVStcRIwecUfypfFaAqDoiQNwb/vL8IsQBZ1cYYK3SIKfhWVYqmKtB02lFqdJPOaL0k31LqR+gdyBSk6b4w+SqL1yIoTsTeFoTtmAGUK8bfDojSAOlT+3yU4B6MwN8R3smKbo6cskUXKZl0h5crmPSDl0/2GzzpekWJmzEF7YX6DeZgpfXm8qhfCnAnHLUt48Ii2Ad+p24GTgiNzRUWlNiPlB+z65f4eE/AizyHjSBqDq4axqV3w20cA0+5Gpw6+luIfrZlDN//WzKdwn62RFjtvrZlCd//WzKl69+drx+mepnU5rL0M+mnNnqZ0dvKmf97GgTZqufRZ7QoXNwkC2sTenUSVntgqsgYlp6iMfg12VDfuDlwxpnYXlb60Y1Lg2FQqHBdsjudhy5OTC+L0HhlzLmrvBL2fJX+I34MlX4pSy5K/xStlwVfuOVy1Lhl5JcgsIvZcxU4Td6Q/kq/EYbL1OFX2QZ7h57ycF7B3LT1RzwHIE5BGYm+hDGVc0OSqHlNW4jbHW6kRpvk/aIrWUPos0lRL4+rxAQmWIu6ovNBnWPU9Jd8b02FT3G6+Pzo8R6j9JZ0RZAa5MahxhhmkCXGb60B7f+Sw9hvRI0ZdyUoyk7fsrRlB8f5WjgZTDGzm2MZZI+n3LikT6fcmKSPh9JOfhqS8Nm6gSyAjWX2AidVXMRMdeexLmqoCvgRQ+LOP59j49f2wFzfNwHwgshJyTAfnt1f7m9r4C7Q9wlsr6uSWPGYCSknLocKdcIHcQV6yPe370DkxyfmGoo3BGmFpCDbRzP7pYFcS/0U312kRjTQC5nYW9QEvx7BXdBayvf8Qm8omtfC4AeXZ4dy3HbyFBhLonUpXfYLgmaeiyAPfsPAd4onBycrQZVn8kesHVYQuuno6QrDzVGaFJApSq8ZouRyxLOi5DLEr4LkctSxpzlsoTnIuSyhC9ruWyyfvnKZQnNxchlCWfOcln6pjKXy9JNmLNctkNrms3NaIvjRqoPaXsC5zZWld5SGnlVu2EQr80nJPSk17Zfh6twsY04/UbIktpRm7nGEabE22k3OUmya2FHRHxSbGRxeoMn4Q3tohsidMwcEb605eqrX18d+uZ7GL9dOuNcOjrk3Leacv4IHqa1+aXIrQnjBcitCdtFyK0pX75ya8JyAXJrwpax3DpZuVzl1oTkQuTWhDFfuTV9Q1nLrenGy1duDQdvCIAJZ++JsaOikQPSYRqUgXZ9yQDfGQyYT0GXzfSdC2wZLQZi6lkPd5vKk4hTiJNPtiVIlECZIoNxhFl55HBEiVaLsFQzwjIpZxET5gZte9ioJtVZtUehIzzVwz46lVJVezgybERD2J/RkzWrFqOlESFEwPXgW8iKhhHkWX8CGizCKCsZPklGecHwKVGCYmmfMYQepedPg57CkE/PVSvmYWdTGhS7E00D/NXmMJIwonYYaNgXHw4azCSMQHuoyEuZFZI2acB5DA1yEFw5cW9p0ofxcACU8ng08fCz+jZ3KgLvcDjxbOAdiSdujS98OFTAl3cwStfQwehx95cgAi25Nje4JBGEZXfkNhg1HOuFLO2nPt3j4Ef3TTuqBHyhSKHBJdSDkFqIsczjtD1bUsCAHPASSTY2SMqJiw2ScmJjgwRSGl+p/ngtIn5Am8x2jQjRfZqV0eCG1KYiAQAdDvZxi/EpUM5L8ClQvsvwKSSMGfsUKM8l+BQoX84+hXT9svUpUJpL8SlQzox9CsmbytunkGzCjH0KwLNfiw6sAPxymCg7jjlMlB+nHKbASx44EuwFhQjS2z+Qv1KxwR8QKywEiXLNRP3ZxxEnFqdNUcrJPHpOANoBaFtcCmzSchpwYoQMt7h74NYP3pxl+C8oY/7+C8p2Cf6LhC9b/wVlyd9/Qdny9V+kK5ep/4KSXIb/gjJm679I3lDO/otk42XrvwgstUQRhChxPcdbj23UtZNnoCQIlwibm7feSJRD0WUGrs8AXW7iekN4HWayj+uSxy9d6hQkokAMw+ZNPyV+V2kUfqULr0Q7i4d8jvyse93spQeTiAYnU7lI+nGJmNNYIlB46jKxUPtrJZW81hGVAlADiThWlcmd0+zJAWSlgpOEKplwIO4rLTi1KOYplMpY0P8Uho40+/oi1PAFEX35ScG1WKY4z1X0manEp75jKIjgEmp/5NT+YmnxTQnwFCaE6bxVLq6RXrdazwGfCidflc/3Ur2ZAcDjGv4UXjWFGkGXk/C3R5dV8LeDhxYcVRwy3eaLrzBmTmZnjg0Psm/1ID0eXw2ze4OCQ6HEpK+cxUN1xOKxN3bj+W0eEadScCoo6Ua2PBKqwTVjVKmXQjnipZB9SzMFoMdQdeNLPqreVI99B4QOprZ/wVm20fkQRl91G9Mx6/H2Z0Kin8DAh6UkQEP5vSDLZ336KcKy/Iw2GDVE75i+YBJPk/NrBIn5QCrvjm1zfB3jO1qtbXEAS8FlirXtweVqte3Q9VCQ03qiYtjiSIj4wu0zXP9tN3Fm1aK6MzvCM8MEnem9pfpnPAWYU+759qk8EazDWo9gJ35N2fETv6b8+IhfIy+sQJZFxjU54DwBLLK+N4EIxREFclweDuaUEw8Hc8qJiYM5kjJwRQxXlWXYb1PO/O23Kd8l2G9HjNnab1Oe/O23KV++9tvx+mVqv01pLsN+m3Jma78dvamc7bejTZit/TbyjHY8xyY/KrJyOPyszxf52tTQFPYQ1MbYo83R+LC1HapU8eEecJiCBVKZ1dur+8vtfQXcBgfUq2IJlU7zmqtVUaOAQm+0LW3mNURqMJDdLcLWmDLmbmtM2fK3NY74MrU1piy52xpTtlxtjeOVy9LWmJJcgq0xZczU1jh6Q/naGkcbL1NbY2RpoSthyz01WjcSyijMobaZbJTTHQPxSA/CyoudOC1/CVw1xrCYA5My4jAHJmXEYg5MoKTRm+tAbph1omiTyGWfW90dkSmcwGFZjnCJDDGrWroptZhxVaPpU2wT3uERuDpCMQ0h2krkDFS4m7dYtR0CkxPFyyY/6SO0vHOv0S8qJZg5ZF88Xv6pwtp2vHwEb+DDoPsi6bu2uYmXr88rDLxMQZep4WcD26KVpkjgNjvMRZq32aB2OEPFlF9kUtNv/BzXgKlUYXhmfLxNUPbgoZJd1lQH0NrMTUKMcF1SsExLGy/1P2KD5M+Ut7FqYzSYaxIu1fn8WSJi33NPVhWyG5zY221sieWoa9JoM5DEMkw+zhmOXCN0kKLi8V44bUQdpdnAL13BhBphKkzRKLJxtvqh4f3uIQQ3n4zQ7Jf28EkMq7EkhqbZT23RRdhvlM2+L8Y0UFDYx+2x+4KLPT7Or6Q9gZKKxe0U5QsFB1RrK99QCbyis2gLgDF1ayIcpdHiDddnYSABfTHqfMJ5Eep8wnch6nzKmLM6n/BchDqf8GWtzk/WL191PqG5GHU+4cxZnU/fVObqfLoJc1bnW7xui40oDlrrknk5jw/g+U+0HRT9VDcdYzbcwvqTuIVB/GoHRbhsIy5NBavT67IQfC7tJvx7SbuowQ3sCV9ZZ6g6wg56maKgl5mmfEETpQq0GPU2YcqJb5MSJ/A2dZj66jc4lqCsFdRkM+tQYag2oluKIYMwXoAhg7BdhCGD8uVryCAsF2DIIGwZGzKSlcvVkEFILsSQQRjzNWTQN5S1IYNuvHwNGR1KbzRYtgsv4o3iIzuIRoP4Q5XVCPVEIrRvK69y+wKcBpr0BXqHfi+/rVITvBuG2+S0uPGq0qAGgOAkZlQUJoJ7EgjeLg48Qpc4AsQWBl8OF111EhfdJ26lsZenyW4q9PK0WGEF0xbnRfgGVSEd9yIx5aeoX7FyYKYQ/SZs6vlimhSNzHEBIVieN0aWSGqkyKmnw9e8jnyagC4S2swGNhgxBNgCSo0YPfVhNLRhOJTZY8clI7yVchbx0Q5k69IVzuGJz7nNsdQNAvvw7ZjUB0vZYPuCQGFToL6ND7HVcdRhrlUPYwGzHSmZhhnVeYQKD/SyK1bV1/+E9tCDkFFWn/+ENvtO9SBtDe+0cEWNvZns+AR1WbtvNrjjC2UVv3sI4cbwHkLYMbqH9GgO0BCmMe2vaPCh9HAgVXoj8qYAmY77qzCMNmLrAVt2VnVE1yL10eNgFw+PLl9KavDxRmxy6LRwERx4LDEqmPSzGJ094bwInT3huxCdPWXMWWdPeC5CZ0/4stbZJ+uXr86e0FyMzp5w5qyzp28qc5093YQ56+w9NrU0JAfluubSNUzWiBjdRuqiBoZLapRvrx92txURlnUr2mBUgNHY0p657EnT3DftmhPwhSqUJ/D1okv8kHBYMtP+dMwgUmDKoc9MFDjxLEFgaEDjtCdZ21tUXrVoIgWAUmIDfCmjEihj/sp8ynYJyvyEL1tlPmXJX5lP2fJV5qcrl6kyn5JchjKfMmarzE/eUM7K/GTjZavMB5YKJB2qZxkFBvw6EHmrrCBNCSJIOzAnrqIqLSBWBlVpRVGGSqp95fzXWAv4GsfbVVFWAegOQGtVerVWMlEIDbWA58epY9i+JF/efWi+iXiJovv8AZkjOrIhzgaeHHYyY/PiPjWKhKwJENO+JadZocAK33w5MTVl4NSfGKfs3JQT5ZSfnHJapFQcWWDyQZUtBFUADoOydVcojhyevJQn8ugf2ZXKI0+V35RA8vR4gblET2q4W+j8JE4OhzJ6X3ySNLr+y41DqMFa1JU63pRtdO5QKDpVcJEttRNJN5O5LgDeozNuSpWq0okSle4TOGlQmIOC48xO4U9hp1A4fMR+QXqeSj+zlXYMBeECcapSkeWjftucQiz0d7QB6UEAbn1WxaE6ouLYG5vx/EqOgBNCBTBMoqRAkkYLDGuyUllEPQpjfKlcQjkil5B9S73EE3rTFfsj6wsmAKWDIEIji7/Ers02pXr8VLhpYcfwnVDzFz74MyHhKAQjXm3B8JpeNzgERZAaQIo+P76mNzPAd7QwmaKroxxJwWXqku3B5cqS7dDhBDfbLcRFk3Lm76JJ+S7BRTNizNZFk/Lk76JJ+fJ10YzXL1MXTUpzGS6alDNbF83oTeXsohltwmxdNI88pZ8cXK5ahBAhPggp6SCWavqe503lQMcUYE7r49uHT0WwPQT+qK7Ej6T8XmZSRYBFdqQmEMEr4yAFlVtJBeRUciMntgItqrfontChQm2qhdjC6hIROswbFouwuqSMuVtdUrb8rS4jvkytLilL7laXlC1Xq8t45bK0uqQkl2B1SRkztbqM3lC+VpfRxsvU6hJYog4qXGiEYabVQW7h6issCA0Km+A6jcmrdjRXGMKgN6rYKaJ1o54sduxhQAQ/AW/Kjp+AN+XHR8AbeGnod8sD02Jnc/vWOBpUbWQ+5VSbrubM0ggMK8nSTJRVjKuqYkqh5XWuI2x1KqsRnYHPCw+TT8qJh8kn5cTE5BNJeTBia8vGN4esZNgkZcyAOKy2jXS0HY4/Ef4BdzLYfJVZvb0KG3yFjQpw6yMRuXu7lc3l1L8+rxBTn2IuamXMBnU/6JxLT8qpByKun+dH6UF+K2RWNQ7Q2mT/IUalYVvUqvh2VPv7b3Dn9hvhZ6u1RpQwAETAAJVZWi1SiDO1WkSQGvp7pst2qkmjes9oUbVZbcDmFO4cAe90679rcukw6G/WY3f9vva/FToHX3vTs4w/RIYKZwNpUeLKGWUwwFqu4MgJMO2QQynscvTYhPMi9NiE70L02JQxZz024bkIPTbhy1qPnaxfvnpsQnMxemzCmbMem76pzPXYdBPmrMe2EOgtomk/c6MwpoEE02LuhIcuakEVztrKjRYCr6jg1gCgwrA4K0pra13SS3tc3M8P02I652J0rITxAnSshO0idKyUL18dK2G5AB0rYctYx5qsXK46VkJyITpWwpivjpW+oax1rHTj5atjtTDkWinInMuqmNLgqzpCoQ7vhgYceXNtEBKI820QBpAWQ61m3Q7uUOjXTYvWXN1J6wgN3Kse2lGFF7xGruoOZ0R2G+2y+omeyCf2JUAVbsyIE3JjfUl/r9/zAlfq73XQhlTYaC1M3PMkcK9d3F6EjuK94z3iPejVXys37EvWz1V8jAjD9i7V1Bg+0chW4NCH0oEaqWQSqchp0cJjrSNGI6DLhpPOBXZUWZpsJISUs4iEcCj4M+BYmFTZSNng3YcvVb/RMi8Cx9f/bPZDSVNGWQn4STLKC8BPiRIE8YuN7LhKpxxE3os+E3l/9Af0p/EDou9Sqfx9VaaOxBpbGkw1gkOtnRSt4eOew7CtiNrAd+L4ZMUDyKtP20KgceC20EdCbQdsvs1BjMA7HGs7G3hHgm1b44PLtRwP3zlxX47H8hBGL7GxuhFObKxuhBMfqxvYlodpJ4tRJxPOi1AnE74LUSdTxpzVyYTnItTJhC9rdXKyfvmqkwnNxaiTCWfO6mT6pjJXJ9NNmLM62Q9pVtJzDCci7FiGExF+rMKJoCyAQ65MVjNffx5vxOYhI013U53BrkWqMsKDOUbd8TJpLPg0GWCOCGEeHGYU5KQK0jWMekCMMD7WFjVXG0h7CMKyTmobjAow6mIVxZP6CH5031RHQcAXSn8aSKc89HyUgtiUhTg7COMFODsI20U4Oyhfvs4OwnIBzg7ClrGzI1m5XJ0dhORCnB2EMV9nB31DWTs76MbL19nhwb8Czg7LL6HcowpIQT5XsWjF6OpVAinA9R9waslrptRArM9Lt/E21Ea5DQB7EGOZrkT/HleIbVMKCVilhFVyPK4xs1zMaSwX2cFeJGU2olAQ5fSBaNqK8mlA3MO939opb4JKc1LrqOECPviG6SO28L0nOwutIYDuELQvLaY0kRoCTg9yoDK/V1wFaRG6Tpk3AA1flvBAdXdEExmfoW+1rz2hOyyJnAu6I4rIpvAktB5hMoMqUxOnBt1h/VV6ryV+wDx0G4o/tq7F+xKAKmiL6An1eKuuSAAIXRGxkSLv2onPsNfNUKKYGCLQjzTmBoBmBgCPa8VTeNXOpgRdTireHl1WKd4OnsG9xhSfIVrc/ORwQzVyqgHbxiouFX5fOrBgFxqEm92PFPqEHdyPCjbFBsPCI8LwXZGW6w1OgaAA5t3ZiQt/m+uQwhOIg8NcSVRI/fHiFGFZUkgbjLiMBbdylTrDSiEoiUo/Gk3ucwoCJ4TFOJmMKEZ1RBSzN8zh+YUxA05fmN/R9G6iBp2RNKXqE+WI+kT2LeUnET1UTKZ0XFq4NutVYYKHWIzFiHJegsWI8l2GxShhzNhiRHkuwWJE+XK2GKXrl63FiNJcisWIcmZsMUreVN4Wo2QTZmwxAp4OB6+64nqTa3MF0xIHZXpeshktzoSEad8CbxPTk/daNBcGkAaGV9s+P3uvRW9hD9/RnmuD6vgIXKbl2h5cruPaDl0P6X7CTfQGGxTsIz6JrUFjl2DyGDFmbvIYsWVv8hjz5WnyGLFkbvIYsWVq8thbuRxNHiOSCzB5jBjzNHmM31C2Jo/xxsvT5DGw1EPHtlR5p9Nk1JrnVw0acOUKLp+1RRwjgDmRum8e7z2ARX/D4cSWfdBVtQQjgEXJBE0gKtQVqfK5KFo3qzqoDuQjuUAPeIz1+8ARHWipBcfQaaCnhsIFjlU5+bD6ESMGYfUjRhzC6gdKqLlTfV5mrU1XceJtRCZxME543ydkKcbVDEMZYZtwbo3A1TllKDhlSCwzFvZBWh3Y1OOQJlWinIpP1A6fR3yN0uSex+PRt9uw0q+eHzy01TwMeihU/DX7tCNUmEnRF2lXtc0Nw3l9/vyzcEagy+Sss4GNI6fEFxz5UkFzbPg9M0wNjTrIIy83VOna+6zGeX39Rohsn67+x0lDHGI41tt+4qhc/9tEoOVV9E0+TRrnW/nCO+U8XmiCueiaORvUPeqni/XobXYhiRmXqCM7rkYHaE0iGiPGONBJmeLjVO2X3qDTqtS+1DU56xm8PcmYxplzdLpW8HAOo5Lc8olH7NjlE4/4scknHnh53MX7Iov+yFMO16Hnt8BEnOB6V+DoXYRLY8SZvUtjxHcBLo0xY64ujRFP9i6NEV+2Lo299cvTpTGiuQiXxogzV5fG+E1l7NIYb8JcXRoDTw13VeWzGggigdizn1eVQRiYXit7CN5gMWBzxInFgM0RJx4DNgdSFuejiwJFU6u6DjTA/EaovJmmVVkHYjk0iFqYDP2NrBQ2oGVfEo7RsCoAHTy7sUelg8fEE19JewLqCYNho2Ijj2epxFX11LSOq+hA8CwkpIYv2tur8GJUKBeDUVJsjMiXYo1pIPSyWCfu4nyD4/tJK2wdtivsdHrRsI5PQQWEtGQHATYL8X0RxgvwfRG2i/B9Ub58fV+E5QJ8X4QtY99XsnK5+r4IyYX4vghjvr4v+oay9n3RjZev78tCPTKcdVVZ3qm1lTV4BF+ZPLQFQgvXX2FKxG4t4RXp2loAdKAYUnqqdhPvWkqcxF0LVOOyMFwciXXJZJXHjeXZH76KcwxsqX6wEUwtYIB3+PesEjfNh60jdkVsShzZQY8vY3USy/iJW6mg7DTZTcnJTotVTPq1E5aqFi9LBxJWAOe+oFYcq6tfSXUKRfsOj41yY6aU+WkoHZSGn//hxyQDbaYMbU2wefhgSl36IWqUnoVAwdyf17L3RMq+3wiocKzqMFm9ZOwTwt3zzFdqHnWYqq5hsGVhqronoerNItUH6Nj3MqLY1thTV2NDU2OHQ2VEZhvee+oNdgTV48hTXXyEbrYlKA8fNWuK5NvN3jWH8nm/H9mzN2tVNHKwOrC1wUBlW3KXliJnwQo/fx0PFgFddMOeDWyJozeELraG1t8EHLrypmRkUs4iSscNZrxw9JuU8UjZ4FkOTnY5dfxr+P6Dq9rDsOWSauRs3qQn1GU1yvngdhi2I/NVFZnmRtRYqz1k6MLQUsuvqEK4MSyqEHaMiio9JuigJGjaettgb++HgWtKLMeZRzgvwplH+C7EmUcZc3bmEZ6LcOYRvqydecn65evMIzQX48wjnDk78+ibytyZRzdhzs68HoNuTHY8STzGtrj2AbZ+7Ao4bS2Ix6njEm5EnOwEfhhVLjqurVyPg3xspkvCTQ1IGC/AMEHYLsIwQfnyNUwQlgswTBC2jA0TycrlapggJBdimCCM+Rom6BvK2jBBN16+hglkKcVGd1njsuySHsCPdxcVclY9dikO3Nb30bUY7uFR64WPjmVQAPILN2vrs9JA6RrG3EaMPWiVShQ2LtEqvb1+2N1WRFgmp2mDUYGSSnWlqkXZE9mib6pbJOALpbZP4KtlHUuBRROxMXpqK/OqhegvAJQWAiyUYBQ5A6wcGG1El43tEkRvdSDmpKLoakDsNsaxCe6inLgEd1FObIK7AimoV3c848QpO45x4pQfpzhx4GXhIiZlifVDSbXvA/saj1FfY2BVFXE6gHYIui89lSiZqFSHY1QFnD2O3ZoYcNDiAhjAgU9WF4wsa3IDlPJMSCwR0Kji/d/VN1EfU3SfP7tzREc+oDOCJ9FMa8qHvEk3kyFvAF6BaFUeCODMfA38KXwNcBoxOOwmvPNqFCBZc+XE6Re2TPRf/75PIRYq/JuAhPKagsJJqWGuwcVewqEr3DBL1BDDe6bUabxncSKgdcUHGKPrHxEUOvAVOKoKrZ/KNtoY1KAwMW4xonbKeQmidsp3GaL2hDFjUTvluQRRO+XLWdSerl+2onZKcymidsqZsag9eVN5i9qTTZixqB14dkNM5JG3NXO9MKdwvVCg7QCztppoiKp0wG+lsqjCnM4OqgxTlcdGDVuFM0K6jVXF97O0hFupRqogRULKTCH89JcyOMXCppTXyaiO6GT2pkM9v1ZmwGmh6lPSR0nz9IbVUql7olA0Y6ErUShIUY4IUmTfUpEC6O2iJlNQxvyNFpTtEowWCV+2RgvKkr/RgrLla7RIVy5TowUluQyjBWXM1miRvKGcjRbJxsvWaAEsHQZJFWeUKtfmvqYlGELCbYZLQ/P/AXhLEvA="

def _v10_embedded_first_trade_df():
    raw=zlib.decompress(base64.b64decode(_EMBEDDED_V10_FIRST_TRADES_ZLIB_B64))
    return pd.read_csv(io.BytesIO(raw),encoding="utf-8-sig",dtype=str,keep_default_na=False)

def _v10_embedded_calendar_df():
    raw=zlib.decompress(base64.b64decode(_EMBEDDED_V10_CALENDAR_ZLIB_B64))
    return pd.read_csv(io.BytesIO(raw),dtype=str,keep_default_na=False)

# ----- First-trade database: prefer the expanded Bucholtz-aware database -----
def load_first_trade_database(path=None):
    global FIRST_TRADE_DB,FIRST_TRADE_DB_PATH
    candidates=[]
    if path:
        candidates.append(Path(path))
    candidates += [
        Path("/content/first_trades_IPOs_clean_Bucholtz_markets.csv"),
        Path("first_trades_IPOs_clean_Bucholtz_markets.csv"),
        Path("/mnt/data/first_trades_IPOs_clean_Bucholtz_markets.csv"),
    ]
    db=None; found=None
    for p in candidates:
        if not p.exists():
            continue
        try:
            trial=pd.read_csv(p,dtype=str,keep_default_na=False)
            if {"Ticker","Date"}.issubset(trial.columns) and len(trial):
                db=trial; found=str(p); break
        except Exception:
            pass
    if db is None:
        db=_v10_embedded_first_trade_df()
        found="EMBEDDED:first_trades_IPOs_clean_Bucholtz_markets.csv"

    for col in ["name","Place","ipo_time","description","market_tz","source_note","source_quality"]:
        if col not in db.columns:
            db[col]=""
    db=db.loc[:,~db.columns.astype(str).str.startswith("Unnamed:")].copy()
    db["Ticker"]=db["Ticker"].astype(str).str.upper().str.strip().str.replace(".","-",regex=False)
    db["_parsed_date"]=pd.to_datetime(db["Date"],errors="coerce",dayfirst=False)
    db=db[db["_parsed_date"].notna()].copy()
    FIRST_TRADE_DB=db
    FIRST_TRADE_DB_PATH=found
    print(f"V10 first-trade database: {len(db):,} rows from {found}; Bucholtz futures/FX anchors included.")
    return db

def lookup_first_trade_record(ticker):
    global FIRST_TRADE_DB
    if FIRST_TRADE_DB is None:
        load_first_trade_database()
    if FIRST_TRADE_DB is None:
        return {"quality":"MISSING_DATABASE","ticker":ticker}
    t=str(ticker).upper().strip().replace(".","-")
    x=FIRST_TRADE_DB[FIRST_TRADE_DB["Ticker"].str.replace(".","-",regex=False)==t].copy()
    if x.empty:
        return {"quality":"NOT_FOUND","ticker":ticker}
    x["time_norm"]=x["ipo_time"].map(_normalize_hhmm)
    x["date_norm"]=x["_parsed_date"].dt.strftime("%Y-%m-%d")
    dates=sorted(x["date_norm"].dropna().unique().tolist())
    if len(dates)>1:
        return {"quality":"AMBIGUOUS","ticker":ticker,"candidate_dates":dates,
                "candidates":x[["Ticker","name","Date","ipo_time","Place"]].to_dict("records")}
    x["_rank"]=(x["source_quality"].str.startswith("BUCHOLTZ").astype(int)*8 +
                x["name"].astype(str).ne("").astype(int)*2 +
                x["time_norm"].notna().astype(int))
    row=x.sort_values("_rank",ascending=False).iloc[0]
    q="VERIFIED_TIME" if row["time_norm"] else "VERIFIED_DATE"
    return {
        "quality":q,"ticker":ticker,
        "company":str(row["name"]) if str(row["name"]) else None,
        "place":str(row["Place"]) if str(row["Place"]) else None,
        "date":row["date_norm"],"time":row["time_norm"],
        "description":str(row["description"]) if str(row["description"]) else None,
        "market_tz":str(row.get("market_tz","")) or None,
        "source_note":str(row.get("source_note","")) or None,
        "source_quality":str(row.get("source_quality","")) or None,
        "candidate_count":len(x),
        "source":"expanded first_trades_IPOs_clean + Bucholtz market anchors",
    }

def first_trade_datetime_utc(record,market_tz="America/New_York"):
    if not record or record.get("quality") not in ("VERIFIED_TIME","VERIFIED_DATE"):
        return None
    d=pd.Timestamp(record["date"])
    hh,mm=12,0
    if record.get("time"):
        hh,mm=map(int,record["time"].split(":"))
    tz_name=record.get("market_tz") or market_tz
    try:
        tz=ZoneInfo(tz_name)
    except Exception:
        tz=ZoneInfo(market_tz)
    return datetime(d.year,d.month,d.day,hh,mm,tzinfo=tz).astimezone(timezone.utc)

def natal_chart_from_record(record,market_tz="America/New_York",latitude=40.7128,longitude=-74.0060):
    dt=first_trade_datetime_utc(record,market_tz)
    if dt is None:
        return None
    snap=astro_snapshot_datetime(dt)
    chart={p:snap[p]["lon"] for p in V10_MOON_NATAL_TARGETS}
    chart["_datetime_utc"]=dt.isoformat()
    chart["_time_resolved"]=record.get("quality")=="VERIFIED_TIME"
    return chart

# Source corrections in the in-memory commodity universe.
_v10_source_dates={
    "WTI Crude Oil":"1983-03-30",
    "Natural Gas":"1990-04-03",
    "Corn":"1877-01-02","Wheat":"1877-01-02","Oats":"1877-01-02",
    "Soybeans":"1936-10-05",
    "Coffee":"1882-03-07",
    "Sugar":"1916-08-31",
    "Cocoa":"1925-10-02",
    "Cotton":"1870-06-20",
    "30Y Treasury Bond":"1977-08-22",
    "10Y Treasury Note":"1982-05-03",
    "British Pound":"1972-05-16","Canadian Dollar":"1972-05-16",
    "Euro FX":"2002-01-01","Australian Dollar":"1987-01-13",
}
for _n,_d in _v10_source_dates.items():
    if _n in COMMODITIES:
        COMMODITIES[_n]["first_trade"]=_d

FIRST_TRADE_DB=None
FIRST_TRADE_DB_PATH=None
# V10.2.3: database loading deferred until the selected tickers actually need it.

# ----- Full session lunar timing -----
def _v10_planet_lon(dt,planet):
    return float(astro_snapshot_datetime(dt)[planet]["lon"])

def _v10_aspect_error(lon1,lon2,angle):
    sep=abs((float(lon1)-float(lon2)+180.0)%360.0-180.0)
    return abs(sep-float(angle))

def _v10_golden_min(func,a,b,iters=42):
    gr=(math.sqrt(5)-1)/2
    c=b-gr*(b-a); d=a+gr*(b-a)
    fc=func(c); fd=func(d)
    for _ in range(iters):
        if fc<fd:
            b,d,fd=d,c,fc; c=b-gr*(b-a); fc=func(c)
        else:
            a,c,fc=c,d,fd; d=a+gr*(b-a); fd=func(d)
    x=(a+b)/2
    return x,func(x)

def _v10_fmt_session_time(dt,qualifier=None):
    txt=dt.astimezone(V10_NYSE_TZ).strftime("%H:%M:%S ET")
    return f"{txt} ({qualifier})" if qualifier else txt

def full_lunar_session_clock(date,record=None):
    """Exhaustive lunar aspect clock for the selected date, NYSE 09:30–16:00 ET only.

    An event is included only when the aspect becomes EXACT during that regular-session window.
    Apply/separate boundaries are clipped to the session when the true orb interval begins before
    09:30 or ends after 16:00.
    """
    d=pd.Timestamp(date).date()
    if pd.Timestamp(d).weekday()>=5:
        return {"date":str(d),"session":"NYSE regular session closed (weekend)",
                "transit":[],"natal":[],"record":record}
    op=datetime(d.year,d.month,d.day,9,30,tzinfo=V10_NYSE_TZ)
    cl=datetime(d.year,d.month,d.day,16,0,tzinfo=V10_NYSE_TZ)
    total=int((cl-op).total_seconds())
    minute_secs=np.arange(0,total+1,60,dtype=float)
    minute_dts=[op+timedelta(seconds=float(x)) for x in minute_secs]

    # One ephemeris pass per minute.
    moon=np.array([_v10_planet_lon(t,"Moon") for t in minute_dts])
    transit_lons={p:np.array([_v10_planet_lon(t,p) for t in minute_dts])
                  for p in V10_MOON_TRANSIT_TARGETS}
    chart=natal_chart_from_record(record) if record and record.get("quality") in ("VERIFIED_TIME","VERIFIED_DATE") else None

    families=[("MOON→TRANSIT",p,transit_lons[p]) for p in V10_MOON_TRANSIT_TARGETS]
    if chart:
        families += [("MOON→NATAL",p,np.full(len(minute_secs),float(chart[p])))
                     for p in V10_MOON_NATAL_TARGETS if p in chart]

    def dt_at(sec):
        return op+timedelta(seconds=float(sec))

    hits=[]
    for family,target,target_arr in families:
        for aspect,spec in V10_LUNAR_ASPECTS.items():
            angle=spec["angle"]; orb=spec["orb"]
            sep=np.abs((moon-target_arr+180.0)%360.0-180.0)
            errs=np.abs(sep-angle)
            idx=int(np.argmin(errs))
            # Moon moves ~0.5°/hour, so a genuine exact hit sampled each minute is much closer than 0.03°.
            if float(errs[idx])>0.03:
                continue

            lo=max(0.0,float(minute_secs[max(0,idx-3)]))
            hi=min(float(total),float(minute_secs[min(len(minute_secs)-1,idx+3)]))
            natal_lon=float(chart[target]) if family=="MOON→NATAL" else None

            def err(sec):
                t=dt_at(sec)
                ml=_v10_planet_lon(t,"Moon")
                tl=_v10_planet_lon(t,target) if family=="MOON→TRANSIT" else natal_lon
                return _v10_aspect_error(ml,tl,angle)

            exact_sec,exact_err=_v10_golden_min(err,lo,hi)
            if exact_err>0.003:
                continue

            def boundary(left=True):
                edge=0.0 if left else float(total)
                if err(edge)<=orb:
                    return edge,True
                step=60.0
                if left:
                    inside=exact_sec
                    outside=max(0.0,inside-step)
                    while outside>0 and err(outside)<=orb:
                        inside=outside; outside=max(0.0,outside-step)
                    a,b=outside,inside
                    for _ in range(36):
                        m=(a+b)/2
                        if err(m)>orb:a=m
                        else:b=m
                    return b,False
                inside=exact_sec
                outside=min(float(total),inside+step)
                while outside<float(total) and err(outside)<=orb:
                    inside=outside; outside=min(float(total),outside+step)
                a,b=inside,outside
                for _ in range(36):
                    m=(a+b)/2
                    if err(m)<=orb:a=m
                    else:b=m
                return a,False

            start_sec,start_clipped=boundary(True)
            end_sec,end_clipped=boundary(False)
            start_dt=dt_at(start_sec); exact_dt=dt_at(exact_sec); end_dt=dt_at(end_sec)
            hits.append({
                "family":family,"target":target,"aspect":aspect,
                "orb_limit":orb,"exact_orb":round(float(exact_err),5),
                "apply_start_et":_v10_fmt_session_time(start_dt,"already applying at open" if start_clipped else None),
                "exact_et":_v10_fmt_session_time(exact_dt),
                "separate_end_et":_v10_fmt_session_time(end_dt,"still separating/in orb at close" if end_clipped else None),
                "apply_start_iso":start_dt.isoformat(),"exact_iso":exact_dt.isoformat(),"separate_end_iso":end_dt.isoformat(),
                "start_clipped":start_clipped,"end_clipped":end_clipped,
                "record_quality":record.get("quality") if record else None,
            })

    hits.sort(key=lambda x:x["exact_iso"])
    return {
        "date":str(d),"session":"09:30–16:00 ET",
        "transit":[x for x in hits if x["family"]=="MOON→TRANSIT"],
        "natal":[x for x in hits if x["family"]=="MOON→NATAL"],
        "record":record,
    }

# ----- Attach full lunar clock to every stock / commodity / FX candidate -----
_V95_enrich_v9=enrich_v9
def enrich_v9(i):
    i=_V95_enrich_v9(i)
    rec=lookup_first_trade_record(i.get("ticker"))
    if rec.get("quality") not in ("VERIFIED_TIME","VERIFIED_DATE"):
        rec=None
    i["v10_first_trade_record"]=rec
    try:
        i["lunar_session_clock"]=full_lunar_session_clock(i.get("date"),rec)
    except Exception as e:
        i["lunar_session_clock"]={"date":i.get("date"),"session":"ERROR","transit":[],"natal":[],"error":str(e),"record":rec}
    return i

# ----- Bucholtz source-rule future calendar -----
BUCHOLTZ_TRIGGER_RULES = {'CL=F': {'market': 'WTI Crude Oil', 'passes': [('Sun', 31, 'natal Mars corner'), ('Sun', 89, 'natal Node corner'), ('Sun', 212, 'natal Saturn/Pluto/Moon corner'), ('Sun', 269, 'natal Neptune corner'), ('Mars', 31, 'natal Mars corner'), ('Mars', 89, 'natal Node corner'), ('Mars', 212, 'natal Saturn/Pluto/Moon corner'), ('Mars', 269, 'natal Neptune corner')], 'retro': ['Mercury']}, 'NG=F': {'market': 'Natural Gas', 'passes': [('Sun', 283, 'natal Mid-Heaven'), ('Sun', 26, 'natal Ascendant'), ('Sun', 116, 'natal Moon'), ('Sun', 294, 'natal Saturn'), ('Mars', 283, 'natal Mid-Heaven'), ('Mars', 26, 'natal Ascendant'), ('Mars', 116, 'natal Moon'), ('Mars', 294, 'natal Saturn')], 'decl': [('Venus', -19, 'natal Venus declination')]}, 'KC=F': {'market': 'Coffee', 'passes': [('Sun', 44, 'natal Neptune'), ('Sun', 166, 'natal Uranus'), ('Sun', 284, 'natal Mid-Heaven'), ('Sun', 346, 'natal Sun'), ('Sun', 27, 'natal Ascendant'), ('Mars', 44, 'natal Neptune'), ('Mars', 166, 'natal Uranus'), ('Mars', 284, 'natal Mid-Heaven'), ('Mars', 346, 'natal Sun'), ('Mars', 27, 'natal Ascendant')], 'decl': [('Mars', 20, 'natal Mars declination'), ('Venus', -7, 'natal Venus declination')]}, 'ZS=F': {'market': 'Soybeans', 'passes': [('Sun', 347, 'natal Saturn'), ('Sun', 76, 'natal Moon'), ('Sun', 104, 'natal Mid-Heaven'), ('Sun', 167, 'natal Neptune'), ('Sun', 192, 'natal Ascendant'), ('Sun', 258, 'natal Jupiter'), ('Mars', 347, 'natal Saturn'), ('Mars', 76, 'natal Moon'), ('Mars', 104, 'natal Mid-Heaven'), ('Mars', 167, 'natal Neptune'), ('Mars', 192, 'natal Ascendant'), ('Mars', 258, 'natal Jupiter')], 'retro': ['Mercury'], 'decl': [('Venus', -22, 'natal Venus declination'), ('Mars', 18, 'natal Mars declination')]}, 'ZC=F': {'market': 'Corn', 'passes': [('Sun', 219, 'natal Mid-Heaven'), ('Sun', 284, 'natal Ascendant'), ('Mars', 219, 'natal Mid-Heaven'), ('Mars', 284, 'natal Ascendant')], 'retro': ['Mercury'], 'decl': [('Venus', -8, 'natal Venus declination'), ('Mars', -8, 'natal Mars declination')]}, 'ZW=F': {'market': 'Wheat', 'passes': [('Sun', 219, 'natal Mid-Heaven'), ('Sun', 284, 'natal Ascendant'), ('Mars', 219, 'natal Mid-Heaven'), ('Mars', 284, 'natal Ascendant')], 'retro': ['Mercury'], 'decl': [('Venus', -8, 'natal Venus declination'), ('Mars', -8, 'natal Mars declination')]}, 'ZO=F': {'market': 'Oats', 'passes': [('Sun', 219, 'natal Mid-Heaven'), ('Sun', 284, 'natal Ascendant'), ('Mars', 219, 'natal Mid-Heaven'), ('Mars', 284, 'natal Ascendant')], 'retro': ['Mercury'], 'decl': [('Venus', -8, 'natal Venus declination'), ('Mars', -8, 'natal Mars declination')]}, '6B=F': {'market': 'British Pound', 'passes': [('Sun', 55, 'natal Sun'), ('Sun', 179, 'natal Pluto'), ('Sun', 299, 'natal Node'), ('Sun', 356, 'published Moon-watch MC point'), ('Sun', 106, 'published Moon-watch Asc point'), ('Mars', 55, 'natal Sun'), ('Mars', 179, 'natal Pluto'), ('Mars', 299, 'natal Node'), ('Mars', 356, 'published Moon-watch MC point'), ('Mars', 106, 'published Moon-watch Asc point')], 'retro': ['Mercury'], 'decl': [('Venus', -11, 'natal Venus declination'), ('Mars', 22, 'natal Mars declination')]}, '6C=F': {'market': 'Canadian Dollar', 'passes': [('Sun', 55, 'natal Sun'), ('Sun', 179, 'natal Pluto'), ('Sun', 299, 'natal Node'), ('Sun', 356, 'published Moon-watch MC point'), ('Sun', 106, 'published Moon-watch Asc point'), ('Mars', 55, 'natal Sun'), ('Mars', 179, 'natal Pluto'), ('Mars', 299, 'natal Node'), ('Mars', 356, 'published Moon-watch MC point'), ('Mars', 106, 'published Moon-watch Asc point')], 'retro': ['Mercury'], 'decl': [('Venus', -11, 'natal Venus declination'), ('Mars', 22, 'natal Mars declination')]}, '6E=F': {'market': 'Euro FX', 'passes': [('Sun', 231, 'natal Mid-Heaven'), ('Sun', 288, 'natal Ascendant'), ('Sun', 347, 'natal Mars'), ('Sun', 307, 'natal Neptune'), ('Sun', 69, 'natal Saturn'), ('Sun', 125, 'natal Moon'), ('Mars', 231, 'natal Mid-Heaven'), ('Mars', 288, 'natal Ascendant'), ('Mars', 347, 'natal Mars'), ('Mars', 307, 'natal Neptune'), ('Mars', 69, 'natal Saturn'), ('Mars', 125, 'natal Moon')], 'decl': [('Venus', -11, 'natal Venus declination'), ('Mars', 22, 'natal Mars declination')]}, '6A=F': {'market': 'Australian Dollar', 'passes': [('Sun', 276, 'natal Neptune'), ('Sun', 3, 'natal Mars'), ('Sun', 96, 'natal Moon'), ('Sun', 251, 'natal Mid-Heaven'), ('Sun', 325, 'natal Ascendant'), ('Mars', 276, 'natal Neptune'), ('Mars', 3, 'natal Mars'), ('Mars', 96, 'natal Moon'), ('Mars', 251, 'natal Mid-Heaven'), ('Mars', 325, 'natal Ascendant')], 'retro': ['Mercury'], 'decl': [('Venus', 12, 'natal Venus declination'), ('Mars', 15, 'natal Mars declination')]}, 'ZB=F': {'market': '30Y Treasury Bond', 'passes': [('Sun', 90, 'Sun 0° to natal Jupiter 0 Cancer'), ('Sun', 180, 'Sun 90° to natal Jupiter 0 Cancer'), ('Sun', 270, 'Sun 180° to natal Jupiter 0 Cancer'), ('Mars', 84, 'Mars 0° to natal Mars 24 Gemini'), ('Mars', 174, 'Mars 90° to natal Mars 24 Gemini'), ('Mars', 354, 'Mars 90° to natal Mars 24 Gemini')]}, 'ZN=F': {'market': '10Y Treasury Note', 'passes': [('Sun', 334, 'natal Mid-Heaven'), ('Sun', 88, 'natal Ascendant'), ('Sun', 214, 'natal Jupiter'), ('Sun', 62, 'natal Mercury'), ('Sun', 243, 'natal Uranus'), ('Sun', 358, 'natal Venus'), ('Sun', 180, 'natal Mars'), ('Sun', 266, 'natal Neptune'), ('Mars', 334, 'natal Mid-Heaven'), ('Mars', 88, 'natal Ascendant'), ('Mars', 214, 'natal Jupiter'), ('Mars', 62, 'natal Mercury'), ('Mars', 243, 'natal Uranus'), ('Mars', 358, 'natal Venus'), ('Mars', 180, 'natal Mars'), ('Mars', 266, 'natal Neptune')], 'retro': ['Mercury'], 'decl': [('Venus', -24, 'natal Venus declination'), ('Mars', -9, 'natal Mars declination')]}, 'CT=F': {'market': 'Cotton', 'passes': [('Sun', 88.83990543595125, 'Sun conjunction natal Sun'), ('Sun', 178.83990543595127, 'Sun square natal Sun'), ('Sun', 268.83990543595127, 'Sun opposition natal Sun'), ('Sun', 358.83990543595127, 'Sun square natal Sun'), ('Venus', 354, 'Venus conjunction natal Moon 24 Pisces')]}, 'SB=F': {'market': 'Sugar', 'passes': [('Sun', 204, 'Sun conjunction natal Mars 24 Libra'), ('Sun', 294, 'Sun square natal Mars 24 Libra'), ('Sun', 24, 'Sun opposition natal Mars 24 Libra'), ('Sun', 114, 'Sun square natal Mars 24 Libra')], 'retro': ['Mercury']}, 'CC=F': {'market': 'Cocoa', 'passes': [], 'retro': ['Mercury']}}

_V10_PLANET_IDS={
    "Sun":swe.SUN,"Moon":swe.MOON,"Mercury":swe.MERCURY,"Venus":swe.VENUS,"Mars":swe.MARS,
    "Jupiter":swe.JUPITER,"Saturn":swe.SATURN,"Uranus":swe.URANUS,"Neptune":swe.NEPTUNE,"Pluto":swe.PLUTO
}
def _v10_jd(dt):
    d=dt.astimezone(timezone.utc)
    return swe.julday(d.year,d.month,d.day,d.hour+d.minute/60+d.second/3600)
def _v10_state(dt,planet):
    xx,_=swe.calc_ut(_v10_jd(dt),_V10_PLANET_IDS[planet],swe.FLG_SWIEPH|swe.FLG_SPEED)
    eq,_=swe.calc_ut(_v10_jd(dt),_V10_PLANET_IDS[planet],swe.FLG_SWIEPH|swe.FLG_SPEED|swe.FLG_EQUATORIAL)
    return float(xx[0]%360),float(xx[3]),float(eq[1])
def _v10_signed(lon,target):
    return (float(lon)-float(target)+180)%360-180
def _v10_root(a,b,func,iters=42):
    fa=func(a)
    for _ in range(iters):
        m=a+(b-a)/2; fm=func(m)
        if fa*fm<=0:b=m
        else:a=m;fa=fm
    return a+(b-a)/2
def _v10_longitude_hits(planet,target,start_year,end_year):
    a=datetime(start_year,1,1,tzinfo=timezone.utc); end=datetime(end_year+1,1,1,tzinfo=timezone.utc)
    step=timedelta(hours=6); hits=[]; fa=_v10_signed(_v10_state(a,planet)[0],target)
    while a<end:
        b=min(a+step,end); fb=_v10_signed(_v10_state(b,planet)[0],target)
        if fa==0 or (fa*fb<0 and abs(fa)<10 and abs(fb)<10):
            r=_v10_root(a,b,lambda d:_v10_signed(_v10_state(d,planet)[0],target))
            if not hits or abs((r-hits[-1]).total_seconds())>12*3600:hits.append(r)
        a=b;fa=fb
    return hits
def _v10_declination_hits(planet,target,start_year,end_year):
    a=datetime(start_year,1,1,tzinfo=timezone.utc); end=datetime(end_year+1,1,1,tzinfo=timezone.utc)
    step=timedelta(hours=12);hits=[];fa=_v10_state(a,planet)[2]-target
    while a<end:
        b=min(a+step,end);fb=_v10_state(b,planet)[2]-target
        if fa==0 or fa*fb<0:
            r=_v10_root(a,b,lambda d:_v10_state(d,planet)[2]-target)
            if not hits or abs((r-hits[-1]).total_seconds())>2*86400:hits.append(r)
        a=b;fa=fb
    return hits
def _v10_retro_intervals(planet,start_year,end_year):
    a=datetime(start_year,1,1,tzinfo=timezone.utc);end=datetime(end_year+1,1,1,tzinfo=timezone.utc)
    step=timedelta(hours=12);prev=_v10_state(a,planet)[1];start=None;out=[];b=a+step
    while b<=end:
        sp=_v10_state(b,planet)[1]
        if prev>=0 and sp<0:start=_v10_root(a,b,lambda d:_v10_state(d,planet)[1])
        elif prev<0 and sp>=0 and start:
            stop=_v10_root(a,b,lambda d:_v10_state(d,planet)[1]);out.append((start,stop));start=None
        a=b;prev=sp;b+=step
    return out

def generate_bucholtz_reversal_calendar(start_year=2026,end_year=2030):
    """Model-derived dates produced by extending source-stated Bucholtz trigger rules.
    These are NOT dates published by Bucholtz for these future years.
    """
    rows=[];cache={};et=ZoneInfo("America/New_York")
    for ticker,rule in BUCHOLTZ_TRIGGER_RULES.items():
        for planet,target,note in rule.get("passes",[]):
            key=("LON",planet,float(target),start_year,end_year)
            if key not in cache:cache[key]=_v10_longitude_hits(planet,float(target),start_year,end_year)
            for dt in cache[key]:
                x=dt.astimezone(et)
                rows.append({"Date":x.date().isoformat(),"Exact_ET":x.strftime("%H:%M"),"Market":rule["market"],"Ticker":ticker,
                             "Trigger_Type":"ZODIAC_PASS","Planet":planet,"Target":f"{float(target):.2f}°","Rule":note,
                             "Source_Status":"MODEL-DERIVED FROM BUCHOLTZ SOURCE RULE",
                             "Interpretation":"Reversal/pivot watch; not directional by itself."})
        for planet,target,note in rule.get("decl",[]):
            key=("DEC",planet,float(target),start_year,end_year)
            if key not in cache:cache[key]=_v10_declination_hits(planet,float(target),start_year,end_year)
            for dt in cache[key]:
                x=dt.astimezone(et)
                rows.append({"Date":x.date().isoformat(),"Exact_ET":x.strftime("%H:%M"),"Market":rule["market"],"Ticker":ticker,
                             "Trigger_Type":"DECLINATION_REPEAT","Planet":planet,"Target":f"{float(target):+.1f}° declination","Rule":note,
                             "Source_Status":"MODEL-DERIVED FROM BUCHOLTZ SOURCE RULE",
                             "Interpretation":"Declination-repeat pivot watch; not directional by itself."})
        for planet in rule.get("retro",[]):
            key=("RET",planet,start_year,end_year)
            if key not in cache:cache[key]=_v10_retro_intervals(planet,start_year,end_year)
            for st,en in cache[key]:
                for typ,dt in [("RETROGRADE_START",st),("RETROGRADE_END",en)]:
                    x=dt.astimezone(et)
                    rows.append({"Date":x.date().isoformat(),"Exact_ET":x.strftime("%H:%M"),"Market":rule["market"],"Ticker":ticker,
                                 "Trigger_Type":typ,"Planet":planet,"Target":"station","Rule":"Mercury retrograde watch",
                                 "Source_Status":"MODEL-DERIVED FROM BUCHOLTZ SOURCE RULE",
                                 "Interpretation":"Retrograde boundary watch; source treats these periods as potential trend-change/volatility windows."})
    return pd.DataFrame(rows).sort_values(["Date","Exact_ET","Ticker","Trigger_Type"]).reset_index(drop=True)

V10_BUCHOLTZ_CALENDAR=None  # V10.2.3: inflate calendar only when first requested.

def v10_bucholtz_calendar_context(idea,n=5):
    ticker=str(idea.get("ticker",""))
    if ticker not in BUCHOLTZ_TRIGGER_RULES:
        return []
    d=str(pd.Timestamp(idea.get("date")).date())
    global V10_BUCHOLTZ_CALENDAR
    if V10_BUCHOLTZ_CALENDAR is None:
        V10_BUCHOLTZ_CALENDAR=_v10_embedded_calendar_df()
    cal=V10_BUCHOLTZ_CALENDAR
    if cal.empty or int(d[:4])<2026 or int(d[:4])>2030:
        cal=generate_bucholtz_reversal_calendar(int(d[:4]),int(d[:4])+2)
    x=cal[(cal["Ticker"]==ticker)&(cal["Date"]>=d)].head(n)
    return x.to_dict("records")

# Add V10 fields to summary without discarding V9.5 columns.
_V95_summary_dataframe=summary_dataframe
def summary_dataframe(ideas):
    df=_V95_summary_dataframe(ideas)
    trans=[];nat=[];next_trig=[]
    for i in ideas:
        if "lunar_session_clock" not in i:enrich_v9(i)
        clk=i.get("lunar_session_clock",{})
        trans.append(len(clk.get("transit",[])));nat.append(len(clk.get("natal",[])))
        ctx=v10_bucholtz_calendar_context(i,1)
        next_trig.append((ctx[0]["Date"]+" "+ctx[0]["Rule"]) if ctx else "")
    df["Moon→Transit session hits"]=trans
    df["Moon→Natal session hits"]=nat
    df["Next Bucholtz source-rule trigger"]=next_trig
    return df

print("V10 loaded: exhaustive NYSE-session lunar apply→exact→separate timing + Bucholtz future trigger calendar.")
```

## V10.1 — Optimised Lunar Engine

The full lunar calculations are now deferred until **after the user has selected/uploaded tickers or a Finviz CSV and the final result list exists**. The Moon/transiting-planet path is calculated once per selected date and cached; each selected ticker then only adds its natal comparisons.


```python

# ================================================================
# V10.1 — LAZY / CACHED LUNAR ENGINE
# ================================================================
V101_VERSION="10.1"

# Session cache keyed by selected date. Each date is calculated once.
V101_SESSION_CACHE = {}

def _v101_session_key(date):
    return str(pd.Timestamp(date).date())

def _v101_build_session_cache(date):
    """Calculate the NYSE-session Moon/planet path ONCE for this date."""
    key=_v101_session_key(date)
    if key in V101_SESSION_CACHE:
        return V101_SESSION_CACHE[key]

    d=pd.Timestamp(date).date()
    if pd.Timestamp(d).weekday()>=5:
        cache={
            "date":key,
            "session":"NYSE regular session closed (weekend)",
            "open":None,"close":None,
            "minute_secs":np.array([]),
            "minute_dts":[],
            "moon":np.array([]),
            "transit_lons":{},
            "transit_hits":[]
        }
        V101_SESSION_CACHE[key]=cache
        return cache

    op=datetime(d.year,d.month,d.day,9,30,tzinfo=V10_NYSE_TZ)
    cl=datetime(d.year,d.month,d.day,16,0,tzinfo=V10_NYSE_TZ)
    total=int((cl-op).total_seconds())
    minute_secs=np.arange(0,total+1,60,dtype=float)
    minute_dts=[op+timedelta(seconds=float(x)) for x in minute_secs]

    # Compute all planetary longitudes once.
    moon=np.empty(len(minute_dts),dtype=float)
    transit_lons={p:np.empty(len(minute_dts),dtype=float) for p in V10_MOON_TRANSIT_TARGETS}

    for idx,t in enumerate(minute_dts):
        snap=astro_snapshot_datetime(t)
        moon[idx]=float(snap["Moon"]["lon"])
        for p in V10_MOON_TRANSIT_TARGETS:
            transit_lons[p][idx]=float(snap[p]["lon"])

    cache={
        "date":key,
        "session":"09:30–16:00 ET",
        "open":op,"close":cl,
        "minute_secs":minute_secs,
        "minute_dts":minute_dts,
        "moon":moon,
        "transit_lons":transit_lons,
        "transit_hits":None
    }

    # Moon -> transit hits are identical for all tickers, so calculate once now.
    cache["transit_hits"]=_v101_extract_hits_from_cache(cache, natal_chart=None, family_mode="transit")
    V101_SESSION_CACHE[key]=cache
    print(f"V10.1 lunar session cache built once for {key}: {len(minute_dts)} one-minute samples.")
    return cache

def _v101_interp_lon(cache_arr, minute_secs, sec):
    """Linear interpolation on an unwrapped longitude series."""
    if len(cache_arr)==0:
        return np.nan
    arr=np.unwrap(np.deg2rad(cache_arr))
    val=np.interp(sec,minute_secs,arr)
    return float(np.rad2deg(val)%360.0)

def _v101_extract_hits_from_cache(cache,natal_chart=None,family_mode="both"):
    """Derive exact aspect events from the already-cached lunar/planetary path.

    No repeated Swiss-Ephemeris minute stepping occurs here.
    """
    if cache.get("open") is None:
        return []

    op=cache["open"]; cl=cache["close"]
    minute_secs=cache["minute_secs"]
    moon=cache["moon"]
    transit_lons=cache["transit_lons"]
    total=float((cl-op).total_seconds())

    families=[]
    if family_mode in ("both","transit"):
        families += [("MOON→TRANSIT",p,transit_lons[p]) for p in V10_MOON_TRANSIT_TARGETS]
    if family_mode in ("both","natal") and natal_chart:
        families += [("MOON→NATAL",p,np.full(len(minute_secs),float(natal_chart[p])))
                     for p in V10_MOON_NATAL_TARGETS if p in natal_chart]

    def dt_at(sec):
        return op+timedelta(seconds=float(sec))

    def lon_at_sec(arr,sec):
        return _v101_interp_lon(arr,minute_secs,sec)

    hits=[]
    for family,target,target_arr in families:
        sep=np.abs((moon-target_arr+180.0)%360.0-180.0)

        for aspect,spec in V10_LUNAR_ASPECTS.items():
            angle=float(spec["angle"]); orb=float(spec["orb"])
            errs=np.abs(sep-angle)

            # Find local minima close enough to an exact hit. More than one exact
            # hit per session is extremely unlikely for the Moon, but support it.
            candidates=[]
            for idx in range(1,len(errs)-1):
                if errs[idx] <= errs[idx-1] and errs[idx] <= errs[idx+1] and errs[idx] < 0.06:
                    candidates.append(idx)
            if errs[0] < 0.03:
                candidates.insert(0,0)
            if errs[-1] < 0.03:
                candidates.append(len(errs)-1)

            used=[]
            for idx in candidates:
                approx=float(minute_secs[idx])
                if any(abs(approx-u)<900 for u in used):
                    continue

                lo=max(0.0,approx-180.0)
                hi=min(total,approx+180.0)
                natal_lon=float(natal_chart[target]) if family=="MOON→NATAL" else None

                def err(sec):
                    ml=lon_at_sec(moon,sec)
                    if family=="MOON→TRANSIT":
                        tl=lon_at_sec(target_arr,sec)
                    else:
                        tl=natal_lon
                    return _v10_aspect_error(ml,tl,angle)

                exact_sec,exact_err=_v10_golden_min(err,lo,hi,36)
                if exact_err>0.01:
                    continue
                used.append(exact_sec)

                def boundary(left=True):
                    edge=0.0 if left else total
                    if err(edge)<=orb:
                        return edge,True

                    step=60.0
                    if left:
                        inside=exact_sec
                        outside=max(0.0,inside-step)
                        while outside>0 and err(outside)<=orb:
                            inside=outside
                            outside=max(0.0,outside-step)
                        a,b=outside,inside
                        for _ in range(32):
                            m=(a+b)/2
                            if err(m)>orb:a=m
                            else:b=m
                        return b,False
                    else:
                        inside=exact_sec
                        outside=min(total,inside+step)
                        while outside<total and err(outside)<=orb:
                            inside=outside
                            outside=min(total,outside+step)
                        a,b=inside,outside
                        for _ in range(32):
                            m=(a+b)/2
                            if err(m)<=orb:a=m
                            else:b=m
                        return a,False

                start_sec,start_clip=boundary(True)
                end_sec,end_clip=boundary(False)

                hits.append({
                    "family":family,
                    "target":target,
                    "aspect":aspect,
                    "orb_limit":orb,
                    "exact_orb":round(float(exact_err),5),
                    "apply_start_et":_v10_fmt_session_time(
                        dt_at(start_sec),
                        "already applying at open" if start_clip else None
                    ),
                    "exact_et":_v10_fmt_session_time(dt_at(exact_sec)),
                    "separate_end_et":_v10_fmt_session_time(
                        dt_at(end_sec),
                        "still separating/in orb at close" if end_clip else None
                    ),
                    "apply_start_iso":dt_at(start_sec).isoformat(),
                    "exact_iso":dt_at(exact_sec).isoformat(),
                    "separate_end_iso":dt_at(end_sec).isoformat(),
                    "start_clipped":start_clip,
                    "end_clipped":end_clip
                })

    hits.sort(key=lambda x:x["exact_iso"])
    return hits

def full_lunar_session_clock_cached(date,record=None):
    """Fast lunar clock using one shared session cache per selected date."""
    cache=_v101_build_session_cache(date)
    natal_chart=None
    if record and record.get("quality") in ("VERIFIED_TIME","VERIFIED_DATE"):
        natal_chart=natal_chart_from_record(record)

    transit=list(cache.get("transit_hits") or [])
    natal=_v101_extract_hits_from_cache(cache,natal_chart=natal_chart,family_mode="natal") if natal_chart else []

    return {
        "date":cache["date"],
        "session":cache["session"],
        "transit":transit,
        "natal":natal,
        "record":record
    }

def _v101_selected_tickers_from_results():
    """Return only the tickers actually selected by the user's scan / Finviz input."""
    ideas=list(LAST_SCAN_RESULTS.get("commodities",[]))+list(LAST_SCAN_RESULTS.get("stocks",[]))
    seen=set();out=[]
    for i in ideas:
        t=str(i.get("ticker","")).strip()
        if t and t not in seen:
            seen.add(t);out.append(t)
    return out

def compute_lunar_clocks_for_selected_results(force=False):
    """Run the heavy lunar layer ONLY after ticker selection is complete."""
    ideas=list(LAST_SCAN_RESULTS.get("commodities",[]))+list(LAST_SCAN_RESULTS.get("stocks",[]))
    if not ideas:
        print("No selected scan results yet. Upload/select tickers or a Finviz CSV and run the scan first.")
        return []

    date=LAST_SCAN_RESULTS.get("date")
    if not date:
        print("No selected scan date is available.")
        return []

    cache=_v101_build_session_cache(date)
    completed=[]

    for i in ideas:
        if (not force) and i.get("lunar_session_clock_v101"):
            completed.append(i)
            continue

        ticker=i.get("ticker")
        rec=lookup_first_trade_record(ticker)
        if rec.get("quality") not in ("VERIFIED_TIME","VERIFIED_DATE"):
            rec=None

        i["v10_first_trade_record"]=rec
        i["lunar_session_clock_v101"]=full_lunar_session_clock_cached(date,rec)
        # Keep the original report key pointed at the optimized result.
        i["lunar_session_clock"]=i["lunar_session_clock_v101"]
        completed.append(i)

    print(f"V10.1 lunar clocks calculated for {len(completed)} selected candidates only.")
    print(f"Shared Moon/transiting-planet session path was calculated once for {date}.")
    return completed

# IMPORTANT: do not run the V10 lunar clock inside every broad scan candidate.
_V101_BASE_ENRICH=enrich_v9
def enrich_v9(i):
    # FAST PATH: bypass the V10 eager lunar wrapper completely.
    # _V95_enrich_v9 is the V9.5 enrichment function captured before V10 wrapped it.
    i=_V95_enrich_v9(i)
    i.pop("lunar_session_clock",None)
    i.pop("lunar_session_clock_v101",None)
    return i

# Wrap reports so the lunar clock is calculated only when the final selected
# result set is known and the user asks for output.
_V101_OLD_MAKE_PDF=make_pdf_report
_V101_OLD_MAKE_CSV=make_csv_report

def make_pdf_report():
    compute_lunar_clocks_for_selected_results()
    return _V101_OLD_MAKE_PDF()

def make_csv_report():
    compute_lunar_clocks_for_selected_results()
    return _V101_OLD_MAKE_CSV()

print("V10.1 optimization loaded: lunar timing is lazy, cached and runs only after selected tickers/Finviz results exist.")
```

## V10.2 — Overnight / Premarket Timing Continuum

The timing engine now examines **both charts**: Moon→first-trade/natal planets and Moon→current transiting planets (including the Sun). It also includes non-lunar **Transit→Transit** aspects as lower-priority background context. The lifecycle is calculated from before the previous close through the selected date so the report can identify prior-close anticipation, overnight orb entry, premarket application, exact perfection and separation.


```python

# ================================================================
# V10.2 — PREVIOUS CLOSE → OVERNIGHT → PREMARKET → SESSION TIMING
# Includes Moon→Natal, Moon→Transit and Transit→Transit
# ================================================================
V102_VERSION="10.2.2"
V102_SKY_CACHE={}

V102_ASPECTS={
    "conjunction":{"angle":0.0,"orb":5.0},
    "opposition":{"angle":180.0,"orb":5.0},
    "trine":{"angle":120.0,"orb":3.0},
    "sextile":{"angle":60.0,"orb":3.0},
    "square":{"angle":90.0,"orb":3.0},
    "quincunx":{"angle":150.0,"orb":1.5},
    "semisquare":{"angle":45.0,"orb":1.5},
    "sesquisquare":{"angle":135.0,"orb":1.5},
}
V102_PLANETS=["Sun","Moon","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]
V102_NATAL_TARGETS=["Sun","Moon","Mercury","Venus","Mars","Jupiter","Saturn","Uranus","Neptune","Pluto"]

def _v102_previous_trading_day(date):
    d=pd.Timestamp(date).date()-timedelta(days=1)
    while pd.Timestamp(d).weekday()>=5:
        d-=timedelta(days=1)
    return d

def _v102_window(date):
    """48h lookback is ample for the requested lunar orbs; slower T→T events may be clipped."""
    d=pd.Timestamp(date).date()
    prev=_v102_previous_trading_day(d)
    # Start 24h before the previous close, so Moon orb-entry is normally found exactly.
    start=datetime(prev.year,prev.month,prev.day,16,0,tzinfo=V10_NYSE_TZ)-timedelta(hours=24)
    end=datetime(d.year,d.month,d.day,23,59,59,tzinfo=V10_NYSE_TZ)
    return start,end

def _v102_period(dt,date):
    d=pd.Timestamp(date).date()
    prev=_v102_previous_trading_day(d)
    x=dt.astimezone(V10_NYSE_TZ)
    mins=x.hour*60+x.minute
    if x.date()<prev:
        return "EARLIER"
    if x.date()==prev:
        if mins<16*60:
            return "PREVIOUS SESSION"
        return "OVERNIGHT / AFTER PREVIOUS CLOSE"
    if x.date()<d:
        return "OVERNIGHT"
    if x.date()>d:
        return "NEXT DAY"
    if mins<4*60:
        return "OVERNIGHT"
    if mins<9*60+30:
        return "PREMARKET"
    if mins<=16*60:
        return "NYSE SESSION"
    if mins<=20*60:
        return "AFTER-HOURS"
    return "AFTER-HOURS / NIGHT"

def _v102_build_cache(date):
    """Shared sky path, calculated once after final ticker selection."""
    key=str(pd.Timestamp(date).date())
    if key in V102_SKY_CACHE:
        return V102_SKY_CACHE[key]

    start,end=_v102_window(date)
    # Ten-minute discovery grid halves ephemeris calls. The Moon moves only ~0.09° in 10 minutes;
    # exact/apply/separate times are then refined by interpolation/bisection.
    step=600.0
    total=(end-start).total_seconds()
    secs=np.arange(0,total+1,step,dtype=float)
    dts=[start+timedelta(seconds=float(s)) for s in secs]
    lons={p:np.empty(len(dts),dtype=float) for p in V102_PLANETS}

    for idx,t in enumerate(dts):
        # Direct Swiss-Ephemeris ecliptic longitude only: faster than full snapshot.
        jd=_jd(t)
        for p in V102_PLANETS:
            xx,_=swe.calc_ut(jd,PLANETS[p],swe.FLG_SWIEPH|swe.FLG_SPEED)
            lons[p][idx]=float(xx[0]%360.0)

    cache={"date":key,"start":start,"end":end,"secs":secs,"dts":dts,"lons":lons}
    cache["moon_transit"]=_v102_shared_events(cache,"MOON→TRANSIT")
    cache["transit_transit"]=_v102_shared_events(cache,"TRANSIT→TRANSIT")
    V102_SKY_CACHE[key]=cache
    print(f"V10.2.2 sky cache built once for {key}: {len(dts):,} ten-minute samples.")
    return cache

def _v102_interp(arr,secs,sec):
    uw=np.unwrap(np.deg2rad(arr))
    return float(np.rad2deg(np.interp(float(sec),secs,uw))%360.0)

def _v102_err(l1,l2,angle):
    sep=abs((float(l1)-float(l2)+180.0)%360.0-180.0)
    return abs(sep-float(angle))

def _v102_minimise(func,a,b,iters=40):
    gr=(math.sqrt(5)-1)/2
    c=b-gr*(b-a); d=a+gr*(b-a); fc=func(c); fd=func(d)
    for _ in range(iters):
        if fc<fd:
            b,d,fd=d,c,fc; c=b-gr*(b-a); fc=func(c)
        else:
            a,c,fc=c,d,fd; d=a+gr*(b-a); fd=func(d)
    x=(a+b)/2
    return x,func(x)

def _v102_lifecycle(cache,arr1,arr2,aspect,family,target):
    spec=V102_ASPECTS[aspect]
    angle=float(spec["angle"]); orb=float(spec["orb"])
    secs=cache["secs"]; start=cache["start"]; total=float(secs[-1])
    sep=np.abs((arr1-arr2+180.0)%360.0-180.0)
    errs=np.abs(sep-angle)

    # Detect local minima near perfection.
    candidates=[]
    for idx in range(1,len(errs)-1):
        if errs[idx]<=errs[idx-1] and errs[idx]<=errs[idx+1] and errs[idx]<0.10:
            candidates.append(idx)

    out=[]; used=[]
    for idx in candidates:
        approx=float(secs[idx])
        if any(abs(approx-u)<3600 for u in used):
            continue
        lo=max(0.0,approx-600.0); hi=min(total,approx+600.0)
        def err(sec):
            return _v102_err(_v102_interp(arr1,secs,sec),_v102_interp(arr2,secs,sec),angle)
        exact_sec,exact_err=_v102_minimise(err,lo,hi)
        if exact_err>0.02:
            continue
        exact_dt=start+timedelta(seconds=float(exact_sec))

        # Keep only aspects which become exact on the SELECTED DATE.
        if exact_dt.astimezone(V10_NYSE_TZ).date()!=pd.Timestamp(cache["date"]).date():
            continue
        used.append(exact_sec)

        def boundary(left=True):
            # Find exact requested-orb crossing. If a slow T→T aspect began before lookback,
            # flag it as clipped rather than inventing a start.
            edge=0.0 if left else total
            if err(edge)<=orb:
                return edge,True
            step=300.0
            if left:
                inside=exact_sec; outside=max(0.0,inside-step)
                while outside>0 and err(outside)<=orb:
                    inside=outside; outside=max(0.0,outside-step)
                a,b=outside,inside
                for _ in range(40):
                    m=(a+b)/2
                    if err(m)>orb:a=m
                    else:b=m
                return b,False
            inside=exact_sec; outside=min(total,inside+step)
            while outside<total and err(outside)<=orb:
                inside=outside; outside=min(total,outside+step)
            a,b=inside,outside
            for _ in range(40):
                m=(a+b)/2
                if err(m)<=orb:a=m
                else:b=m
            return a,False

        apply_sec,apply_clip=boundary(True)
        sep_sec,sep_clip=boundary(False)
        apply_dt=start+timedelta(seconds=float(apply_sec))
        sep_dt=start+timedelta(seconds=float(sep_sec))

        out.append({
            "family":family,"target":target,"aspect":aspect,"orb_limit":orb,
            "apply_start_et":apply_dt.astimezone(V10_NYSE_TZ).strftime("%Y-%m-%d %H:%M:%S ET"),
            "exact_et":exact_dt.astimezone(V10_NYSE_TZ).strftime("%Y-%m-%d %H:%M:%S ET"),
            "separate_end_et":sep_dt.astimezone(V10_NYSE_TZ).strftime("%Y-%m-%d %H:%M:%S ET"),
            "apply_period":_v102_period(apply_dt,cache["date"]),
            "exact_period":_v102_period(exact_dt,cache["date"]),
            "separate_period":_v102_period(sep_dt,cache["date"]),
            "apply_start_iso":apply_dt.isoformat(),"exact_iso":exact_dt.isoformat(),"separate_end_iso":sep_dt.isoformat(),
            "apply_clipped":apply_clip,"separate_clipped":sep_clip,
        })
    return out

def _v102_shared_events(cache,family):
    events=[]; lons=cache["lons"]
    if family=="MOON→TRANSIT":
        for p in [x for x in V102_PLANETS if x!="Moon"]:
            for asp in V102_ASPECTS:
                events += _v102_lifecycle(cache,lons["Moon"],lons[p],asp,family,f"transiting {p}")
    elif family=="TRANSIT→TRANSIT":
        nonmoon=[p for p in V102_PLANETS if p!="Moon"]
        for ai,a in enumerate(nonmoon):
            for b in nonmoon[ai+1:]:
                for asp in V102_ASPECTS:
                    events += _v102_lifecycle(cache,lons[a],lons[b],asp,family,f"{a} ↔ {b}")
    return sorted(events,key=lambda x:x["exact_iso"])

def _v102_natal_events(cache,record):
    if not record or record.get("quality") not in ("VERIFIED_TIME","VERIFIED_DATE"):
        return []
    chart=natal_chart_from_record(record)
    if not chart:
        return []
    moon=cache["lons"]["Moon"]; secs=cache["secs"]; events=[]
    for p in V102_NATAL_TARGETS:
        if p not in chart:
            continue
        target=np.full(len(secs),float(chart[p]),dtype=float)
        for asp in V102_ASPECTS:
            events += _v102_lifecycle(cache,moon,target,asp,"MOON→NATAL",f"natal {p}")
    return sorted(events,key=lambda x:x["exact_iso"])

def _v102_primary_timing_plan(idea,continuum):
    """Choose the stock-specific lunar event that should drive timing presentation.

    Moon→Natal outranks Moon→Transit. Transit→Transit is supporting context only.
    Preference is given to events whose orb begins after the previous close and before
    the NYSE open, because those can be anticipated at the prior close.
    """
    date=pd.Timestamp(idea.get("date")).date()
    prev=_v102_previous_trading_day(date)
    prev_close=datetime(prev.year,prev.month,prev.day,16,0,tzinfo=V10_NYSE_TZ)
    open_dt=datetime(date.year,date.month,date.day,9,30,tzinfo=V10_NYSE_TZ)

    lunar=list(continuum.get("moon_natal",[]))+list(continuum.get("moon_transit",[]))
    if not lunar:
        return {
            "classification":"NO QUALIFYING LUNAR PERFECTION",
            "preferred_window":"No qualifying Moon exact hit on selected date",
            "reason":"No Moon→natal or Moon→transit aspect in the requested aspect/orb set becomes exact on the selected date."
        }

    aspect_weight={"conjunction":1.0,"opposition":1.0,"trine":0.9,"square":0.9,"sextile":0.8,
                   "quincunx":0.65,"semisquare":0.6,"sesquisquare":0.6}

    ranked=[]
    for e in lunar:
        apply_dt=pd.Timestamp(e["apply_start_iso"]).to_pydatetime()
        exact_dt=pd.Timestamp(e["exact_iso"]).to_pydatetime()
        fam=3.0 if e["family"]=="MOON→NATAL" else 2.0
        overnight_bonus=2.5 if prev_close < apply_dt < open_dt else 0.0
        premarket_exact=1.0 if e["exact_period"]=="PREMARKET" else 0.0
        session_exact=0.8 if e["exact_period"]=="NYSE SESSION" else 0.0
        score=fam+overnight_bonus+premarket_exact+session_exact+aspect_weight.get(e["aspect"],0.5)
        ranked.append((score,e))
    ranked.sort(key=lambda x:(-x[0],x[1]["exact_iso"]))
    e=ranked[0][1]
    apply_dt=pd.Timestamp(e["apply_start_iso"]).to_pydatetime()
    exact_dt=pd.Timestamp(e["exact_iso"]).to_pydatetime()

    # Build a practical timing classification.
    if prev_close < apply_dt < open_dt:
        cls="PRIOR-CLOSE ANTICIPATION / OVERNIGHT GAP WATCH"
        window=f"{prev.strftime('%Y-%m-%d')} 15:30–16:00 ET prior close; next milestone orb-entry {apply_dt.astimezone(V10_NYSE_TZ).strftime('%H:%M:%S ET')}"
        reason=(f"{e['family']} {e['aspect']} {e['target']} enters its requested {e['orb_limit']:.1f}° orb after the previous close "
                f"and becomes exact {e['exact_et']}. This condition is knowable before the prior close.")
    elif apply_dt <= prev_close:
        cls="ALREADY APPLYING AT PRIOR CLOSE"
        window=f"Aspect already active by prior close; exact {e['exact_et']}"
        reason=(f"{e['family']} {e['aspect']} {e['target']} had already entered orb before the prior close. "
                "It is not a fresh overnight orb-entry signal, but remains active context.")
    elif e["apply_period"]=="PREMARKET":
        cls="PREMARKET APPLYING"
        window=f"Before orb-entry {e['apply_start_et']} if using premarket access; exact {e['exact_et']}"
        reason=f"{e['family']} {e['aspect']} {e['target']} begins applying premarket and perfects on the selected date."
    elif e["apply_period"]=="NYSE SESSION":
        cls="INTRADAY APPLYING"
        window=f"Before orb-entry {e['apply_start_et']}; exact {e['exact_et']}"
        reason=f"{e['family']} {e['aspect']} {e['target']} begins applying during the regular session."
    else:
        cls="SELECTED-DATE LUNAR CONTEXT"
        window=f"Orb-entry {e['apply_start_et']} → exact {e['exact_et']}"
        reason=f"{e['family']} {e['aspect']} {e['target']} is the highest-priority selected-date lunar event."

    return {
        "classification":cls,
        "preferred_window":window,
        "reason":reason,
        "primary_event":e
    }

def compute_timing_continuum_for_selected_results(force=False):
    """Heavy timing work runs only after ticker/Finviz selection and final scan results exist."""
    ideas=list(LAST_SCAN_RESULTS.get("commodities",[]))+list(LAST_SCAN_RESULTS.get("stocks",[]))
    if not ideas:
        print("No selected results yet. Select tickers or upload Finviz CSV, run the scan, then calculate timing.")
        return []

    date=LAST_SCAN_RESULTS.get("date")
    cache=_v102_build_cache(date)
    for i in ideas:
        if i.get("v102_timing_continuum") and not force:
            continue
        rec=lookup_first_trade_record(i.get("ticker"))
        if rec.get("quality") not in ("VERIFIED_TIME","VERIFIED_DATE"):
            rec=None
        continuum={
            "moon_natal":_v102_natal_events(cache,rec),
            "moon_transit":list(cache.get("moon_transit",[])),
            "transit_transit":list(cache.get("transit_transit",[])),
            "record":rec,
            "date":str(pd.Timestamp(date).date())
        }
        i["v102_timing_continuum"]=continuum
        i["v102_timing_plan"]=_v102_primary_timing_plan(i,continuum)

        # Keep backward-compatible lunar clock for other report code.
        i["lunar_session_clock"]={
            "date":continuum["date"],"session":"V10.2 continuum",
            "transit":continuum["moon_transit"],"natal":continuum["moon_natal"],"record":rec
        }

        # Preserve legacy V8.4 timing but prevent after-close perfection from being presented
        # as the preferred V10.2 window.
        d=i.get("decision_v84") or i.get("decision_v83")
        if isinstance(d,dict):
            d.setdefault("legacy_execution_window",d.get("execution_window"))
            d.setdefault("legacy_best_time_et",d.get("best_time_et"))
            d["execution_window"]=i["v102_timing_plan"].get("preferred_window")
    print(f"V10.2 timing continuum calculated for {len(ideas)} selected candidates only.")
    print("Shared Moon→Transit and Transit→Transit sky paths were calculated once; only Moon→Natal varies by ticker.")
    return ideas

def _v102_rows(events):
    rows=[['Aspect','Orb','Apply starts','Exact','Separate ends']]
    for e in events:
        rows.append([
            f"{e['family']} {e['aspect']} {e['target']}",
            f"{e['orb_limit']:.1f}°",
            f"{e['apply_start_et']} [{e['apply_period']}]",
            f"{e['exact_et']} [{e['exact_period']}]",
            f"{e['separate_end_et']} [{e['separate_period']}]"
        ])
    return rows

def v102_add_report_sections(story,styles,i):
    """Render all three timing families in the PDF."""
    cont=i.get("v102_timing_continuum",{})
    plan=i.get("v102_timing_plan",{})

    story.append(Paragraph('B2. V10.2 Timing Continuum — Previous Close → Overnight → Premarket → Session',styles['V9Sec']))
    story.append(Paragraph(
        '<b>Priority rule:</b> Moon→Natal is stock-specific and highest priority; Moon→Transit examines the live transit chart; '
        'Transit→Transit is supporting background context and does not overrule stock-specific lunar/natal evidence. '
        'All events shown become exact on the selected date. Applying starts and separation ends may fall before/after the NYSE session.',
        styles['V9Small']))
    if plan:
        story.append(Paragraph(
            f"<b>Timing classification:</b> {_html.escape(str(plan.get('classification','')))}<br/>"
            f"<b>Preferred timing window:</b> {_html.escape(str(plan.get('preferred_window','')))}<br/>"
            f"<b>Reason:</b> {_html.escape(str(plan.get('reason','')))}",
            styles['V9Small']))

    def add_table(title,events,limit=20):
        story.append(Paragraph(title,styles['V9Small']))
        if not events:
            story.append(Paragraph('• No qualifying exact aspect on the selected date.',styles['V9Small']))
            return
        data=_v102_rows(events[:limit])
        tb=Table(data,repeatRows=1,colWidths=[48*mm,12*mm,42*mm,37*mm,42*mm])
        tb.setStyle(TableStyle([
            ('BACKGROUND',(0,0),(-1,0),colors.lightgrey),('GRID',(0,0),(-1,-1),0.25,colors.grey),
            ('FONTSIZE',(0,0),(-1,-1),5.4),('VALIGN',(0,0),(-1,-1),'TOP'),
            ('LEFTPADDING',(0,0),(-1,-1),2),('RIGHTPADDING',(0,0),(-1,-1),2)
        ]))
        story.append(tb);story.append(Spacer(1,3))

    add_table('<b>Moon → Natal / first-trade chart</b>',cont.get("moon_natal",[]))
    add_table('<b>Moon → Transiting planetary chart (including Sun)</b>',cont.get("moon_transit",[]))
    add_table('<b>Transit → Transit background aspects</b>',cont.get("transit_transit",[]),12)

# Report wrappers: calculate only after the selected result list exists.
_V102_OLD_MAKE_PDF=make_pdf_report
_V102_OLD_MAKE_CSV=make_csv_report

def make_pdf_report():
    compute_timing_continuum_for_selected_results()
    return _V102_OLD_MAKE_PDF()

def make_csv_report():
    compute_timing_continuum_for_selected_results()
    return _V102_OLD_MAKE_CSV()

print("V10.2 loaded: Moon→Natal + Moon→Transit + Transit→Transit, with prior-close/overnight/premarket/session timing.")
```

## V10.2.1 — Startup / Database Order Fix

The old V9 sanity checks no longer try to load the first-trade database before the V10 loader exists. Run All now completes first; first-trade validation happens only after you select/upload tickers and run the scan.


```python

# V10.2.1 on-demand first-trade validation
def validate_selected_first_trade_records():
    ideas=list(LAST_SCAN_RESULTS.get("commodities",[]))+list(LAST_SCAN_RESULTS.get("stocks",[]))
    if not ideas:
        print("No selected scan results yet. Select/upload tickers and run the scan first.")
        return pd.DataFrame()
    rows=[]
    for i in ideas:
        t=i.get("ticker")
        r=lookup_first_trade_record(t)
        rows.append({
            "Ticker":t,
            "Quality":r.get("quality"),
            "Date":r.get("date"),
            "Time":r.get("time"),
            "Source":r.get("source"),
        })
    out_df=pd.DataFrame(rows)
    display(out_df)
    bad=out_df[~out_df["Quality"].isin(["VERIFIED_TIME","VERIFIED_DATE"])]
    if len(bad):
        print(f"WARNING: {len(bad)} selected ticker(s) do not have a verified first-trade record.")
    else:
        print("Selected first-trade validation passed.")
    return out_df
```

## V10.2.2 — Performance Fix

This build removes a hidden duplicate calculation: V10.1 was still running the old full lunar engine inside each candidate before discarding the result. V10.2.2 bypasses that eager wrapper, calculates the shared sky once after the final ticker/Finviz selection, and prevents PDF/CSV export from running a second lunar clock.


```python

# ================================================================
# V10.2.2 — FINAL FAST WRAPPERS
# ================================================================
# The old wrapper chain was:
# V10.2 timing -> V10.1 lunar clock -> V10 report
# which duplicated the expensive lunar calculations.
#
# This final override does:
# V10.2 timing continuum ONCE -> original V10 report renderer.
# The report already reads v102_timing_continuum, so no second clock is needed.

def make_pdf_report():
    compute_timing_continuum_for_selected_results()
    return _V101_OLD_MAKE_PDF()

def make_csv_report():
    compute_timing_continuum_for_selected_results()
    return _V101_OLD_MAKE_CSV()

print("V10.2.2 fast wrappers active: duplicate V10.1 lunar calculation bypassed.")
```

```python

def v1022_timing_benchmark():
    """Optional diagnostic: reports how long the selected timing calculation takes."""
    import time
    t0=time.perf_counter()
    compute_timing_continuum_for_selected_results(force=True)
    elapsed=time.perf_counter()-t0
    n=len(LAST_SCAN_RESULTS.get("commodities",[]))+len(LAST_SCAN_RESULTS.get("stocks",[]))
    print(f"Timing continuum: {elapsed:.1f}s for {n} selected candidate(s).")
    return elapsed
```
